````python
# %% Cell 1 (markdown)
# # # Developing a Multi-Agent AI Agent System
# # ## Week 6 Final Project — AI Business Analyst (LangGraph + Groq)
# #
# # ---
# # # 1. Project Overview
# #
# # This notebook implements an **AI Business Analyst Multi-Agent System**. It accepts a natural-language business request and autonomously runs a multi-step workflow:
# #
# # 1. **Retrieves** data from three sources: the *Online Retail* Excel dataset, the *Target 2025 Annual Report* (PDF, lightweight RAG) and *Open-Meteo* weather data (live API with local CSV fallback).
# # 2. **Validates and cleans** the sales data with a documented data-quality pipeline.
# # 3. **Analyses** it with deterministic Pandas tools (KPIs, trends, top products/countries, customers, statistical anomalies).
# # 4. **Interprets** the results with a Groq LLM, separating *FACT / CALCULATED RESULT / LLM INTERPRETATION / RECOMMENDATION*.
# # 5. **Generates** a professional PDF report (ReportLab + Matplotlib charts).
# # 6. **Acts**: saves the report and JSON artifacts and prepares an e-mail-ready summary.
# #
# # **Design principle:** *Python computes, the LLM interprets.* Every number in the report comes from Pandas. The LLM only (a) decides which tools to call, (b) summarises retrieved annual-report evidence, (c) writes interpretations/recommendations. If the LLM is unavailable, the workflow still runs and produces a complete deterministic report.
# #
# # # 2. Objectives
# #
# # | # | Objective | Where |
# # |---|-----------|-------|
# # | 1 | Multiple specialised agents | Section 16 |
# # | 2 | Real LLM tool calling (Groq) | Sections 13, 14, 16 |
# # | 3 | LangGraph orchestration with conditional edges | Section 17 |
# # | 4 | External API integration (Open-Meteo) + fallback | Section 12 |
# # | 5 | PDF retrieval (TF-IDF RAG) with page citations | Section 11 |
# # | 6 | Deterministic analytics and anomaly detection | Sections 8-10, 14 |
# # | 7 | Automated PDF report + e-mail summary | Sections 21-22 |
# # | 8 | Error handling, logging, testing, evaluation | Sections 23-24 |
# #
# # # 3. Architecture
# #
# # The diagram is drawn in Section 17 (after the graph is compiled) and also exported as Mermaid text.
# #
# # ```
# # User request -> Orchestrator (LangGraph)
# #   -> Request check -> Data Retrieval Agent -> Data Quality check --(not acceptable)--> Cleaning -> (re-check)
# #   -> Data Analysis Agent -> Research/Insight Agent -> Report Agent -> Action Agent -> PDF + e-mail summary
# # ```
# #
# # Large DataFrames are **not** placed in the LangGraph state (they are not JSON-serialisable and would bloat it). They live in a `DataStore`; the state carries only small, serialisable summaries.

# %% Cell 2 (markdown)
# # # 4. Environment Setup
# # The next cell installs any missing packages (needs **Internet = On** in Kaggle: *Settings → Internet*). If an import error appears right after installation, restart the session and *Run All*.

# %% Cell 3 (code)
import importlib, importlib.util, subprocess, sys

REQUIRED_PACKAGES = {
    "pandas": "pandas", "numpy": "numpy", "openpyxl": "openpyxl", "matplotlib": "matplotlib",
    "sklearn": "scikit-learn", "pypdf": "pypdf", "reportlab": "reportlab", "requests": "requests",
    "pydantic": "pydantic", "langgraph": "langgraph", "langchain_core": "langchain-core",
    "langchain_groq": "langchain-groq",
}
missing = [pip_name for mod, pip_name in REQUIRED_PACKAGES.items() if importlib.util.find_spec(mod) is None]
if missing:
    print("Installing missing packages:", missing)
    proc = subprocess.run([sys.executable, "-m", "pip", "install", "-q", *missing], capture_output=True, text=True)
    importlib.invalidate_caches()
    if proc.returncode != 0:
        print(proc.stderr[-1500:])
        print("\nInstallation failed. In Kaggle enable: Settings -> Internet -> On, then re-run.")
    else:
        print("Installation finished.")
else:
    print("All required packages are already installed.")


# %% Cell 4 (code)
import os, re, io, sys, json, time, math, random, functools, traceback, operator, warnings
import datetime as dt
from dataclasses import dataclass, field
from pathlib import Path
from typing import Any, Annotated, Callable, Literal, Optional, TypedDict

import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")          # charts are saved to PNG and displayed explicitly
import matplotlib.pyplot as plt
from IPython.display import display, Image, Markdown

warnings.filterwarnings("ignore", category=FutureWarning)
pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 200)
print("Python", sys.version.split()[0], "| pandas", pd.__version__, "| numpy", np.__version__)


# %% Cell 5 (markdown)
# # # 5. Configuration
# #
# # **This is the only place you need to edit** (plus the request cell below).
# #
# # ```
# # GROQ API KEY LOCATION:
# # 1. Kaggle Secrets  ->  GROQ_API_KEY      (recommended: Add-ons -> Secrets)
# # 2. Or the configuration cell near the beginning of this notebook  (variable GROQ_API_KEY)
# # ```
# # The key is never printed, never logged and never written to any output file.

# %% Cell 6 (code)
# ============================== CONFIGURATION ==============================
# 1) DATA LOCATION ---------------------------------------------------------
#    Attach your Kaggle Dataset to the notebook; its files appear under /kaggle/input/<dataset-slug>/.
#    The notebook also auto-searches all of /kaggle/input/, so this only needs to be right if you want to be explicit.
DATA_DIR = "/kaggle/input/odc-project-data"

# 2) GROQ API KEY ------------------------------------------------------------
#    Preferred: Kaggle Secrets named GROQ_API_KEY.  Fallback: paste the key between the quotes below.
#    (Do NOT share/commit the notebook with a key pasted here.)
GROQ_API_KEY = "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"

# 3) MODELS (defined once; nothing else in the notebook hardcodes a model name) ---
GROQ_MODEL = "openai/gpt-oss-20b"                 # preferred: supports tool calling
GROQ_FALLBACK_MODEL = "llama-3.3-70b-versatile"   # used automatically if the preferred model is unavailable

# 4) WEATHER LOCATION (default = the coordinates in the provided Open-Meteo CSV file name: 52.52N 13.42E = Berlin)
WEATHER_LAT, WEATHER_LON = 52.52, 13.42

# 5) E-MAIL (placeholder only; real sending is OFF by default, see Section 22)
REPORT_RECIPIENT = "recipient@example.com"
ENABLE_EMAIL_SENDING = False

# 6) MISC
SEED = 42
OUTPUT_DIR = Path("/kaggle/working") if Path("/kaggle/working").exists() else Path("./kaggle_working_local")
# ============================================================================
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
random.seed(SEED); np.random.seed(SEED)
print("Outputs will be written to:", OUTPUT_DIR.resolve())


# %% Cell 7 (markdown)
# # ### Workflow request (edit freely — nothing else needs to change)

# %% Cell 8 (code)
USER_REQUEST = """
Analyze the sales data, identify the key business KPIs,
top-performing products and countries, customer trends,
statistical anomalies, and relevant insights from the
annual report. Enrich the analysis with weather data
when relevant, generate recommendations, and produce
a professional business report.
"""


# %% Cell 9 (code)
import os


def resolve_groq_key() -> tuple[str, str]:
  """Return (key, source). Order: Kaggle Secrets -> environment variable -> GROQ_API_KEY variable above."""
  # 1. Try Kaggle Secrets (suppress linter warning with # type: ignore)
  try:
    from kaggle_secrets import UserSecretsClient  # type: ignore

    value = UserSecretsClient().get_secret("GROQ_API_KEY")
    if value and value.strip():
      return value.strip(), "Kaggle Secrets"
  except Exception:
    pass

  # 2. Try Environment Variable (Local .env or system environment)
  value = os.environ.get("GROQ_API_KEY", "").strip()
  if value:
    return value, "environment variable"

  # 3. Try Notebook Configuration Variable if defined
  config_key = globals().get("GROQ_API_KEY", "")
  if isinstance(config_key, str) and config_key.strip():
    return config_key.strip(), "configuration cell"

  return "", "not found"


_GROQ_KEY, _GROQ_KEY_SOURCE = (
    resolve_groq_key()
)  # kept in memory only; never printed or saved

print("GROQ API KEY LOCATION:")
print("1. Kaggle Secrets -> GROQ_API_KEY")
print("2. Environment variable / .env file")
print("3. Configuration cell near the beginning of this notebook")
print(
    "\nKey status:"
    f" {'FOUND via ' + _GROQ_KEY_SOURCE if _GROQ_KEY else 'NOT FOUND - LLM features will fall back to deterministic mode'}"
)
# %% Cell 10 (markdown)
# # # 6. Data Discovery
# # The notebook looks for the three files in `DATA_DIR` and then recursively in `/kaggle/input/` (and `./data` when run locally).

# %% Cell 11 (code)
def discover_files(data_dir: str) -> dict[str, Optional[Path]]:
    """Locate the three project files by name pattern. Returns {'sales','weather','report'} -> Path | None."""
    roots = [Path(data_dir), Path("/kaggle/input"), Path("./data"), Path(".")]
    seen, candidates = set(), []
    for root in roots:
        if not root.exists():
            continue
        try:
            for p in root.rglob("*"):
                if p.is_file() and p.suffix.lower() in {".xlsx", ".xls", ".csv", ".pdf"} and str(p.resolve()) not in seen:
                    seen.add(str(p.resolve())); candidates.append(p)
        except Exception:
            continue
        if len(candidates) > 5000:
            break
    def pick(pred):
        hits = [p for p in candidates if pred(p)]
        return hits[0] if hits else None
    sales = pick(lambda p: p.suffix.lower() in {".xlsx", ".xls"} and "retail" in p.name.lower())
    weather = pick(lambda p: p.suffix.lower() == ".csv" and "meteo" in p.name.lower())
    report = pick(lambda p: p.suffix.lower() == ".pdf" and ("annual" in p.name.lower() or "target" in p.name.lower())) \
        or pick(lambda p: p.suffix.lower() == ".pdf")
    return {"sales": sales, "weather": weather, "report": report}

DATA_PATHS = discover_files(DATA_DIR)
EXPECTED_NAMES = {"sales": "Online Retail.xlsx", "weather": "open-meteo-52.52N13.42E38m.csv",
                  "report": "2025-Annual-Report-Target-Corporation.pdf"}
for key, path in DATA_PATHS.items():
    print(f"{'FOUND  ' if path else 'MISSING'} {key:8s} -> {path if path else EXPECTED_NAMES[key]}")
if not all(DATA_PATHS.values()):
    print(f"""
Some files were not found. Fix it like this:
  1. Create a Kaggle Dataset containing the three files above (Kaggle -> Datasets -> New Dataset).
  2. Attach it to this notebook (Add Input). Files will appear at /kaggle/input/<dataset-slug>/.
  3. Set DATA_DIR in the configuration cell to that folder (auto-search also scans /kaggle/input).
Missing components will be skipped gracefully and listed in the report's Limitations section.""")


# %% Cell 12 (markdown)
# # ### Structured logging, secret redaction and tool timing
# # Every step writes a structured entry (`[LEVEL] [component] message`). Entries are redacted for the API key as a safety net and copied into the LangGraph state.

# %% Cell 13 (code)
class ExecutionLogger:
    """Collects structured log entries, prints them, and hands new entries to the graph state via flush()."""
    def __init__(self) -> None:
        self.entries: list[dict] = []
        self._cursor = 0

    @staticmethod
    def _redact(text: str) -> str:
        return text.replace(_GROQ_KEY, "***") if _GROQ_KEY else text

    def _add(self, level: str, component: str, message: str) -> None:
        entry = {"timestamp": dt.datetime.now().isoformat(timespec="seconds"), "level": level,
                 "component": component, "message": self._redact(str(message))}
        self.entries.append(entry)
        print(f"[{level}] [{component}] {entry['message']}")

    def info(self, component: str, message: str) -> None: self._add("INFO", component, message)
    def warning(self, component: str, message: str) -> None: self._add("WARNING", component, message)
    def error(self, component: str, message: str) -> None: self._add("ERROR", component, message)

    def flush(self) -> list[dict]:
        new = self.entries[self._cursor:]
        self._cursor = len(self.entries)
        return new

    def reset(self) -> None:
        self.entries, self._cursor = [], 0

LOG = ExecutionLogger()
TOOL_STATS: dict[str, list[float]] = {}     # tool name -> list of durations (seconds)
TOOL_ERRORS: dict[str, int] = {}            # tool name -> number of raised exceptions

def track_tool(name: str) -> Callable:
    """Decorator: measures execution time of a tool and counts exceptions (which are re-raised)."""
    def decorator(fn: Callable) -> Callable:
        @functools.wraps(fn)
        def wrapper(*args, **kwargs):
            t0 = time.perf_counter()
            try:
                return fn(*args, **kwargs)
            except Exception as exc:
                TOOL_ERRORS[name] = TOOL_ERRORS.get(name, 0) + 1
                LOG.error(name, f"{type(exc).__name__}: {exc}")
                raise
            finally:
                TOOL_STATS.setdefault(name, []).append(time.perf_counter() - t0)
        return wrapper
    return decorator

def jsonable(obj: Any) -> Any:
    """Recursively convert numpy/pandas/Path objects into plain JSON-serialisable Python objects."""
    if isinstance(obj, dict): return {str(k): jsonable(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple, set)): return [jsonable(v) for v in obj]
    if isinstance(obj, (np.integer,)): return int(obj)
    if isinstance(obj, (np.floating,)): return None if np.isnan(obj) else float(obj)
    if isinstance(obj, float): return None if math.isnan(obj) else obj
    if isinstance(obj, (np.bool_,)): return bool(obj)
    if isinstance(obj, (pd.Timestamp, dt.datetime, dt.date)): return str(obj)
    if isinstance(obj, Path): return str(obj)
    if obj is pd.NA or obj is pd.NaT: return None
    return obj


# %% Cell 14 (markdown)
# # ### Pydantic models (validation of request, quality report, KPIs and insights)

# %% Cell 15 (code)
from pydantic import BaseModel, Field, field_validator

class WorkflowRequest(BaseModel):
    """Validates the user's business-analysis request."""
    user_request: str

    @field_validator("user_request")
    @classmethod
    def _check(cls, v: str) -> str:
        v = " ".join(str(v).split())
        if len(v) < 15:
            raise ValueError("Request is too short (min 15 characters). Describe what should be analysed.")
        if len(v) > 4000:
            raise ValueError("Request is too long (max 4000 characters).")
        return v

class DataQualityReport(BaseModel):
    """Result of validate_sales_data()."""
    rows: int
    columns: int
    missing_values: dict = Field(default_factory=dict)
    duplicate_rows: int = 0
    invalid_quantity_rows: int = 0
    negative_quantity_rows: int = 0
    invalid_price_rows: int = 0
    invalid_date_rows: int = 0
    missing_customer_id_rows: int = 0
    empty_description_rows: int = 0
    cancellation_rows: int = 0
    outliers: dict = Field(default_factory=dict)
    defect_rows: int = 0
    quality_score_pct: float = 0.0
    needs_cleaning: bool = False
    blocking_issues: list = Field(default_factory=list)
    acceptable: bool = False
    is_cleaned: bool = False
    notes: list = Field(default_factory=list)

class KPIResult(BaseModel):
    """Deterministic KPI set (definitions in Section 14)."""
    total_revenue: float = Field(ge=0)
    returns_value: float = Field(ge=0)
    net_revenue: float
    total_orders: int = Field(ge=0)
    cancellation_invoices: int = Field(ge=0)
    unique_customers: int = Field(ge=0)
    unique_products: int = Field(ge=0)
    average_order_value: float = Field(ge=0)
    average_items_per_order: float = Field(ge=0)
    average_revenue_per_customer: float = Field(ge=0)
    return_rate_invoices_pct: float = Field(ge=0)
    return_rate_value_pct: float = Field(ge=0)
    transaction_lines: int = Field(ge=0)
    date_start: str = ""
    date_end: str = ""

CATEGORIES = ("FACT", "CALCULATED RESULT", "LLM INTERPRETATION", "RECOMMENDATION")

class Insight(BaseModel):
    """One labelled statement. category separates facts, calculations, interpretations and recommendations."""
    category: Literal["FACT", "CALCULATED RESULT", "LLM INTERPRETATION", "RECOMMENDATION"]
    text: str
    evidence: list = Field(default_factory=list)
    source: Literal["deterministic", "llm"] = "deterministic"
    note: str = ""


# %% Cell 16 (markdown)
# # # 7. Data Loading
# # `DataStore` keeps the heavy objects (DataFrames, the RAG index, weather data). The LangGraph state only carries small JSON summaries.

# %% Cell 17 (code)
@dataclass
class DataStore:
    """In-memory container for large objects shared by tools (kept out of the graph state on purpose)."""
    paths: dict = field(default_factory=dict)
    raw_df: Optional[pd.DataFrame] = None
    clean_df: Optional[pd.DataFrame] = None
    cleaning_report: dict = field(default_factory=dict)
    weather_df: Optional[pd.DataFrame] = None
    weather_meta: dict = field(default_factory=dict)
    rag_index: Any = None
    evidence: list = field(default_factory=list)     # retrieved annual-report chunks (deduplicated)

    def reset_run(self) -> None:
        self.raw_df = self.clean_df = self.weather_df = None
        self.cleaning_report, self.weather_meta, self.evidence = {}, {}, []

STORE = DataStore(paths={k: (str(v) if v else None) for k, v in DATA_PATHS.items()})
REQUIRED_COLUMNS = ["InvoiceNo", "StockCode", "Description", "Quantity", "InvoiceDate", "UnitPrice", "CustomerID", "Country"]
_LOAD_CACHE: dict = {}

@track_tool("load_sales_data")
def load_sales_data(path: str | Path) -> pd.DataFrame:
    """Load the Online Retail dataset (.xlsx/.xls/.csv) and verify the expected schema.

    Raises FileNotFoundError for a missing file and ValueError for a corrupted file or missing columns.
    Results are cached by (path, mtime) because reading the Excel file takes a while.
    """
    p = Path(path)
    if not p.exists():
        raise FileNotFoundError(f"Sales file not found: {p}. Expected a file like '{EXPECTED_NAMES['sales']}'.")
    cache_key = (str(p.resolve()), p.stat().st_mtime)
    if cache_key in _LOAD_CACHE:
        return _LOAD_CACHE[cache_key].copy()
    try:
        df = pd.read_csv(p, encoding="latin-1") if p.suffix.lower() == ".csv" else pd.read_excel(p, engine="openpyxl" if p.suffix.lower() == ".xlsx" else None)
    except Exception as exc:
        raise ValueError(f"Could not read '{p.name}' (corrupted or unsupported file): {type(exc).__name__}: {exc}") from exc
    df.columns = [str(c).strip() for c in df.columns]
    missing_cols = [c for c in REQUIRED_COLUMNS if c not in df.columns]
    if missing_cols:
        raise ValueError(f"Sales file is missing required columns: {missing_cols}. Found: {list(df.columns)}")
    _LOAD_CACHE.clear()
    _LOAD_CACHE[cache_key] = df
    return df.copy()

if DATA_PATHS["sales"]:
    t0 = time.perf_counter()
    try:
        _preview = load_sales_data(DATA_PATHS["sales"])
        print(f"Loaded {len(_preview):,} rows x {_preview.shape[1]} columns in {time.perf_counter()-t0:.1f}s")
        display(_preview.head())
        display(_preview.dtypes.to_frame("dtype"))
    except Exception as exc:
        print("Loading preview failed:", exc)
else:
    print("Sales file not found - see Section 6.")


# %% Cell 18 (markdown)
# # # 8. Data Quality
# #
# # `validate_sales_data()` checks missing values, duplicates, invalid quantities/prices/dates, missing customer IDs, negative transactions, empty descriptions and outliers (IQR, k=3).
# #
# # **Important decisions (documented, not hidden):**
# # * **Negative quantities are NOT blindly deleted.** Invoices starting with `C` are genuine cancellations/returns. They are kept and flagged (`IsReturn`) so they form a separate analytical subset (return rate) and are excluded from sales KPIs.
# # * **Missing `CustomerID`** (guest checkouts) is *not* a defect: rows stay in revenue KPIs but are excluded from customer-level analysis.
# # * **Defect rows** = duplicates, unparsable dates, non-positive prices, zero/invalid quantities, empty descriptions. The quality score is `100 x (1 - defect_rows / rows)`.
# # * The dataset is *acceptable* only when no blocking issue exists (empty data, missing columns, >50% defects) **and** nothing needs cleaning. Otherwise the workflow routes to the cleaning node and re-checks.

# %% Cell 19 (code)
def _iqr_upper(series: pd.Series, k: float = 3.0) -> tuple[float, int]:
    s = series.dropna()
    if s.empty:
        return float("nan"), 0
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    upper = float(q3 + k * (q3 - q1))
    return upper, int((s > upper).sum())

@track_tool("validate_sales_data")
def validate_sales_data(df: pd.DataFrame, is_cleaned: bool = False) -> dict:
    """Run the data-quality checks and return a DataQualityReport as a dict (see Section 8 for definitions)."""
    n = len(df)
    missing_cols = [c for c in REQUIRED_COLUMNS if c not in df.columns]
    if n == 0 or missing_cols:
        issues = (["Dataset is empty"] if n == 0 else []) + ([f"Missing columns: {missing_cols}"] if missing_cols else [])
        return DataQualityReport(rows=n, columns=df.shape[1], blocking_issues=issues, acceptable=False,
                                 is_cleaned=is_cleaned).model_dump()

    dates = pd.to_datetime(df["InvoiceDate"], errors="coerce")
    qty = pd.to_numeric(df["Quantity"], errors="coerce")
    price = pd.to_numeric(df["UnitPrice"], errors="coerce")
    invoice = df["InvoiceNo"].astype(str)
    is_cancel = invoice.str.startswith("C")
    empty_desc = df["Description"].isna() | df["Description"].fillna("").astype(str).str.strip().eq("")

    dup_mask = df.duplicated(keep="first")
    bad_date = dates.isna()
    bad_price = price.isna() | (price <= 0)
    bad_qty = qty.isna() | (qty == 0)
    defect = dup_mask | bad_date | bad_price | bad_qty | empty_desc

    valid_sales = (qty > 0) & (price > 0) & ~is_cancel
    q_up, q_n = _iqr_upper(qty[valid_sales]); p_up, p_n = _iqr_upper(price[valid_sales])
    r_up, r_n = _iqr_upper((qty * price)[valid_sales])
    outliers = {
        "method": "IQR, upper fence = Q3 + 3.0 x IQR (extreme outliers only), computed on valid sales lines",
        "quantity": {"upper_fence": q_up, "count": q_n}, "unit_price": {"upper_fence": p_up, "count": p_n},
        "line_revenue": {"upper_fence": r_up, "count": r_n},
    }
    defect_rows = int(defect.sum())
    blocking = []
    if defect_rows / n > 0.5:
        blocking.append(f"More than 50% of rows are defective ({defect_rows:,}/{n:,})")
    needs_cleaning = defect_rows > 0
    report = DataQualityReport(
        rows=n, columns=df.shape[1],
        missing_values={c: int(v) for c, v in df.isna().sum().items()},
        duplicate_rows=int(dup_mask.sum()), invalid_quantity_rows=int(bad_qty.sum()),
        negative_quantity_rows=int((qty < 0).sum()), invalid_price_rows=int(bad_price.sum()),
        invalid_date_rows=int(bad_date.sum()), missing_customer_id_rows=int(df["CustomerID"].isna().sum()),
        empty_description_rows=int(empty_desc.sum()), cancellation_rows=int(is_cancel.sum()),
        outliers=outliers, defect_rows=defect_rows, quality_score_pct=round(100 * (1 - defect_rows / n), 3),
        needs_cleaning=needs_cleaning, blocking_issues=blocking,
        acceptable=(not blocking) and (not needs_cleaning), is_cleaned=is_cleaned,
        notes=["Negative quantities are kept as returns/cancellations (flag IsReturn), not deleted.",
               "Missing CustomerID rows stay in revenue KPIs and are excluded from customer analytics.",
               "Outliers are reported, not removed; they are analysed as statistical anomalies."],
    )
    return jsonable(report.model_dump())


# %% Cell 20 (markdown)
# # # 9. Data Preprocessing
# # `clean_sales_data()` applies only defensible fixes and reports what it did:
# # 1. Normalise text columns (trim; `StockCode` upper-cased so `85123a` = `85123A`).
# # 2. Parse dates; drop unparsable ones. Coerce numerics; drop unparsable.
# # 3. Drop exact duplicate rows.
# # 4. Drop zero quantities and non-positive prices (these are manual adjustments / free samples, not sales).
# # 5. Fill empty descriptions with `UNKNOWN` (rows kept).
# # 6. **Keep** cancellations/returns (negative quantities on `C...` invoices).

# %% Cell 21 (code)
@track_tool("clean_sales_data")
def clean_sales_data(df: pd.DataFrame) -> tuple[pd.DataFrame, dict]:
    """Clean the raw dataset. Returns (clean_df, report) where report documents every removal/fix."""
    rep: dict[str, Any] = {"rows_before": int(len(df))}
    out = df.copy()
    out["InvoiceNo"] = out["InvoiceNo"].astype(str).str.strip()
    out["StockCode"] = out["StockCode"].astype(str).str.strip().str.upper()
    out["Country"] = out["Country"].astype(str).str.strip()
    out["Description"] = out["Description"].fillna("").astype(str).str.strip()
    empty = out["Description"].eq("")
    rep["empty_descriptions_filled"] = int(empty.sum())
    out.loc[empty, "Description"] = "UNKNOWN"

    out["InvoiceDate"] = pd.to_datetime(out["InvoiceDate"], errors="coerce")
    out["Quantity"] = pd.to_numeric(out["Quantity"], errors="coerce")
    out["UnitPrice"] = pd.to_numeric(out["UnitPrice"], errors="coerce")
    bad = out["InvoiceDate"].isna() | out["Quantity"].isna() | out["UnitPrice"].isna()
    rep["unparsable_rows_dropped"] = int(bad.sum())
    out = out[~bad]

    before = len(out)
    out = out.drop_duplicates()
    rep["duplicates_dropped"] = int(before - len(out))

    zero_qty = out["Quantity"] == 0
    rep["zero_quantity_dropped"] = int(zero_qty.sum())
    out = out[~zero_qty]
    neg_price = out["UnitPrice"] < 0
    zero_price = out["UnitPrice"] == 0
    rep["negative_price_dropped"] = int(neg_price.sum())
    rep["zero_price_dropped"] = int(zero_price.sum())
    out = out[~(neg_price | zero_price)]

    cid = pd.to_numeric(out["CustomerID"], errors="coerce")
    out["CustomerID"] = cid.astype("Int64").astype("string")
    out["Quantity"] = out["Quantity"].astype("int64")
    rep["returns_kept"] = int(((out["Quantity"] < 0) | out["InvoiceNo"].str.startswith("C")).sum())
    rep["rows_after"] = int(len(out))
    rep["rows_removed_total"] = rep["rows_before"] - rep["rows_after"]
    rep["decision_note"] = ("Cancellations/returns (negative quantity, invoices starting with 'C') were kept and flagged; "
                            "zero/negative-price lines were removed because they are adjustments or free items, not sales.")
    return out.reset_index(drop=True), rep


# %% Cell 22 (markdown)
# # # 10. Feature Engineering
# #
# # | Feature | Definition |
# # |---|---|
# # | `Revenue` | `Quantity x UnitPrice` (negative for returns) |
# # | `Date`, `Year`, `Month`, `Day`, `Hour`, `DayOfWeek`, `YearMonth` | calendar parts of `InvoiceDate` |
# # | `IsReturn` | `InvoiceNo` starts with `C` **or** `Quantity < 0` |
# # | `IsNonProduct` | `StockCode` does not start with 5 digits (POST, DOT, M, BANK CHARGES ... = service/fee lines; excluded from product rankings) |
# # | `CustomerTotalSpend` | sum of sales revenue per customer (returns excluded) |
# # | `CustomerOrderCount` | distinct sales invoices per customer |
# # | `CustomerAverageOrderValue` | `CustomerTotalSpend / CustomerOrderCount` |

# %% Cell 23 (code)
@track_tool("engineer_features")
def engineer_features(df: pd.DataFrame) -> pd.DataFrame:
    """Add the derived features described in Section 10 and return a new DataFrame."""
    out = df.copy()
    out["Revenue"] = out["Quantity"] * out["UnitPrice"]
    d = out["InvoiceDate"]
    out["Date"] = d.dt.normalize()
    out["Year"], out["Month"], out["Day"], out["Hour"] = d.dt.year, d.dt.month, d.dt.day, d.dt.hour
    out["DayOfWeek"] = d.dt.day_name()
    out["YearMonth"] = d.dt.strftime("%Y-%m")
    out["IsReturn"] = out["InvoiceNo"].astype(str).str.startswith("C") | (out["Quantity"] < 0)
    out["IsNonProduct"] = ~out["StockCode"].astype(str).str.match(r"^\d{5}")
    sales_with_id = out[(~out["IsReturn"]) & out["CustomerID"].notna()]
    grp = sales_with_id.groupby("CustomerID")
    spend, orders = grp["Revenue"].sum(), grp["InvoiceNo"].nunique()
    out["CustomerTotalSpend"] = out["CustomerID"].map(spend)
    out["CustomerOrderCount"] = out["CustomerID"].map(orders)
    out["CustomerAverageOrderValue"] = out["CustomerID"].map(spend / orders)
    return out

def _ensure_features(df: pd.DataFrame) -> pd.DataFrame:
    return df if {"Revenue", "IsReturn", "IsNonProduct", "YearMonth"} <= set(df.columns) else engineer_features(df)

def _sales(df: pd.DataFrame) -> pd.DataFrame:
    """Sales subset: non-return lines with positive quantity."""
    return df[(~df["IsReturn"]) & (df["Quantity"] > 0)]

def _returns(df: pd.DataFrame) -> pd.DataFrame:
    """Returns/cancellations subset."""
    return df[df["IsReturn"]]

def _safe_div(a: float, b: float) -> float:
    return float(a) / float(b) if b else 0.0

if DATA_PATHS["sales"] and "_preview" in globals():
    _c, _r = clean_sales_data(_preview)
    _f = engineer_features(_c)
    print("Cleaning report:", json.dumps(jsonable(_r), indent=2))
    display(_f.head())
    del _preview, _c, _f


# %% Cell 24 (markdown)
# # # 11. PDF Retrieval / RAG
# #
# # Lightweight, dependency-free-on-GPU RAG over the Target annual report:
# #
# # ```
# # PDF -> page text (pypdf) -> overlapping word chunks (page-aware) -> TF-IDF (1-2 grams) -> cosine similarity -> top-k evidence -> LLM summary
# # ```
# # TF-IDF was chosen over neural embeddings because it is fast, deterministic, needs no model download and works reliably offline on Kaggle. Each chunk keeps its **PDF page number**, so citations read `Annual Report - Page X` (physical PDF page, which may differ from the printed page number).

# %% Cell 25 (code)
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel

class AnnualReportIndex:
    """Page-aware TF-IDF retrieval index over a PDF."""

    def __init__(self, pdf_path: str | Path, chunk_words: int = 180, overlap: int = 40, min_chunk_words: int = 25) -> None:
        self.pdf_path = Path(pdf_path)
        self.chunk_words, self.overlap, self.min_chunk_words = chunk_words, overlap, min_chunk_words
        self.chunks: list[dict] = []
        self.n_pages = 0
        self.pages_without_text = 0
        self._vectorizer: Optional[TfidfVectorizer] = None
        self._matrix = None

    def build(self) -> "AnnualReportIndex":
        """Extract text, chunk it and fit the TF-IDF index. Raises on unreadable/empty PDFs."""
        if not self.pdf_path.exists():
            raise FileNotFoundError(f"Annual report not found: {self.pdf_path}")
        try:
            from pypdf import PdfReader
            reader = PdfReader(str(self.pdf_path))
            self.n_pages = len(reader.pages)
        except Exception as exc:
            raise ValueError(f"Could not open PDF '{self.pdf_path.name}': {type(exc).__name__}: {exc}") from exc
        stride = max(1, self.chunk_words - self.overlap)
        for page_no, page in enumerate(reader.pages, start=1):
            try:
                text = page.extract_text() or ""
            except Exception:
                text = ""
            words = re.sub(r"\s+", " ", text).strip().split(" ")
            words = [w for w in words if w]
            if len(words) < self.min_chunk_words:
                self.pages_without_text += 1 if not words else 0
                if len(words) >= 10:
                    self.chunks.append({"chunk_id": len(self.chunks), "page": page_no, "text": " ".join(words)})
                continue
            for start in range(0, len(words), stride):
                piece = words[start:start + self.chunk_words]
                if len(piece) >= self.min_chunk_words:
                    self.chunks.append({"chunk_id": len(self.chunks), "page": page_no, "text": " ".join(piece)})
        if not self.chunks:
            raise ValueError("PDF text extraction produced no text (scanned/image-only PDF? OCR would be required).")
        texts = [c["text"] for c in self.chunks]
        self._vectorizer = TfidfVectorizer(stop_words="english", ngram_range=(1, 2), sublinear_tf=True, max_df=0.9)
        try:
            self._matrix = self._vectorizer.fit_transform(texts)
        except ValueError:
            # Very small or repetitive PDFs can prune every term with the
            # default English stop-word/max_df settings. Relax only the
            # vectorizer, never the page-aware chunking or citation logic.
            self._vectorizer = TfidfVectorizer(stop_words=None, ngram_range=(1, 2), sublinear_tf=True, max_df=1.0)
            self._matrix = self._vectorizer.fit_transform(texts)
        return self

    def search(self, query: str, top_k: int = 3, min_score: float = 0.02) -> list[dict]:
        """Return the top_k chunks most similar to the query, each with page, score, text and citation."""
        if not query or not query.strip():
            raise ValueError("Empty search query.")
        if self._vectorizer is None:
            raise RuntimeError("Index not built.")
        sims = linear_kernel(self._vectorizer.transform([query]), self._matrix).ravel()
        order = np.argsort(-sims)[:max(1, int(top_k))]
        return [{"chunk_id": self.chunks[i]["chunk_id"], "page": self.chunks[i]["page"], "score": round(float(sims[i]), 4),
                 "text": self.chunks[i]["text"], "citation": f"Annual Report - Page {self.chunks[i]['page']}"}
                for i in order if sims[i] >= min_score]

def get_rag_index() -> AnnualReportIndex:
    """Build the index lazily (once) from the discovered PDF path."""
    if STORE.rag_index is None:
        if not STORE.paths.get("report"):
            raise FileNotFoundError(f"Annual report PDF not found. Expected: {EXPECTED_NAMES['report']}")
        t0 = time.perf_counter()
        STORE.rag_index = AnnualReportIndex(STORE.paths["report"]).build()
        LOG.info("rag", f"Index built: {STORE.rag_index.n_pages} pages, {len(STORE.rag_index.chunks)} chunks in {time.perf_counter()-t0:.1f}s")
    return STORE.rag_index

@track_tool("search_annual_report")
def search_annual_report(query: str, top_k: int = 3) -> list[dict]:
    """Retrieve the annual-report passages most relevant to `query`. New chunks are added to STORE.evidence."""
    results = get_rag_index().search(query, top_k=top_k)
    known = {e["chunk_id"]: e for e in STORE.evidence}
    for r in results:
        item = dict(r, query=query)
        if r["chunk_id"] not in known or r["score"] > known[r["chunk_id"]]["score"]:
            known[r["chunk_id"]] = item
    STORE.evidence = sorted(known.values(), key=lambda e: -e["score"])
    return results

if STORE.paths.get("report"):
    try:
        _demo = search_annual_report("net sales growth and comparable sales", top_k=3)
        print(f"Indexed {STORE.rag_index.n_pages} pages / {len(STORE.rag_index.chunks)} chunks. Example retrieval:\n")
        for r in _demo:
            print(f"[{r['citation']}]  score={r['score']}\n   {r['text'][:350]}...\n")
        STORE.evidence = []   # demo retrieval is not part of the workflow evidence
    except Exception as exc:
        print("RAG demo failed:", exc)
else:
    print("Annual report PDF not found - RAG skipped.")


# %% Cell 26 (markdown)
# # # 12. External API Tools (Open-Meteo)
# #
# # `get_weather_data()` calls the **live Open-Meteo API** (no key needed):
# # * historical daily data from `archive-api.open-meteo.com` (temperature mean/max/min, precipitation),
# # * or current conditions from `api.open-meteo.com`.
# #
# # Robustness: timeout + retries; empty/invalid responses are treated as failures; on any failure the tool **falls back to the local CSV** and says so in its result (`fallback_used=True`).
# #
# # *Important caveat:* the provided coordinates are Berlin, while the retailer's customers are mostly in the UK. Weather analysis is therefore an **exploratory proxy**, not evidence of causation (this is repeated in the report).

# %% Cell 27 (code)
import requests

OPEN_METEO_ARCHIVE = "https://archive-api.open-meteo.com/v1/archive"
OPEN_METEO_FORECAST = "https://api.open-meteo.com/v1/forecast"

def _parse_open_meteo_csv(path: str | Path) -> pd.DataFrame:
    """Parse an Open-Meteo CSV export (metadata header + data table) into a DAILY DataFrame.

    Output columns (when available): date, temp_mean, temp_max, temp_min, precip_sum. Hourly data is aggregated to daily.
    """
    lines = Path(path).read_text(encoding="utf-8-sig", errors="replace").splitlines()
    idx = next((i for i, l in enumerate(lines) if l.strip().lower().startswith("time")), None)
    if idx is None:
        raise ValueError("Open-Meteo CSV has no 'time' header row.")
    raw = pd.read_csv(io.StringIO("\n".join(lines[idx:])))
    tcol = raw.columns[0]
    raw[tcol] = pd.to_datetime(raw[tcol], errors="coerce")
    raw = raw.dropna(subset=[tcol])
    if raw.empty:
        raise ValueError("Open-Meteo CSV contains no valid timestamps.")
    mapping: dict[str, str] = {}
    for c in raw.columns[1:]:
        lc = c.lower()
        if "temperature" in lc and "max" in lc: mapping[c] = "temp_max"
        elif "temperature" in lc and "min" in lc: mapping[c] = "temp_min"
        elif "temperature" in lc: mapping.setdefault(c, "temp_mean")
        elif "precip" in lc or "rain" in lc: mapping.setdefault(c, "precip_sum")
    if not mapping:
        raise ValueError(f"No temperature/precipitation columns recognised in {list(raw.columns)}")
    data = pd.DataFrame({"date": raw[tcol].dt.normalize()})
    for src, dst in mapping.items():
        data[dst] = pd.to_numeric(raw[src], errors="coerce")
    g = data.groupby("date")
    daily = pd.DataFrame(index=g.size().index)
    if "temp_mean" in data: daily["temp_mean"] = g["temp_mean"].mean()
    if "temp_max" in data: daily["temp_max"] = g["temp_max"].max()
    elif "temp_mean" in data: daily["temp_max"] = g["temp_mean"].max()
    if "temp_min" in data: daily["temp_min"] = g["temp_min"].min()
    elif "temp_mean" in data: daily["temp_min"] = g["temp_mean"].min()
    if "temp_mean" not in daily and {"temp_max", "temp_min"} <= set(daily.columns):
        daily["temp_mean"] = (daily["temp_max"] + daily["temp_min"]) / 2
    if "precip_sum" in data: daily["precip_sum"] = g["precip_sum"].sum(min_count=1)
    return daily.reset_index()

@track_tool("load_local_weather")
def load_local_weather(path: str | Path) -> pd.DataFrame:
    """Load the local Open-Meteo CSV as a daily DataFrame."""
    if not path or not Path(path).exists():
        raise FileNotFoundError(f"Local weather CSV not found. Expected: {EXPECTED_NAMES['weather']}")
    return _parse_open_meteo_csv(path)

@track_tool("fetch_open_meteo_api")
def fetch_open_meteo_api(start_date: str, end_date: str, latitude: float, longitude: float,
                         mode: str = "historical", timeout: int = 20, retries: int = 2) -> pd.DataFrame | dict:
    """Call the Open-Meteo API. historical -> daily DataFrame; current -> dict. Raises on timeout/HTTP error/empty payload."""
    if mode == "current":
        url, params = OPEN_METEO_FORECAST, {"latitude": latitude, "longitude": longitude, "current_weather": "true", "timezone": "auto"}
    else:
        url = OPEN_METEO_ARCHIVE
        params = {"latitude": latitude, "longitude": longitude, "start_date": start_date, "end_date": end_date, "timezone": "auto",
                  "daily": "temperature_2m_mean,temperature_2m_max,temperature_2m_min,precipitation_sum"}
    last_exc: Optional[Exception] = None
    for attempt in range(retries + 1):
        try:
            resp = requests.get(url, params=params, timeout=timeout)
            resp.raise_for_status()
            payload = resp.json()
            if mode == "current":
                if not payload.get("current_weather"):
                    raise ValueError("Empty current_weather payload")
                return payload["current_weather"]
            daily = payload.get("daily") or {}
            if not daily.get("time"):
                raise ValueError("Empty daily payload from Open-Meteo")
            df = pd.DataFrame({"date": pd.to_datetime(daily["time"]), "temp_mean": daily.get("temperature_2m_mean"),
                               "temp_max": daily.get("temperature_2m_max"), "temp_min": daily.get("temperature_2m_min"),
                               "precip_sum": daily.get("precipitation_sum")})
            return df
        except Exception as exc:
            last_exc = exc
            if attempt < retries:
                time.sleep(1.5 * (attempt + 1))
    raise RuntimeError(f"Open-Meteo API failed after {retries + 1} attempts: {type(last_exc).__name__}: {last_exc}")

def _weather_summary(df: pd.DataFrame) -> dict:
    out = {"rows": int(len(df)), "date_start": str(df["date"].min().date()), "date_end": str(df["date"].max().date())}
    if "temp_mean" in df: out.update(temp_mean_avg=round(float(df["temp_mean"].mean()), 2))
    if "temp_min" in df: out.update(temp_min=round(float(df["temp_min"].min()), 2))
    if "temp_max" in df: out.update(temp_max=round(float(df["temp_max"].max()), 2))
    if "precip_sum" in df: out.update(precip_total_mm=round(float(df["precip_sum"].sum()), 1), rainy_days=int((df["precip_sum"] > 0.1).sum()))
    return out

@track_tool("get_weather_data")
def get_weather_data(start_date: str = "", end_date: str = "", latitude: float = WEATHER_LAT, longitude: float = WEATHER_LON,
                     mode: str = "historical", store: bool = True) -> dict:
    """Retrieve weather via the live Open-Meteo API, falling back to the local CSV if the API is unavailable.

    Args:
        start_date/end_date: 'YYYY-MM-DD'. Empty = the sales-data date range. (Ignored for mode='current'.)
        latitude/longitude: location (defaults to the coordinates of the provided CSV).
        mode: 'historical' (daily archive) or 'current' (latest conditions).
        store: keep the daily DataFrame in STORE.weather_df for later correlation analysis.
    """
    mode = (mode or "historical").lower()
    if mode not in {"historical", "current"}:
        raise ValueError("mode must be 'historical' or 'current'")
    latitude = WEATHER_LAT if latitude is None else float(latitude)
    longitude = WEATHER_LON if longitude is None else float(longitude)
    if not (-90 <= latitude <= 90 and -180 <= longitude <= 180):
        raise ValueError("Invalid coordinates")

    if mode == "current":
        try:
            cur = fetch_open_meteo_api("", "", latitude, longitude, mode="current")
            return jsonable({"source": "open-meteo API (current)", "api_available": True, "fallback_used": False, "mode": mode,
                             "latitude": latitude, "longitude": longitude, "current": cur})
        except Exception as exc:
            LOG.warning("get_weather_data", f"Current-weather API unavailable ({exc}); using latest row of local CSV (NOT current).")
            local = load_local_weather(STORE.paths.get("weather"))
            return jsonable({"source": "local CSV (API fallback, latest available day - not current)", "api_available": False,
                             "fallback_used": True, "mode": mode, "api_error": str(exc)[:300], "latest_local_day": _weather_summary(local.tail(1))})

    if not start_date or not end_date:
        ref = STORE.clean_df if STORE.clean_df is not None else STORE.raw_df
        if ref is None:
            raise ValueError("No dates given and no sales data loaded to derive them from.")
        d = pd.to_datetime(ref["InvoiceDate"], errors="coerce")
        start_date, end_date = str(d.min().date()), str(d.max().date())
    s, e = pd.to_datetime(start_date), pd.to_datetime(end_date)
    if pd.isna(s) or pd.isna(e) or s > e:
        raise ValueError(f"Invalid date range: {start_date} -> {end_date}")
    latest_allowed = pd.Timestamp.today().normalize() - pd.Timedelta(days=5)   # archive lags a few days
    e = min(e, latest_allowed)
    s_str, e_str = str(s.date()), str(e.date())

    api_error, df, source = None, None, None
    try:
        df = fetch_open_meteo_api(s_str, e_str, latitude, longitude, mode="historical")
        source = "open-meteo API (historical archive)"
    except Exception as exc:
        api_error = str(exc)
        LOG.warning("get_weather_data", f"API unavailable -> falling back to local CSV. Reason: {api_error[:200]}")
        local = load_local_weather(STORE.paths.get("weather"))
        df = local[(local["date"] >= s) & (local["date"] <= e)]
        source = "local CSV (API fallback)"
        if df.empty:
            raise ValueError("Local CSV has no rows in the requested range either.") from exc
    df = df.dropna(how="all", subset=[c for c in df.columns if c != "date"]).reset_index(drop=True)
    if df.empty:
        raise ValueError("Weather data is empty after cleaning.")
    summary = _weather_summary(df)
    result = {"source": source, "api_available": api_error is None, "fallback_used": api_error is not None, "mode": mode,
              "latitude": latitude, "longitude": longitude, "api_error": api_error[:300] if api_error else None, **summary}
    if store:
        STORE.weather_df, STORE.weather_meta = df, result
    return jsonable(result)


# %% Cell 28 (markdown)
# # # 13. LLM Configuration (Groq)
# #
# # `init_llm()` creates a `ChatGroq` client with `GROQ_MODEL`, runs a tiny connectivity test and **automatically falls back** to `GROQ_FALLBACK_MODEL`. If no key / no model works, `LLM` stays `None` and every LLM-dependent step uses its deterministic fallback.
# #
# # | Component | Needs Groq? | Fallback when Groq is unavailable |
# # |---|---|---|
# # | Retrieval Agent tool selection | Optional | deterministic tool sequence |
# # | Analysis Agent tool selection | Optional | deterministic tool sequence |
# # | Annual-report summary | **Yes** | extractive snippets with page citations |
# # | Insights / interpretations | **Yes** | rule-based calculated results and recommendations |
# # | Executive summary / conclusion | **Yes** | template text built from KPIs |
# # | All numbers, charts, PDF, e-mail summary | No | - |

# %% Cell 29 (code)
LLM = None
LLM_STATUS: dict[str, Any] = {"available": False, "model": None, "key_source": _GROQ_KEY_SOURCE, "reason": "not initialised", "calls": 0, "errors": 0}

def init_llm() -> None:
    """Initialise the Groq chat model (preferred model, then fallback model). Never prints the key."""
    global LLM
    LLM = None
    LLM_STATUS.update(available=False, model=None, reason="", calls=0, errors=0)
    if not _GROQ_KEY:
        LLM_STATUS["reason"] = "No GROQ_API_KEY found (Kaggle Secrets / configuration cell)."
        LOG.warning("llm", LLM_STATUS["reason"] + " Running in deterministic mode.")
        return
    try:
        from langchain_groq import ChatGroq
    except Exception as exc:
        LLM_STATUS["reason"] = f"langchain_groq import failed: {exc}"
        LOG.warning("llm", LLM_STATUS["reason"]); return
    reasons = []
    for model in [GROQ_MODEL, GROQ_FALLBACK_MODEL]:
        try:
            candidate = ChatGroq(model=model, api_key=_GROQ_KEY, temperature=0, max_retries=2, timeout=60)
            reply = candidate.invoke("Reply with the single word: OK")
            if not str(getattr(reply, "content", "")).strip():
                raise ValueError("empty reply")
            LLM = candidate
            LLM_STATUS.update(available=True, model=model, reason="ok")
            LOG.info("llm", f"Groq model ready: {model}" + ("" if model == GROQ_MODEL else f" (fallback; preferred {GROQ_MODEL} unavailable)"))
            return
        except Exception as exc:
            reasons.append(f"{model}: {type(exc).__name__}: {str(exc)[:160]}")
            LOG.warning("llm", f"Model {model} unusable - {type(exc).__name__}")
    LLM_STATUS["reason"] = " | ".join(reasons)
    LOG.warning("llm", "No Groq model available. Running in deterministic mode.")

init_llm()
print("LLM status:", {k: v for k, v in LLM_STATUS.items()})


# %% Cell 30 (code)
def llm_invoke(runnable: Any, messages: Any, retries: int = 3) -> Any:
    """Invoke an LLM runnable with retries for rate limits/timeouts/5xx (other errors are raised immediately)."""
    last: Optional[Exception] = None
    for attempt in range(retries):
        try:
            LLM_STATUS["calls"] += 1
            return runnable.invoke(messages)
        except Exception as exc:
            LLM_STATUS["errors"] += 1
            last = exc
            text = f"{type(exc).__name__} {exc}".lower()
            if any(k in text for k in ("rate", "429", "timeout", "timed out", "503", "502", "500", "overloaded", "connection")):
                time.sleep(min(25, 6 * (attempt + 1)))
                continue
            raise
    raise last  # type: ignore[misc]

def extract_json(text: str) -> Any:
    """Extract the first JSON object/array from an LLM reply (handles markdown fences and surrounding prose)."""
    if not isinstance(text, str) or not text.strip():
        raise ValueError("empty LLM reply")
    cleaned = re.sub(r"```(?:json)?", "", text).strip()
    for opener, closer in (("{", "}"), ("[", "]")):
        a, b = cleaned.find(opener), cleaned.rfind(closer)
        if a != -1 and b > a:
            try:
                return json.loads(cleaned[a:b + 1])
            except json.JSONDecodeError:
                continue
    raise ValueError("no valid JSON found in LLM reply")

def llm_text(system: str, user: str) -> str:
    """Single LLM call returning text. Raises if the LLM is unavailable."""
    if LLM is None:
        raise RuntimeError("LLM unavailable")
    from langchain_core.messages import SystemMessage, HumanMessage
    reply = llm_invoke(LLM, [SystemMessage(content=system), HumanMessage(content=user)])
    return str(reply.content).strip()

def llm_json(system: str, user: str, attempts: int = 2) -> Any:
    """LLM call that must return JSON; re-asks once if the reply cannot be parsed."""
    last: Optional[Exception] = None
    for i in range(attempts):
        text = llm_text(system, user + ("" if i == 0 else "\n\nYour previous reply was not valid JSON. Reply with ONLY one valid JSON object."))
        try:
            return extract_json(text)
        except Exception as exc:
            last = exc
    raise ValueError(f"LLM did not return valid JSON: {last}")


# %% Cell 31 (markdown)
# # # 14. Agent Tools
# #
# # ### 14.1 Deterministic analytics (Pandas) — *the LLM never does this arithmetic*
# #
# # **KPI definitions**
# # * **Sales lines** = non-return lines with `Quantity > 0` (after cleaning). **Returns** = lines flagged `IsReturn`.
# # * `Total Revenue` = sum of `Revenue` over sales lines (gross, before returns). `Net Revenue` = Total Revenue - value of returns.
# # * `Total Orders` = distinct sales `InvoiceNo`. `Unique Customers` = distinct non-missing `CustomerID` in sales.
# # * `Unique Products` = distinct product `StockCode` (service/fee codes excluded).
# # * `Average Order Value` = Total Revenue / Total Orders. `Average Items per Order` = units sold / Total Orders.
# # * `Return Rate (invoices)` = cancellation invoices / (sales invoices + cancellation invoices). `Return Rate (value)` = returns value / Total Revenue.
# # * The **last month in the data may be partial**; it is flagged and excluded from growth/trend statistics.

# %% Cell 32 (code)
@track_tool("calculate_kpis")
def calculate_kpis(df: pd.DataFrame) -> dict:
    """Compute the deterministic KPI set (see definitions above). Returns a KPIResult dict."""
    df = _ensure_features(df)
    s, r = _sales(df), _returns(df)
    if s.empty:
        raise ValueError("No sales lines available for KPI calculation.")
    total_revenue = float(s["Revenue"].sum())
    returns_value = float(abs(r["Revenue"].sum()))
    orders, canc = int(s["InvoiceNo"].nunique()), int(r["InvoiceNo"].nunique())
    customers = int(s["CustomerID"].nunique())
    kpi = KPIResult(
        total_revenue=round(total_revenue, 2), returns_value=round(returns_value, 2), net_revenue=round(total_revenue - returns_value, 2),
        total_orders=orders, cancellation_invoices=canc, unique_customers=customers,
        unique_products=int(s.loc[~s["IsNonProduct"], "StockCode"].nunique()),
        average_order_value=round(_safe_div(total_revenue, orders), 2),
        average_items_per_order=round(_safe_div(s["Quantity"].sum(), orders), 2),
        average_revenue_per_customer=round(_safe_div(s.loc[s["CustomerID"].notna(), "Revenue"].sum(), customers), 2),
        return_rate_invoices_pct=round(100 * _safe_div(canc, orders + canc), 2),
        return_rate_value_pct=round(100 * _safe_div(returns_value, total_revenue), 2),
        transaction_lines=int(len(s)), date_start=str(s["InvoiceDate"].min().date()), date_end=str(s["InvoiceDate"].max().date()))
    return jsonable(kpi.model_dump())

@track_tool("analyze_sales_trends")
def analyze_sales_trends(df: pd.DataFrame) -> dict:
    """Monthly revenue/orders/customers/AOV, MoM growth, partial-month flags, trend direction, weekday and hour profiles."""
    s = _sales(_ensure_features(df))
    if s.empty:
        raise ValueError("No sales lines for trend analysis.")
    m = (s.groupby("YearMonth").agg(revenue=("Revenue", "sum"), orders=("InvoiceNo", "nunique"),
                                   customers=("CustomerID", "nunique"), units=("Quantity", "sum"))
           .reset_index().sort_values("YearMonth").reset_index(drop=True))
    m["aov"] = m["revenue"] / m["orders"]
    first, last = s["InvoiceDate"].min(), s["InvoiceDate"].max()
    partial_first = first.normalize() > first.to_period("M").start_time
    partial_last = last.normalize() < (last + pd.offsets.MonthEnd(0)).normalize()
    m["is_partial"] = False
    if partial_first: m.loc[m.index[0], "is_partial"] = True
    if partial_last: m.loc[m.index[-1], "is_partial"] = True
    m["mom_growth_pct"] = m["revenue"].pct_change() * 100
    m.loc[m["is_partial"] | m["is_partial"].shift(1, fill_value=False), "mom_growth_pct"] = np.nan
    complete = m[~m["is_partial"]]
    result: dict[str, Any] = {"monthly": m.round(2).to_dict("records"), "partial_months": m.loc[m["is_partial"], "YearMonth"].tolist()}
    if len(complete) >= 3:
        slope = float(np.polyfit(np.arange(len(complete)), complete["revenue"].to_numpy(), 1)[0])
        rel = slope / float(complete["revenue"].mean())
        result.update(trend_slope_per_month=round(slope, 2), trend_relative_slope=round(rel, 4),
                      trend_direction="increasing" if rel > 0.02 else "decreasing" if rel < -0.02 else "roughly stable")
    if not complete.empty:
        pk, lo = complete.loc[complete["revenue"].idxmax()], complete.loc[complete["revenue"].idxmin()]
        result.update(peak_month={"month": pk["YearMonth"], "revenue": round(float(pk["revenue"]), 2)},
                      lowest_month={"month": lo["YearMonth"], "revenue": round(float(lo["revenue"]), 2)})
    order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
    dow = s.groupby("DayOfWeek")["Revenue"].sum().reindex(order).fillna(0)
    result["revenue_by_weekday"] = {k: round(float(v), 2) for k, v in dow.items()}
    hour = s.groupby("Hour")["Revenue"].sum()
    result["peak_hour"] = int(hour.idxmax())
    result["revenue_by_hour"] = {int(k): round(float(v), 2) for k, v in hour.items()}
    return jsonable(result)

@track_tool("get_top_products")
def get_top_products(df: pd.DataFrame, n: int = 10) -> dict:
    """Top-n products (service/fee lines excluded) by revenue and by units."""
    n = max(1, min(int(n), 50))
    s = _sales(_ensure_features(df))
    non_product_revenue = float(s.loc[s["IsNonProduct"], "Revenue"].sum())
    p = s[~s["IsNonProduct"]]
    if p.empty:
        raise ValueError("No product lines available.")
    desc = p.groupby("StockCode")["Description"].agg(lambda x: x.value_counts().index[0])
    agg = p.groupby("StockCode").agg(revenue=("Revenue", "sum"), quantity=("Quantity", "sum"), orders=("InvoiceNo", "nunique"))
    agg["description"] = desc
    total = float(agg["revenue"].sum())
    def fmt(frame: pd.DataFrame) -> list[dict]:
        out = frame.reset_index()[["StockCode", "description", "revenue", "quantity", "orders"]]
        out["revenue_share_pct"] = 100 * out["revenue"] / total
        return out.round(2).to_dict("records")
    by_rev = agg.sort_values("revenue", ascending=False).head(n)
    by_qty = agg.sort_values("quantity", ascending=False).head(n)
    return jsonable({"by_revenue": fmt(by_rev), "by_quantity": fmt(by_qty), "n_products": int(len(agg)),
                     "top_n_revenue_share_pct": round(100 * float(by_rev["revenue"].sum()) / total, 2),
                     "non_product_lines_revenue": round(non_product_revenue, 2)})

@track_tool("get_top_countries")
def get_top_countries(df: pd.DataFrame, n: int = 10) -> dict:
    """Top-n countries by revenue with orders, customers and revenue share."""
    n = max(1, min(int(n), 50))
    s = _sales(_ensure_features(df))
    agg = s.groupby("Country").agg(revenue=("Revenue", "sum"), orders=("InvoiceNo", "nunique"), customers=("CustomerID", "nunique"))
    total = float(agg["revenue"].sum())
    agg["revenue_share_pct"] = 100 * agg["revenue"] / total
    agg["aov"] = agg["revenue"] / agg["orders"]
    top = agg.sort_values("revenue", ascending=False).head(n).reset_index()
    non_home = agg.drop(index="United Kingdom", errors="ignore").sort_values("revenue", ascending=False)
    return jsonable({"top": top.round(2).to_dict("records"), "n_countries": int(len(agg)),
                     "top_country": str(top.iloc[0]["Country"]), "top_country_share_pct": round(float(top.iloc[0]["revenue_share_pct"]), 2),
                     "non_uk_share_pct": round(float(100 * non_home["revenue"].sum() / total), 2),
                     "top_non_uk": non_home.head(5).reset_index().round(2).to_dict("records")})

@track_tool("analyze_customers")
def analyze_customers(df: pd.DataFrame) -> dict:
    """Customer behaviour: spend distribution, concentration, repeat rate, frequency segments (identified customers only)."""
    s = _sales(_ensure_features(df))
    ident = s[s["CustomerID"].notna()]
    if ident.empty:
        raise ValueError("No sales lines with CustomerID.")
    per = ident.groupby("CustomerID").agg(spend=("Revenue", "sum"), orders=("InvoiceNo", "nunique"), units=("Quantity", "sum"))
    per = per.sort_values("spend", ascending=False)
    k = max(1, int(math.ceil(0.10 * len(per))))
    orders = per["orders"]
    return jsonable({
        "identified_customers": int(len(per)),
        "revenue_share_from_identified_pct": round(100 * float(ident["Revenue"].sum()) / float(s["Revenue"].sum()), 2),
        "revenue_share_without_customer_id_pct": round(100 * float(s.loc[s["CustomerID"].isna(), "Revenue"].sum()) / float(s["Revenue"].sum()), 2),
        "mean_spend": round(float(per["spend"].mean()), 2), "median_spend": round(float(per["spend"].median()), 2),
        "spend_percentiles": {f"p{q}": round(float(per["spend"].quantile(q / 100)), 2) for q in (50, 75, 90, 99)},
        "top10pct_customers_count": k, "top10pct_revenue_share_pct": round(100 * float(per["spend"].iloc[:k].sum()) / float(per["spend"].sum()), 2),
        "repeat_customer_rate_pct": round(100 * float((orders >= 2).mean()), 2),
        "avg_orders_per_customer": round(float(orders.mean()), 2),
        "segments": {"one_time": int((orders == 1).sum()), "occasional_2_5": int(((orders >= 2) & (orders <= 5)).sum()), "loyal_6_plus": int((orders >= 6).sum())},
        "top_customers": per.head(10).reset_index().round(2).to_dict("records"),
    })

@track_tool("detect_anomalies")
def detect_anomalies(df: pd.DataFrame, method: str = "iqr", iqr_k: float = 3.0, z_threshold: float = 4.0) -> dict:
    """Detect **statistical anomalies** (NOT fraud) on product sales lines.

    Line level: Quantity, UnitPrice and line Revenue above an upper threshold
      * method='iqr':    Q3 + iqr_k x IQR  (default k=3 -> extreme outliers only)
      * method='zscore': mean + z_threshold x std
    Day level: robust z-score (median/MAD, Iglewicz-Hoaglin) on daily revenue, |z| > 3.5.
    """
    method = (method or "iqr").lower()
    if method not in {"iqr", "zscore"}:
        raise ValueError("method must be 'iqr' or 'zscore'")
    full = _ensure_features(df)
    p = _sales(full)
    p = p[~p["IsNonProduct"]].copy()
    if p.empty:
        raise ValueError("No product sales lines for anomaly detection.")
    thresholds, flags = {}, pd.DataFrame(index=p.index)
    for col in ("Quantity", "UnitPrice", "Revenue"):
        x = p[col].astype(float)
        thr = float(x.quantile(0.75) + iqr_k * (x.quantile(0.75) - x.quantile(0.25))) if method == "iqr" else float(x.mean() + z_threshold * x.std())
        thresholds[col] = round(thr, 2)
        flags[col] = x > thr
    any_flag = flags.any(axis=1)
    flagged = p[any_flag].copy()
    flagged["reasons"] = flags[any_flag].apply(lambda r: ", ".join(c for c in flags.columns if r[c]), axis=1)
    top = flagged.sort_values("Revenue", ascending=False).head(10)
    examples = [{"InvoiceNo": r.InvoiceNo, "date": str(r.InvoiceDate.date()), "description": str(r.Description)[:45], "quantity": int(r.Quantity),
                 "unit_price": round(float(r.UnitPrice), 2), "revenue": round(float(r.Revenue), 2), "country": r.Country, "reasons": r.reasons}
                for r in top.itertuples()]
    daily = p.groupby("Date")["Revenue"].sum()
    med = float(daily.median()); mad = float((daily - med).abs().median())
    rz = 0.6745 * (daily - med) / mad if mad > 0 else pd.Series(0.0, index=daily.index)
    bad_days = rz[rz.abs() > 3.5].sort_values(key=lambda x: -x.abs())
    returns = _returns(full).sort_values("Revenue").head(5)
    return jsonable({
        "label": "Statistical Anomalies (unusual values, not proof of fraud or error)", "method": method,
        "parameters": {"iqr_k": iqr_k} if method == "iqr" else {"z_threshold": z_threshold},
        "thresholds": thresholds, "lines_analysed": int(len(p)),
        "per_metric_counts": {c: int(flags[c].sum()) for c in flags.columns},
        "flagged_lines": int(any_flag.sum()), "flagged_lines_pct": round(100 * float(any_flag.mean()), 3),
        "flagged_revenue": round(float(flagged["Revenue"].sum()), 2),
        "flagged_revenue_share_pct": round(100 * float(flagged["Revenue"].sum()) / float(p["Revenue"].sum()), 2),
        "top_examples": examples,
        "daily": {"method": "robust z-score (median/MAD), |z|>3.5", "days_analysed": int(len(daily)), "flagged_days": int(len(bad_days)),
                  "examples": [{"date": str(d.date()), "revenue": round(float(daily[d]), 2), "robust_z": round(float(z), 2)} for d, z in bad_days.head(8).items()]},
        "largest_returns": [{"InvoiceNo": r.InvoiceNo, "description": str(r.Description)[:45], "quantity": int(r.Quantity), "value": round(float(r.Revenue), 2)} for r in returns.itertuples()],
    })

@track_tool("analyze_weather_impact")
def analyze_weather_impact(df: pd.DataFrame, weather_df: Optional[pd.DataFrame]) -> dict:
    """Exploratory correlation between daily sales revenue and daily weather (association only, not causation)."""
    if weather_df is None or weather_df.empty:
        return {"available": False, "reason": "No weather data retrieved."}
    daily = _sales(_ensure_features(df)).groupby("Date")["Revenue"].sum().rename("revenue").reset_index().rename(columns={"Date": "date"})
    w = weather_df.copy(); w["date"] = pd.to_datetime(w["date"]).dt.normalize()
    j = daily.merge(w, on="date", how="inner").dropna(subset=["revenue"])
    if len(j) < 20:
        return {"available": False, "reason": f"Only {len(j)} overlapping days (need >= 20)."}
    out: dict[str, Any] = {"available": True, "overlap_days": int(len(j)),
                           "caveat": "Weather location may differ from the customer base (provided coordinates = Berlin; most sales are UK). Correlation is not causation."}
    for col, key in (("temp_mean", "temperature"), ("precip_sum", "precipitation")):
        if col in j and j[col].notna().sum() >= 20 and j[col].nunique() > 1:
            out[f"pearson_{key}"] = round(float(j["revenue"].corr(j[col])), 3)
            out[f"spearman_{key}"] = round(float(j["revenue"].corr(j[col], method="spearman")), 3)
    if "temp_mean" in j and j["temp_mean"].nunique() >= 4:
        try:
            j["band"] = pd.qcut(j["temp_mean"], 4, duplicates="drop")
            g = j.groupby("band", observed=True).agg(avg_daily_revenue=("revenue", "mean"), days=("revenue", "size"), t_min=("temp_mean", "min"), t_max=("temp_mean", "max"))
            out["by_temperature_quartile"] = [{"temp_range_c": f"{r.t_min:.1f} to {r.t_max:.1f}", "avg_daily_revenue": round(float(r.avg_daily_revenue), 2), "days": int(r.days)} for r in g.itertuples()]
        except Exception:
            pass
    return jsonable(out)


# %% Cell 33 (markdown)
# # ### 14.2 LLM-facing tool wrappers (what the Groq model can actually call)
# # Each wrapper has a docstring that becomes the tool description sent to the model. Wrappers catch exceptions and return `{"error": ...}` so one failing tool never crashes the agent loop. Full results are cached in `RESULTS`; the LLM receives a size-limited JSON copy (to respect free-tier token limits).

# %% Cell 34 (code)
from langchain_core.tools import StructuredTool

RESULTS: dict[str, Any] = {}          # tool name -> latest full result (used by agents/nodes)
RESULT_ERRORS: dict[str, str] = {}    # tool name -> latest error message
SEARCH_QUERIES: list[str] = []       # queries actually executed by the report-search tool
ANALYSIS_DF_ERROR = "Clean sales data is not available yet (run retrieval + quality/cleaning first)."

def _run_tool(name: str, fn: Callable, *args, **kwargs) -> Any:
    try:
        res = fn(*args, **kwargs)
        RESULTS[name] = jsonable(res)
        RESULT_ERRORS.pop(name, None)
        return RESULTS[name]
    except Exception as exc:
        RESULT_ERRORS[name] = f"{type(exc).__name__}: {exc}"
        return {"error": RESULT_ERRORS[name]}

def _clean_df() -> pd.DataFrame:
    if STORE.clean_df is None:
        raise RuntimeError(ANALYSIS_DF_ERROR)
    return STORE.clean_df

# ---- retrieval tools ----
def load_sales_data_tool() -> dict:
    """Load the Online Retail sales dataset (Excel) into memory. Call this first. Returns row count, columns and date range."""
    def _do():
        if not STORE.paths.get("sales"):
            raise FileNotFoundError(f"Sales file not found. Expected: {EXPECTED_NAMES['sales']}")
        STORE.raw_df = load_sales_data(STORE.paths["sales"])
        d = pd.to_datetime(STORE.raw_df["InvoiceDate"], errors="coerce")
        return {"rows": len(STORE.raw_df), "columns": list(STORE.raw_df.columns), "date_start": str(d.min().date()), "date_end": str(d.max().date()),
                "file": Path(STORE.paths["sales"]).name}
    return _run_tool("load_sales_data", _do)

def search_annual_report_tool(query: str, top_k: int = 3) -> dict:
    """Search the Target annual report (PDF) for passages relevant to a query, e.g. 'net sales growth' or 'supply chain risk'. Returns snippets with page numbers."""
    res = _run_tool("search_annual_report", search_annual_report, query, int(top_k or 3))
    if isinstance(res, dict) and "error" in res:
        return res
    SEARCH_QUERIES.append(query)
    RESULTS["search_annual_report"] = {"queries_run": list(SEARCH_QUERIES)}
    return {"query": query, "results": [{"page": r["page"], "score": r["score"], "snippet": r["text"][:280]} for r in res]}

def get_weather_data_tool(start_date: str = "", end_date: str = "", latitude: float = WEATHER_LAT, longitude: float = WEATHER_LON, mode: str = "historical") -> dict:
    """Get weather data from the live Open-Meteo API (falls back to a local CSV). mode='historical' needs start_date/end_date (YYYY-MM-DD; leave empty to use the sales date range); mode='current' returns current conditions."""
    return _run_tool("get_weather_data", get_weather_data, start_date or "", end_date or "", latitude, longitude, mode or "historical")

# ---- analysis tools ----
def calculate_kpis_tool() -> dict:
    """Calculate the deterministic business KPIs (revenue, orders, customers, AOV, items per order, return rates)."""
    return _run_tool("calculate_kpis", lambda: calculate_kpis(_clean_df()))
def analyze_sales_trends_tool() -> dict:
    """Analyse monthly revenue trends, growth, seasonality peaks, weekday and hourly profiles."""
    return _run_tool("analyze_sales_trends", lambda: analyze_sales_trends(_clean_df()))
def get_top_products_tool(n: int = 10) -> dict:
    """Get the top-n products by revenue and by units sold."""
    return _run_tool("get_top_products", lambda: get_top_products(_clean_df(), int(n or 10)))
def get_top_countries_tool(n: int = 10) -> dict:
    """Get the top-n countries by revenue with orders, customers and revenue share."""
    return _run_tool("get_top_countries", lambda: get_top_countries(_clean_df(), int(n or 10)))
def analyze_customers_tool() -> dict:
    """Analyse customer behaviour: spend distribution, concentration, repeat purchase rate and segments."""
    return _run_tool("analyze_customers", lambda: analyze_customers(_clean_df()))
def detect_anomalies_tool(method: str = "iqr") -> dict:
    """Detect statistical anomalies in transactions (unusually high quantity/price/revenue) using 'iqr' or 'zscore'."""
    return _run_tool("detect_anomalies", lambda: detect_anomalies(_clean_df(), method or "iqr"))
def analyze_weather_impact_tool() -> dict:
    """Correlate daily revenue with the retrieved weather data (exploratory, association only)."""
    return _run_tool("analyze_weather_impact", lambda: analyze_weather_impact(_clean_df(), STORE.weather_df))

def _mk(fn: Callable, name: str) -> StructuredTool:
    return StructuredTool.from_function(func=fn, name=name, description=(fn.__doc__ or name).strip())

RETRIEVAL_TOOLS = [_mk(load_sales_data_tool, "load_sales_data"), _mk(search_annual_report_tool, "search_annual_report"), _mk(get_weather_data_tool, "get_weather_data")]
ANALYSIS_TOOLS = [_mk(calculate_kpis_tool, "calculate_kpis"), _mk(analyze_sales_trends_tool, "analyze_sales_trends"), _mk(get_top_products_tool, "get_top_products"),
                  _mk(get_top_countries_tool, "get_top_countries"), _mk(analyze_customers_tool, "analyze_customers"), _mk(detect_anomalies_tool, "detect_anomalies"),
                  _mk(analyze_weather_impact_tool, "analyze_weather_impact")]
INSIGHT_TOOLS = [RETRIEVAL_TOOLS[1]]
print("LLM-callable tools:", [t.name for t in RETRIEVAL_TOOLS + ANALYSIS_TOOLS])


# %% Cell 35 (markdown)
# # ### 14.3 Report tools: charts, PDF (ReportLab) and e-mail summary
# # Defined here (before the agents that use them); Sections 20-22 display their outputs.

# %% Cell 36 (code)
@track_tool("generate_charts")
def generate_charts(df: pd.DataFrame, analysis: dict, out_dir: Path, weather_df: Optional[pd.DataFrame] = None) -> dict:
    """Create the Matplotlib charts used in the report. Returns {chart_name: png_path}. A failing chart is logged and skipped."""
    out_dir = Path(out_dir); out_dir.mkdir(parents=True, exist_ok=True)
    plt.rcParams.update({"figure.dpi": 100, "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False, "font.size": 10})
    full = _ensure_features(df); s = _sales(full)
    paths: dict[str, str] = {}
    NAVY, TEAL, ORANGE, GREY, RED = "#1F3A5F", "#2A9D8F", "#E76F51", "#B0B7C3", "#C0392B"

    def save(fig, name: str) -> None:
        p = out_dir / f"{name}.png"
        fig.tight_layout(); fig.savefig(p, dpi=130); plt.close(fig); paths[name] = str(p)

    def attempt(name: str, fn: Callable) -> None:
        try:
            fn()
        except Exception as exc:
            plt.close("all")
            LOG.warning("generate_charts", f"Chart '{name}' skipped: {type(exc).__name__}: {exc}")

    def revenue_trend():
        m = pd.DataFrame(analysis["trends"]["monthly"])
        fig, ax = plt.subplots(figsize=(10, 4.8))
        cols = [GREY if p else NAVY for p in m["is_partial"]]
        ax.bar(m["YearMonth"], m["revenue"], color=cols, edgecolor="white")
        ax.set_ylabel("Revenue (GBP)"); ax.set_xlabel("Month"); ax.set_title("Monthly Revenue Trend (grey = partial month)")
        ax.tick_params(axis="x", rotation=45)
        ax2 = ax.twinx(); ax2.plot(m["YearMonth"], m["orders"], color=ORANGE, marker="o", label="Orders"); ax2.set_ylabel("Orders"); ax2.grid(False)
        ax2.legend(loc="upper left"); save(fig, "revenue_trend")

    def top_products():
        t = pd.DataFrame(analysis["top_products"]["by_revenue"]).iloc[::-1]
        fig, ax = plt.subplots(figsize=(10, 5.2))
        ax.barh([str(d)[:38] for d in t["description"]], t["revenue"], color=TEAL)
        for i, v in enumerate(t["revenue"]): ax.text(v, i, f" {v:,.0f}", va="center", fontsize=8)
        ax.set_xlabel("Revenue (GBP)"); ax.set_title("Top 10 Products by Revenue"); save(fig, "top_products")

    def top_countries():
        c = pd.DataFrame(analysis["top_countries"]["top"])
        fig, axes = plt.subplots(1, 2, figsize=(11, 4.8))
        a = c.iloc[::-1]; axes[0].barh(a["Country"], a["revenue"], color=NAVY); axes[0].set_title("Top 10 Countries by Revenue"); axes[0].set_xlabel("Revenue (GBP)")
        b = c[c["Country"] != "United Kingdom"].iloc[::-1]
        axes[1].barh(b["Country"], b["revenue"], color=TEAL); axes[1].set_title("Top Countries excluding United Kingdom"); axes[1].set_xlabel("Revenue (GBP)")
        save(fig, "top_countries")

    def customer_distribution():
        spend = s[s["CustomerID"].notna()].groupby("CustomerID")["Revenue"].sum().sort_values(ascending=False)
        spend = spend[spend > 0]
        fig, axes = plt.subplots(1, 2, figsize=(11, 4.6))
        axes[0].hist(spend, bins=np.logspace(np.log10(max(spend.min(), 1)), np.log10(spend.max()), 40), color=TEAL, edgecolor="white")
        axes[0].set_xscale("log"); axes[0].axvline(spend.median(), color=ORANGE, ls="--", label=f"Median {spend.median():,.0f}")
        axes[0].set_xlabel("Customer total spend (GBP, log scale)"); axes[0].set_ylabel("Customers"); axes[0].set_title("Customer Revenue Distribution"); axes[0].legend()
        cum = spend.cumsum() / spend.sum(); x = np.arange(1, len(cum) + 1) / len(cum)
        axes[1].plot(x * 100, cum.values * 100, color=NAVY); axes[1].axvline(10, color=GREY, ls=":"); 
        axes[1].set_xlabel("Top x% of customers"); axes[1].set_ylabel("Cumulative % of revenue"); axes[1].set_title("Revenue Concentration (Pareto curve)")
        save(fig, "customer_distribution")

    def order_metrics():
        m = pd.DataFrame(analysis["trends"]["monthly"])
        fig, axes = plt.subplots(1, 2, figsize=(11, 4.4))
        axes[0].plot(m["YearMonth"], m["orders"], marker="o", color=NAVY, label="Orders"); axes[0].plot(m["YearMonth"], m["customers"], marker="s", color=ORANGE, label="Unique customers")
        axes[0].set_title("Orders and Active Customers per Month"); axes[0].tick_params(axis="x", rotation=60); axes[0].legend()
        axes[1].plot(m["YearMonth"], m["aov"], marker="o", color=TEAL); axes[1].set_title("Average Order Value per Month (GBP)"); axes[1].tick_params(axis="x", rotation=60)
        save(fig, "order_metrics")

    def anomalies():
        thr = analysis["anomalies"]["thresholds"]
        p = s[~s["IsNonProduct"]]
        flag = (p["Quantity"] > thr["Quantity"]) | (p["UnitPrice"] > thr["UnitPrice"]) | (p["Revenue"] > thr["Revenue"])
        normal, bad = p[~flag], p[flag]
        normal = normal.sample(min(len(normal), 15000), random_state=SEED)
        fig, axes = plt.subplots(1, 2, figsize=(11.5, 4.8))
        axes[0].scatter(normal["Quantity"], normal["UnitPrice"], s=4, alpha=0.25, color=GREY, label=f"Normal (sample of {len(normal):,})")
        axes[0].scatter(bad["Quantity"], bad["UnitPrice"], s=10, alpha=0.7, color=RED, label=f"Statistical anomalies ({len(bad):,})")
        axes[0].set_xscale("log"); axes[0].set_yscale("log"); axes[0].set_xlabel("Quantity"); axes[0].set_ylabel("Unit price (GBP)")
        axes[0].set_title("Transaction Anomalies"); axes[0].legend(fontsize=8)
        daily = p.groupby("Date")["Revenue"].sum(); med = daily.median(); mad = (daily - med).abs().median()
        rz = 0.6745 * (daily - med) / mad if mad > 0 else daily * 0
        axes[1].plot(daily.index, daily.values, color=NAVY, lw=1, label="Daily revenue")
        hot = daily[rz.abs() > 3.5]; axes[1].scatter(hot.index, hot.values, color=RED, zorder=3, label=f"|robust z|>3.5 ({len(hot)} days)")
        axes[1].set_title("Daily Revenue with Anomalous Days"); axes[1].set_ylabel("Revenue (GBP)"); axes[1].tick_params(axis="x", rotation=30); axes[1].legend(fontsize=8)
        save(fig, "anomalies")

    def weather_relation():
        if weather_df is None or weather_df.empty or "temp_mean" not in weather_df: raise ValueError("no weather data")
        daily = s.groupby("Date")["Revenue"].sum().rename("revenue").reset_index().rename(columns={"Date": "date"})
        w = weather_df.copy(); w["date"] = pd.to_datetime(w["date"]).dt.normalize()
        j = daily.merge(w, on="date").dropna(subset=["temp_mean"])
        if len(j) < 20: raise ValueError("too little overlap")
        r = j["revenue"].corr(j["temp_mean"]); coef = np.polyfit(j["temp_mean"], j["revenue"], 1)
        fig, ax = plt.subplots(figsize=(8, 4.6)); ax.scatter(j["temp_mean"], j["revenue"], s=14, alpha=0.6, color=TEAL)
        xs = np.linspace(j["temp_mean"].min(), j["temp_mean"].max(), 50); ax.plot(xs, np.polyval(coef, xs), color=ORANGE, label=f"Linear fit (Pearson r = {r:.2f}, n = {len(j)} days)")
        ax.set_xlabel("Mean daily temperature (deg C)"); ax.set_ylabel("Daily revenue (GBP)"); ax.set_title("Daily Revenue vs Temperature (exploratory)"); ax.legend(); save(fig, "weather_relation")

    for name, fn in [("revenue_trend", revenue_trend), ("top_products", top_products), ("top_countries", top_countries), ("customer_distribution", customer_distribution),
                     ("order_metrics", order_metrics), ("anomalies", anomalies), ("weather_relation", weather_relation)]:
        attempt(name, fn)
    return paths


# %% Cell 37 (code)
from reportlab.lib.pagesizes import A4
from reportlab.lib import colors
from reportlab.lib.units import cm
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle, Image as RLImage, PageBreak, KeepTogether
from reportlab.lib.utils import ImageReader
from xml.sax.saxutils import escape as xml_escape

_PDF_REPL = {"\u2013": "-", "\u2014": "-", "\u2018": "'", "\u2019": "'", "\u201c": '"', "\u201d": '"', "\u2022": "-", "\u2192": "->", "\u2265": ">=",
             "\u2264": "<=", "\u00a0": " ", "\u2026": "...", "\u2248": "~", "\u00d7": "x", "\u2212": "-", "\u202f": " "}

def pdf_safe(text: Any) -> str:
    """Make text safe for ReportLab's built-in Helvetica (latin-1): replace common unicode, drop the rest."""
    t = str(text)
    for k, v in _PDF_REPL.items():
        t = t.replace(k, v)
    return t.encode("latin-1", "replace").decode("latin-1")

def gbp(x: Any) -> str:
    return f"GBP {float(x):,.0f}"

TAG_COLORS = {"FACT": "#1B7F3B", "CALCULATED RESULT": "#1F3A5F", "LLM INTERPRETATION": "#B9770E", "RECOMMENDATION": "#8E2A6B"}

@track_tool("generate_pdf_report")
def generate_pdf_report(rd: dict, chart_paths: dict, output_path: str | Path) -> Path:
    """Render the 15-section business report with ReportLab and return the saved path. Raises if the PDF cannot be built.

    `rd` holds: generated_at, user_request, sources, data_quality, cleaning_report, analysis, evidence, ar_summary,
    weather, insights, recommendations, exec_summary, limitations, llm_meta.
    """
    output_path = Path(output_path); output_path.parent.mkdir(parents=True, exist_ok=True)
    ss = getSampleStyleSheet()
    st = {"title": ParagraphStyle("t", parent=ss["Title"], fontSize=22, textColor=colors.HexColor("#1F3A5F"), spaceAfter=6),
          "sub": ParagraphStyle("s", parent=ss["Normal"], fontSize=10, textColor=colors.HexColor("#555555"), spaceAfter=4),
          "h1": ParagraphStyle("h1", parent=ss["Heading1"], fontSize=15, textColor=colors.HexColor("#1F3A5F"), spaceBefore=14, spaceAfter=6),
          "h2": ParagraphStyle("h2", parent=ss["Heading2"], fontSize=11.5, textColor=colors.HexColor("#2A9D8F"), spaceBefore=8, spaceAfter=3),
          "p": ParagraphStyle("p", parent=ss["Normal"], fontSize=9.5, leading=13, spaceAfter=5),
          "b": ParagraphStyle("b", parent=ss["Normal"], fontSize=9.5, leading=13, leftIndent=14, bulletIndent=3, spaceAfter=3),
          "cell": ParagraphStyle("c", parent=ss["Normal"], fontSize=8, leading=10), "cellh": ParagraphStyle("ch", parent=ss["Normal"], fontSize=8, leading=10, textColor=colors.white, fontName="Helvetica-Bold"),
          "note": ParagraphStyle("n", parent=ss["Normal"], fontSize=8, leading=10, textColor=colors.HexColor("#666666"), spaceAfter=6)}
    W = 17 * cm
    story: list = []
    def P(t, style="p"): story.append(Paragraph(xml_escape(pdf_safe(t)), st[style]))
    def PM(markup, style="p"): story.append(Paragraph(markup, st[style]))
    def bullet(t): story.append(Paragraph(xml_escape(pdf_safe(t)), st["b"], bulletText="-"))
    def table(rows, widths=None, header=True):
        data = [[Paragraph(xml_escape(pdf_safe(c)), st["cellh" if (header and i == 0) else "cell"]) for c in r] for i, r in enumerate(rows)]
        cw = [W * w for w in widths] if widths else None
        t = Table(data, colWidths=cw, repeatRows=1 if header else 0)
        style = [("GRID", (0, 0), (-1, -1), 0.25, colors.HexColor("#B0B7C3")), ("VALIGN", (0, 0), (-1, -1), "TOP"),
                 ("ROWBACKGROUNDS", (0, 1 if header else 0), (-1, -1), [colors.white, colors.HexColor("#F2F5F9")]),
                 ("LEFTPADDING", (0, 0), (-1, -1), 4), ("RIGHTPADDING", (0, 0), (-1, -1), 4)]
        if header: style.append(("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#1F3A5F")))
        t.setStyle(TableStyle(style)); story.append(t); story.append(Spacer(1, 6))
    def chart(name, caption, width=W):
        p = chart_paths.get(name)
        if not p or not Path(p).exists():
            P(f"(Chart '{name}' unavailable.)", "note"); return
        iw, ih = ImageReader(p).getSize()
        story.append(KeepTogether([RLImage(p, width=width, height=width * ih / iw), Paragraph(xml_escape(pdf_safe(caption)), st["note"])]))
    def section(title, fn):
        story.append(Paragraph(xml_escape(pdf_safe(title)), st["h1"]))
        try:
            fn()
        except Exception as exc:
            LOG.error("generate_pdf_report", f"Section '{title}' failed: {type(exc).__name__}: {exc}")
            P(f"This section could not be generated: {type(exc).__name__}: {exc}", "note")

    A = rd.get("analysis", {}); K = A.get("kpis", {}); TR = A.get("trends", {}); TP = A.get("top_products", {}); TC = A.get("top_countries", {})
    CU = A.get("customers", {}); AN = A.get("anomalies", {}); WI = A.get("weather_impact", {}); DQ = rd.get("data_quality", {}); CR = rd.get("cleaning_report", {})
    LM = rd.get("llm_meta", {}); ES = rd.get("exec_summary", {})

    P("AI Business Analyst - Multi-Agent Business Report", "title")
    P(f"Generated {rd.get('generated_at', '')} by a LangGraph multi-agent system. LLM mode: "
      f"{('Groq ' + str(LM.get('model'))) if LM.get('available') else 'deterministic (no LLM available)'}.", "sub")
    P("Request: " + " ".join(str(rd.get("user_request", "")).split()), "sub")
    story.append(Spacer(1, 6))

    def s1():
        P(ES.get("executive_summary", "No executive summary available."))
        P("Executive summary authored by: " + ("LLM (from the calculated results only)" if ES.get("mode") == "llm" else "deterministic template (numbers from Pandas)"), "note")
        if K:
            table([["Total revenue", "Orders", "Customers", "Avg. order value"], [gbp(K["total_revenue"]), f"{K['total_orders']:,}", f"{K['unique_customers']:,}", f"GBP {K['average_order_value']:,.2f}"]])
    section("1. Executive Summary", s1)

    def s2():
        table([["Source", "Status", "Details"]] + [[x["name"], x["status"], x["detail"]] for x in rd.get("sources", [])], [0.3, 0.14, 0.56])
        P("Currency is assumed to be GBP (the Online Retail dataset is from a UK-based retailer; the file itself carries no currency field).", "note")
    section("2. Data Sources", s2)

    def s3():
        if not DQ: P("Data-quality report unavailable."); return
        table([["Check", "Raw data result"], ["Rows / columns", f"{DQ.get('rows', 0):,} / {DQ.get('columns', 0)}"], ["Duplicate rows", f"{DQ.get('duplicate_rows', 0):,}"],
               ["Invalid or zero quantity rows", f"{DQ.get('invalid_quantity_rows', 0):,}"], ["Negative quantity rows (returns kept)", f"{DQ.get('negative_quantity_rows', 0):,}"],
               ["Invalid/non-positive price rows", f"{DQ.get('invalid_price_rows', 0):,}"], ["Invalid dates", f"{DQ.get('invalid_date_rows', 0):,}"],
               ["Missing CustomerID rows", f"{DQ.get('missing_customer_id_rows', 0):,}"], ["Empty descriptions", f"{DQ.get('empty_description_rows', 0):,}"],
               ["Cancellation rows (invoice starts with C)", f"{DQ.get('cancellation_rows', 0):,}"], ["Defect rows / quality score", f"{DQ.get('defect_rows', 0):,} / {DQ.get('quality_score_pct', 0)}%"]], [0.6, 0.4])
        if CR:
            P(f"Cleaning: {CR.get('rows_before', 0):,} rows -> {CR.get('rows_after', 0):,} rows. Dropped: duplicates {CR.get('duplicates_dropped', 0):,}, "
              f"zero/negative prices {CR.get('zero_price_dropped', 0) + CR.get('negative_price_dropped', 0):,}, unparsable {CR.get('unparsable_rows_dropped', 0):,}, zero quantity {CR.get('zero_quantity_dropped', 0):,}. "
              f"Empty descriptions filled: {CR.get('empty_descriptions_filled', 0):,}.")
            P(CR.get("decision_note", ""), "note")
        ol = DQ.get("outliers", {})
        if ol: P("Outliers (reported, not removed): " + ol.get("method", "") + f". Lines above the fence - quantity {ol['quantity']['count']:,}, unit price {ol['unit_price']['count']:,}, line revenue {ol['line_revenue']['count']:,}.", "note")
    section("3. Data Quality", s3)

    def s4():
        table([["KPI", "Value", "Definition"],
               ["Total revenue (gross sales)", gbp(K["total_revenue"]), "Sum of Quantity x UnitPrice over sales lines (returns excluded)"],
               ["Returns value", gbp(K["returns_value"]), "Absolute value of cancellation/return lines"], ["Net revenue", gbp(K["net_revenue"]), "Total revenue - returns value"],
               ["Total orders", f"{K['total_orders']:,}", "Distinct sales invoices"], ["Unique customers", f"{K['unique_customers']:,}", "Distinct non-missing CustomerID in sales"],
               ["Unique products", f"{K['unique_products']:,}", "Distinct product StockCodes (fee/service codes excluded)"], ["Average order value", f"GBP {K['average_order_value']:,.2f}", "Total revenue / total orders"],
               ["Avg. items per order", f"{K['average_items_per_order']:,.2f}", "Units sold / total orders"], ["Avg. revenue per customer", f"GBP {K['average_revenue_per_customer']:,.2f}", "Identified-customer revenue / unique customers"],
               ["Return rate (invoices)", f"{K['return_rate_invoices_pct']}%", "Cancellation invoices / (sales + cancellation invoices)"], ["Return rate (value)", f"{K['return_rate_value_pct']}%", "Returns value / total revenue"],
               ["Period", f"{K['date_start']} to {K['date_end']}", "First and last sales date"]], [0.27, 0.23, 0.5])
    section("4. Key KPIs", s4)

    def s5():
        chart("revenue_trend", "Figure 1 - Monthly revenue (bars) and orders (line). Grey bars are partial months.")
        pm = TR.get("peak_month"); lo = TR.get("lowest_month")
        if pm: P(f"Peak complete month: {pm['month']} ({gbp(pm['revenue'])}); lowest complete month: {lo['month']} ({gbp(lo['revenue'])}). Trend across complete months: {TR.get('trend_direction', 'n/a')}.")
        if TR.get("partial_months"): P("Partial months (excluded from growth and trend statistics): " + ", ".join(TR["partial_months"]) + ".", "note")
        wd = TR.get("revenue_by_weekday", {}); 
        if wd: P("Revenue by weekday: " + "; ".join(f"{k} {gbp(v)}" for k, v in wd.items()) + f". Peak hour of day: {TR.get('peak_hour')}:00.")
        chart("order_metrics", "Figure 2 - Orders, active customers and average order value per month.")
    section("5. Sales Analysis", s5)

    def s6():
        chart("top_products", "Figure 3 - Top 10 products by revenue (fee/service lines excluded).")
        table([["StockCode", "Description", "Revenue", "Units", "Share %"]] + [[r["StockCode"], r["description"], f"{r['revenue']:,.0f}", f"{r['quantity']:,}", f"{r['revenue_share_pct']}"] for r in TP["by_revenue"]], [0.14, 0.44, 0.17, 0.13, 0.12])
        P(f"The top {len(TP['by_revenue'])} products account for {TP['top_n_revenue_share_pct']}% of product revenue across {TP['n_products']:,} products. Fee/service lines (postage, manual adjustments, etc.) contributed {gbp(TP['non_product_lines_revenue'])} and are excluded from rankings.")
        P("Top 5 by units sold: " + "; ".join(f"{r['description']} ({r['quantity']:,})" for r in TP["by_quantity"][:5]) + ".")
    section("6. Product Analysis", s6)

    def s7():
        chart("customer_distribution", "Figure 4 - Distribution of customer spend and revenue concentration.")
        table([["Metric", "Value"], ["Identified customers", f"{CU['identified_customers']:,}"], ["Revenue from identified customers", f"{CU['revenue_share_from_identified_pct']}%"],
               ["Revenue without CustomerID", f"{CU['revenue_share_without_customer_id_pct']}%"], ["Mean / median spend", f"GBP {CU['mean_spend']:,.0f} / GBP {CU['median_spend']:,.0f}"],
               [f"Top 10% of customers ({CU['top10pct_customers_count']:,})", f"{CU['top10pct_revenue_share_pct']}% of identified revenue"], ["Repeat-customer rate (>=2 orders)", f"{CU['repeat_customer_rate_pct']}%"],
               ["Average orders per customer", f"{CU['avg_orders_per_customer']}"], ["Segments (1 / 2-5 / 6+ orders)", f"{CU['segments']['one_time']:,} / {CU['segments']['occasional_2_5']:,} / {CU['segments']['loyal_6_plus']:,}"]], [0.55, 0.45])
    section("7. Customer Analysis", s7)

    def s8():
        chart("top_countries", "Figure 5 - Revenue by country (right: excluding the United Kingdom).")
        table([["Country", "Revenue", "Share %", "Orders", "Customers", "AOV"]] + [[r["Country"], f"{r['revenue']:,.0f}", f"{r['revenue_share_pct']}", f"{r['orders']:,}", f"{r['customers']:,}", f"{r['aov']:,.0f}"] for r in TC["top"]], [0.26, 0.16, 0.12, 0.14, 0.16, 0.16])
        P(f"{TC['top_country']} generates {TC['top_country_share_pct']}% of revenue across {TC['n_countries']} countries; non-UK markets together contribute {TC['non_uk_share_pct']}%.")
    section("8. Country Analysis", s8)

    def s9():
        P(AN.get("label", "Statistical Anomalies") + ". Method: " + AN.get("method", "") + f" with parameters {AN.get('parameters', {})}. Thresholds: " + ", ".join(f"{k} > {v:,}" for k, v in AN.get("thresholds", {}).items()) + ".")
        P(f"{AN.get('flagged_lines', 0):,} of {AN.get('lines_analysed', 0):,} product lines ({AN.get('flagged_lines_pct', 0)}%) are flagged, representing {gbp(AN.get('flagged_revenue', 0))} ({AN.get('flagged_revenue_share_pct', 0)}% of product revenue). "
          "Flagged lines are unusual values (e.g. wholesale-size orders), not proof of fraud or error.")
        chart("anomalies", "Figure 6 - Left: quantity vs unit price (log-log). Right: daily revenue with anomalous days.")
        if AN.get("top_examples"):
            table([["Invoice", "Date", "Description", "Qty", "Price", "Revenue", "Country"]] + [[r["InvoiceNo"], r["date"], r["description"], f"{r['quantity']:,}", f"{r['unit_price']}", f"{r['revenue']:,.0f}", r["country"]] for r in AN["top_examples"]], [0.11, 0.12, 0.29, 0.08, 0.08, 0.12, 0.2])
        d = AN.get("daily", {}); P(f"Daily level ({d.get('method', '')}): {d.get('flagged_days', 0)} of {d.get('days_analysed', 0)} days flagged.")
        if AN.get("largest_returns"): P("Largest return/cancellation lines: " + "; ".join(f"{r['InvoiceNo']} {r['description']} ({gbp(abs(r['value']))})" for r in AN["largest_returns"][:3]) + ".", "note")
    section("9. Anomaly Analysis", s9)

    def s10():
        ar = rd.get("ar_summary", {})
        PM(f"<b>Summary ({xml_escape(ar.get('mode', 'n/a'))}):</b>")
        for line in ar.get("bullets", []) or ["No annual-report evidence was retrieved."]: bullet(line)
        ev = rd.get("evidence", [])[:6]
        if ev:
            P("Retrieved evidence (TF-IDF cosine similarity; citations are physical PDF pages):", "h2")
            table([["Citation", "Score", "Excerpt"]] + [[e["citation"], f"{e['score']:.3f}", e["text"][:320] + "..."] for e in ev], [0.2, 0.08, 0.72])
    section("10. Annual Report Insights", s10)

    def s11():
        wm = rd.get("weather", {})
        if not wm or "error" in wm: P("Weather data was not available. " + str(wm.get("error", ""))); return
        P(f"Source: {wm.get('source')} ({wm.get('date_start')} to {wm.get('date_end')}, {wm.get('rows')} days, lat {wm.get('latitude')}, lon {wm.get('longitude')}). API fallback used: {wm.get('fallback_used')}.")
        if wm.get("api_error"): P("API error message: " + str(wm["api_error"]), "note")
        if WI.get("available"):
            P(f"Over {WI['overlap_days']} overlapping days, Pearson r (revenue vs mean temperature) = {WI.get('pearson_temperature')}, Spearman = {WI.get('spearman_temperature')}; "
              f"vs precipitation: Pearson = {WI.get('pearson_precipitation')}. {WI['caveat']}")
            if WI.get("by_temperature_quartile"): table([["Temperature band (C)", "Avg daily revenue", "Days"]] + [[r["temp_range_c"], f"{r['avg_daily_revenue']:,.0f}", str(r["days"])] for r in WI["by_temperature_quartile"]], [0.4, 0.35, 0.25])
            chart("weather_relation", "Figure 7 - Daily revenue vs mean temperature (exploratory).", width=13 * cm)
        else: P("Weather/sales correlation not available: " + str(WI.get("reason", "")))
    section("11. Weather/API Insights", s11)

    def s12():
        P("Each statement is labelled: FACT (quoted from the annual report, with page), CALCULATED RESULT (computed by Pandas), LLM INTERPRETATION (hypothesis by the model), RECOMMENDATION.", "note")
        for ins in rd.get("insights", []):
            col = TAG_COLORS.get(ins["category"], "#000000")
            ev = ("  <font size=7 color='#666666'>[" + xml_escape(pdf_safe("; ".join(map(str, ins.get("evidence", []))[:200] if isinstance(ins.get("evidence"), str) else [str(x) for x in ins.get("evidence", [])]))) + "]</font>") if ins.get("evidence") else ""
            story.append(Paragraph(f"<font color='{col}'><b>[{ins['category']}]</b></font> {xml_escape(pdf_safe(ins['text']))}{ev}", st["b"], bulletText="-"))
    section("12. AI-generated Business Insights", s12)

    def s13():
        for i, rec in enumerate(rd.get("recommendations", []), 1):
            tag = "LLM" if rec.get("source") == "llm" else "rule-based"
            story.append(Paragraph(f"<b>{i}.</b> {xml_escape(pdf_safe(rec['text']))} <font size=7 color='#666666'>[{tag}]</font>", st["p"]))
        if not rd.get("recommendations"): P("No recommendations generated.")
    section("13. Recommendations", s13)

    def s14():
        for l in rd.get("limitations", []): bullet(l)
    section("14. Limitations", s14)

    def s15():
        P(ES.get("conclusion", "Analysis complete. See sections above for details."))
    section("15. Conclusion", s15)

    def footer(canvas, doc):
        canvas.saveState(); canvas.setFont("Helvetica", 8); canvas.setFillColor(colors.HexColor("#666666"))
        canvas.drawString(2 * cm, 1.1 * cm, "Multi-Agent AI Business Analyst Report"); canvas.drawRightString(A4[0] - 2 * cm, 1.1 * cm, f"Page {doc.page}"); canvas.restoreState()
    doc = SimpleDocTemplate(str(output_path), pagesize=A4, leftMargin=2 * cm, rightMargin=2 * cm, topMargin=1.8 * cm, bottomMargin=1.8 * cm,
                            title="Multi-Agent Business Report", author="AI Business Analyst Multi-Agent System")
    doc.build(story, onFirstPage=footer, onLaterPages=footer)
    if not output_path.exists() or output_path.stat().st_size == 0:
        raise RuntimeError("PDF build finished but the file is missing or empty.")
    return output_path


import datetime as dt
from email.message import EmailMessage
import os
from pathlib import Path
import smtplib

# المتغيرات الافتراضية للتشغيل المحلي (تكون في أول الخلية أو في ملف Config)
ENABLE_EMAIL_SENDING = os.getenv("ENABLE_EMAIL_SENDING", "False").lower() in (
    "true",
    "1",
)
REPORT_RECIPIENT = os.getenv("REPORT_RECIPIENT", "client@example.com")


@track_tool("generate_email_summary")
def generate_email_summary(
    rd: dict,
    report_path: str | Path,
    recipient: str = REPORT_RECIPIENT,
) -> dict:
  """Create an e-mail-ready summary (subject, recipient placeholder, body, attachment path).

  Does NOT send anything.
  """
  analysis = rd.get("analysis", {})
  kpis = analysis.get("kpis", {})
  top_products = analysis.get("top_products", {}).get("by_revenue", [{}])
  top_countries = analysis.get("top_countries", {})
  recs = rd.get("recommendations", [])

  lines = [
      "Hello,",
      "",
      (
          "The automated multi-agent business analysis has finished. The full"
          " PDF report is attached."
      ),
      "",
  ]

  if kpis:
    top_prod_desc = top_products[0].get("description", "n/a") if top_products else "n/a"
    top_country = top_countries.get("top_country", "n/a")
    top_country_share = top_countries.get("top_country_share_pct", "n/a")

    lines += [
        "KEY FIGURES (calculated with Pandas)",
        f"- Total revenue: GBP {kpis.get('total_revenue', 0):,.0f} ({kpis.get('date_start', '')} to {kpis.get('date_end', '')})",
        (
            f"- Orders: {kpis.get('total_orders', 0):,}; customers:"
            f" {kpis.get('unique_customers', 0):,}; average order value: GBP"
            f" {kpis.get('average_order_value', 0):,.2f}"
        ),
        (
            f"- Top product: {top_prod_desc}; top country: {top_country}"
            f" ({top_country_share}% of revenue)"
        ),
        f"- Return rate (value): {kpis.get('return_rate_value_pct', 0)}%",
        "",
    ]

  lines += ["TOP RECOMMENDATIONS"] + [
      f"{i}. {r.get('text', '')}" for i, r in enumerate(recs[:3], 1)
  ]

  llm = rd.get("llm_meta", {})
  analysis_mode = (
      f"LLM-assisted ({llm.get('model')})"
      if llm.get("available")
      else "deterministic (no LLM available)"
  )

  lines += [
      "",
      f"Analysis mode: {analysis_mode}.",
      "Statistical anomalies are unusual values, not proof of fraud.",
      "",
      "Regards,",
      "AI Business Analyst (automated)",
  ]

  return {
      "subject": (
          f"Business Analysis Report - {kpis.get('date_start', '')} to"
          f" {kpis.get('date_end', '')}"
      ).strip(),
      "to": recipient,
      "body": "\n".join(lines),
      "attachment_path": str(report_path),
      "generated_at": dt.datetime.now().isoformat(timespec="seconds"),
  }


def save_email_summary(email: dict, path: Path) -> Path:
  """Write the e-mail summary as a readable text file."""
  summary_text = (
      f"To: {email['to']}\n"
      f"Subject: {email['subject']}\n"
      f"Attachment: {email['attachment_path']}\n\n"
      f"{email['body']}\n"
  )
  out_path = Path(path)
  out_path.write_text(summary_text, encoding="utf-8")
  return out_path


def send_email_if_configured(email: dict) -> dict:
  """OPTIONAL real sending via SMTP. Disabled unless ENABLE_EMAIL_SENDING=True AND SMTP_* secrets exist."""
  if not ENABLE_EMAIL_SENDING:
    return {
        "sent": False,
        "reason": (
            "ENABLE_EMAIL_SENDING is False (default). Email summary was"
            " generated only."
        ),
    }

  def secret(name: str) -> str:
    try:
      from kaggle_secrets import UserSecretsClient  # type: ignore

      return UserSecretsClient().get_secret(name)
    except Exception:
      return os.environ.get(name, "")

  host = secret("SMTP_HOST")
  port = secret("SMTP_PORT") or "587"
  user = secret("SMTP_USER")
  pwd = secret("SMTP_PASSWORD")

  if not (host and user and pwd) or "example.com" in email["to"]:
    return {
        "sent": False,
        "reason": (
            "SMTP_* secrets missing or recipient is still the placeholder."
        ),
    }

  try:
    msg = EmailMessage()
    msg["Subject"] = email["subject"]
    msg["From"] = user
    msg["To"] = email["to"]
    msg.set_content(email["body"])

    attachment_path = Path(email["attachment_path"])
    if attachment_path.exists():
      msg.add_attachment(
          attachment_path.read_bytes(),
          maintype="application",
          subtype="pdf",
          filename=attachment_path.name,
      )

    with smtplib.SMTP(host, int(port), timeout=30) as smtp:
      smtp.starttls()
      smtp.login(user, pwd)
      smtp.send_message(msg)

    return {"sent": True, "reason": "sent via SMTP"}
  except Exception as exc:
    return {"sent": False, "reason": f"{type(exc).__name__}: {exc}"}


# %% Cell 39 (markdown)
# # # 15. Agent State
# #
# # The LangGraph state is a `TypedDict`. Fields marked *(reducer)* are **appended/merged** by LangGraph when several nodes write to them (e.g. `execution_log`); all other fields are overwritten by the node that owns them. The state contains only small, JSON-serialisable summaries - DataFrames stay in `STORE`.
# #
# # | Field | Meaning |
# # |---|---|
# # | `user_request`, `request_valid` | the validated business request |
# # | `data_summary` | availability/shape of the three sources (sales, annual report, weather) |
# # | `data_quality`, `cleaning_report`, `quality_attempts` | latest quality check, what cleaning did, loop counter |
# # | `analysis_results` | KPIs, trends, top products/countries, customers, anomalies, weather impact |
# # | `annual_report_evidence`, `annual_report_summary` | retrieved chunks (with pages) and their summary |
# # | `weather_results` | weather tool result (source, fallback flag, summary) |
# # | `insights`, `recommendations` | labelled statements (FACT / CALCULATED RESULT / LLM INTERPRETATION / RECOMMENDATION) |
# # | `exec_summary`, `chart_paths`, `report_path` | report content and output paths |
# # | `email_summary`, `artifacts` | automated action results |
# # | `llm_meta`, `status` | LLM usage/grounding information and final workflow status |
# # | `execution_log` *(reducer)*, `errors` *(reducer)*, `agents_executed` *(reducer)* | structured log, errors/warnings, executed agents |
# # | `agent_timings` *(reducer: sum)*, `agent_traces` *(reducer: merge)* | seconds per node; per-agent tool-calling trace |

# %% Cell 40 (code)
def _merge_sum(a: dict, b: dict) -> dict:
    out = dict(a or {})
    for k, v in (b or {}).items():
        out[k] = round(out.get(k, 0) + v, 3)
    return out

def _merge_dict(a: dict, b: dict) -> dict:
    return {**(a or {}), **(b or {})}

class AgentState(TypedDict, total=False):
    user_request: str
    request_valid: bool
    data_summary: dict
    data_quality: dict
    cleaning_report: dict
    quality_attempts: int
    analysis_results: dict
    annual_report_evidence: list
    annual_report_summary: dict
    weather_results: dict
    insights: list
    recommendations: list
    exec_summary: dict
    chart_paths: dict
    report_path: str
    email_summary: dict
    artifacts: dict
    llm_meta: dict
    status: str
    execution_log: Annotated[list, operator.add]
    errors: Annotated[list, operator.add]
    agents_executed: Annotated[list, operator.add]
    agent_timings: Annotated[dict, _merge_sum]
    agent_traces: Annotated[dict, _merge_dict]

def agent_node(name: str, is_agent: bool = True) -> Callable:
    """Wrap a node: logs start/end, measures time, converts exceptions into state errors (the graph never crashes)."""
    def decorator(fn: Callable) -> Callable:
        @functools.wraps(fn)
        def wrapper(state: dict) -> dict:
            t0 = time.perf_counter()
            LOG.info(name, "started")
            try:
                update = fn(state) or {}
            except Exception as exc:
                LOG.error(name, f"{type(exc).__name__}: {exc}")
                update = {"errors": [{"node": name, "error": f"{type(exc).__name__}: {exc}", "severity": "error"}]}
            dur = time.perf_counter() - t0
            update.setdefault("errors", [])
            update["agent_timings"] = {name: round(dur, 3)}
            if is_agent:
                update["agents_executed"] = [name]
            LOG.info(name, f"completed in {dur:.2f}s")
            update["execution_log"] = LOG.flush()
            return update
        return wrapper
    return decorator


# %% Cell 41 (markdown)
# # # 16. Multi-Agent Definitions
# #
# # | Agent | Responsibility | LLM role | Deterministic guarantee |
# # |---|---|---|---|
# # | **Data Retrieval Agent** | load Excel, search annual report, fetch weather, validate availability | *decides which retrieval tools to call* (Groq tool calling) | required tools that the LLM skipped/failed are run by Python |
# # | *Data Quality check / Cleaning* (graph nodes) | validate, clean, re-validate | none | pure Pandas |
# # | **Data Analysis Agent** | KPIs, trends, top products/countries, customers, anomalies, weather impact | *decides which analysis tools to call* | missing analyses are run by Python; **all arithmetic is Pandas** |
# # | **Research / Insight Agent** | retrieve annual-report evidence, combine sources, interpret, recommend | chooses search queries; summarises evidence; writes interpretations and recommendations | rule-based calculated results and recommendations always exist |
# # | **Report Agent** | charts, executive summary, 15-section PDF | writes executive summary/conclusion (number-checked) | template fallback text |
# # | **Action Agent** | verify and save PDF, save JSON, prepare e-mail summary (+optional SMTP) | none | - |
# #
# # **Tool-calling loop** (`ToolAgent.run`): `LLM -> tool_call -> Python tool -> tool result -> LLM -> ... -> final answer`. The trace of which tools the *LLM* called vs which were run as *deterministic fallback* is stored in the state and used in the evaluation.

# %% Cell 42 (code)
@dataclass
class ToolAgent:
    """An agent = system prompt + tools + required deterministic runners (guarantee of completeness)."""
    name: str
    system_prompt: str
    tools: list
    required: dict                      # tool name -> zero-arg callable that runs the tool deterministically
    result_chars: int = 900             # max characters of each tool result shown to the LLM (free-tier token limits)
    max_iterations: int = 4

    def run(self, task: str) -> dict:
        trace: dict[str, Any] = {"agent": self.name, "llm_used": False, "llm_model": LLM_STATUS.get("model"), "llm_tool_calls": [], "llm_tool_errors": [],
                                 "deterministic_fallback": [], "llm_error": None, "llm_final_text": "", "tool_errors": []}
        ok_called: set[str] = set()
        if LLM is not None:
            trace["llm_used"] = True
            try:
                from langchain_core.messages import SystemMessage, HumanMessage, ToolMessage
                tmap = {t.name: t for t in self.tools}
                bound = LLM.bind_tools(self.tools)
                messages: list = [SystemMessage(content=self.system_prompt), HumanMessage(content=task)]
                for _ in range(self.max_iterations):
                    ai = llm_invoke(bound, messages)
                    messages.append(ai)
                    calls = getattr(ai, "tool_calls", None) or []
                    if not calls:
                        trace["llm_final_text"] = str(ai.content)[:500]
                        break
                    for call in calls:
                        tname, args, cid = call["name"], (call.get("args") or {}), call.get("id")
                        if tname not in tmap:
                            result: Any = {"error": f"unknown tool '{tname}'"}
                        else:
                            try:
                                result = tmap[tname].invoke(args)
                            except Exception as exc:
                                result = {"error": f"{type(exc).__name__}: {exc}"}
                        failed = isinstance(result, dict) and "error" in result
                        (trace["llm_tool_errors"] if failed else trace["llm_tool_calls"]).append({"tool": tname, "args": jsonable(args), **({"error": result["error"][:200]} if failed else {})})
                        if not failed:
                            ok_called.add(tname)
                        LOG.info(self.name, f"LLM tool call: {tname}({json.dumps(jsonable(args))[:120]}) -> {'ERROR' if failed else 'ok'}")
                        messages.append(ToolMessage(content=json.dumps(result, default=str)[:self.result_chars], tool_call_id=cid))
            except Exception as exc:
                trace["llm_error"] = f"{type(exc).__name__}: {str(exc)[:300]}"
                LOG.warning(self.name, f"LLM tool loop failed ({trace['llm_error']}); deterministic fallback will complete the tools.")
        for tname, runner in self.required.items():
            if tname not in ok_called:
                LOG.info(self.name, f"Deterministic run of tool '{tname}'" + ("" if LLM is None else " (LLM did not successfully call it)"))
                runner()
                trace["deterministic_fallback"].append(tname)
        trace["tool_errors"] = [{"tool": t, "error": RESULT_ERRORS[t]} for t in self.required if t in RESULT_ERRORS]
        return trace

DEFAULT_RETRIEVAL_QUERIES = ["net sales revenue growth comparable sales", "inventory supply chain shipping costs"]
INSIGHT_QUERIES = ["net sales revenue growth comparable sales trends", "guests customers loyalty membership Target Circle", "seasonality holiday weather impact on sales",
                   "inventory supply chain risks costs", "digital e-commerce same-day fulfillment services"]

def _run_searches(queries: list[str]) -> dict:
    last: dict = {}
    for q in queries:
        last = search_annual_report_tool(q, 3)
        if isinstance(last, dict) and "error" in last:
            return last
    return last

def build_retrieval_agent() -> ToolAgent:
    return ToolAgent(
        name="Data Retrieval Agent", tools=RETRIEVAL_TOOLS, result_chars=1000,
        system_prompt=("You are the Data Retrieval Agent of a business-analysis system. Use the tools to gather data: "
                       "(1) load_sales_data, (2) search_annual_report with 1-2 focused queries relevant to the request, "
                       "(3) get_weather_data in 'historical' mode (leave dates empty to use the sales period). "
                       "Call each needed tool, then reply with the single word DONE. Never invent data."),
        required={"load_sales_data": load_sales_data_tool,
                  "search_annual_report": lambda: _run_searches(DEFAULT_RETRIEVAL_QUERIES),
                  "get_weather_data": get_weather_data_tool})

def build_analysis_agent() -> ToolAgent:
    return ToolAgent(
        name="Data Analysis Agent", tools=ANALYSIS_TOOLS, result_chars=500,
        system_prompt=("You are the Data Analysis Agent. You do NOT compute numbers yourself; you call deterministic Python tools. "
                       "For a full business analysis call ALL of: calculate_kpis, analyze_sales_trends, get_top_products, get_top_countries, "
                       "analyze_customers, detect_anomalies, analyze_weather_impact (each once). Then reply with the single word DONE."),
        required={"calculate_kpis": calculate_kpis_tool, "analyze_sales_trends": analyze_sales_trends_tool, "get_top_products": get_top_products_tool,
                  "get_top_countries": get_top_countries_tool, "analyze_customers": analyze_customers_tool, "detect_anomalies": detect_anomalies_tool,
                  "analyze_weather_impact": analyze_weather_impact_tool})


# %% Cell 43 (code)
# ---------- Research / Insight Agent helpers ----------
def build_digest(analysis: dict, weather: Optional[dict] = None) -> dict:
    """Compact, LLM-friendly digest of the calculated results (the ONLY numbers the LLM may use)."""
    K, TR, TP, TC, CU, AN, WI = (analysis.get(k) or {} for k in ("kpis", "trends", "top_products", "top_countries", "customers", "anomalies", "weather_impact"))
    monthly = [{"month": m["YearMonth"], "revenue": round(m["revenue"])} for m in TR.get("monthly", []) if not m.get("is_partial")]
    return jsonable({
        "currency": "GBP (assumed)", "kpis": {k: K.get(k) for k in ("total_revenue", "returns_value", "net_revenue", "total_orders", "unique_customers", "unique_products", "average_order_value",
                                                                     "average_items_per_order", "return_rate_invoices_pct", "return_rate_value_pct", "date_start", "date_end")},
        "top_products": [{"name": r["description"], "revenue": round(r["revenue"]), "share_pct": r["revenue_share_pct"]} for r in TP.get("by_revenue", [])[:5]],
        "top_countries": [{"country": r["Country"], "share_pct": r["revenue_share_pct"]} for r in TC.get("top", [])[:5]], "non_uk_share_pct": TC.get("non_uk_share_pct"),
        "monthly_revenue_complete_months": monthly, "trend_direction": TR.get("trend_direction"), "peak_month": TR.get("peak_month"), "partial_months": TR.get("partial_months"),
        "customers": {k: CU.get(k) for k in ("identified_customers", "top10pct_revenue_share_pct", "repeat_customer_rate_pct", "revenue_share_without_customer_id_pct", "median_spend")},
        "anomalies": {k: AN.get(k) for k in ("flagged_lines", "flagged_lines_pct", "flagged_revenue_share_pct")},
        "weather": {"available": WI.get("available"), "pearson_temperature": WI.get("pearson_temperature"), "overlap_days": WI.get("overlap_days"),
                    "source": (weather or {}).get("source")}})

def _numbers(text: str) -> set[str]:
    out = set()
    for tok in re.findall(r"\d[\d,]*\.?\d*", str(text)):
        tok = tok.replace(",", "").rstrip(".")
        if len(tok.replace(".", "")) >= 3:
            out.add(tok)
            try: out.add(str(int(round(float(tok)))))
            except ValueError: pass
    return out

def unsupported_numbers(text: str, allowed_text: str) -> list[str]:
    """Numbers (>=3 digits) in `text` that do not occur in `allowed_text` (used to flag possibly fabricated figures)."""
    allowed = _numbers(allowed_text)
    bad = []
    for n in _numbers(text):
        near = any(abs(float(n) - float(a)) <= max(1.0, 0.005 * abs(float(a))) for a in allowed) if allowed else False
        if not near: bad.append(n)
    return sorted(set(bad))

def summarize_annual_report(evidence: list[dict]) -> dict:
    """Summarise retrieved annual-report chunks. LLM if available (page-cited bullets), else extractive snippets."""
    if not evidence:
        return {"mode": "none", "bullets": [], "note": "No evidence retrieved."}
    pages = {e["page"] for e in evidence}
    if LLM is not None:
        try:
            excerpts = "\n\n".join(f"[Page {e['page']}] {e['text'][:650]}" for e in evidence[:6])
            out = llm_json("You summarise excerpts from Target Corporation's annual report. Use ONLY the excerpts; never add outside facts or numbers. "
                           "Return JSON: {\"bullets\": [{\"text\": \"<=35 words\", \"page\": <page number from the excerpt label>}]} with 4-6 bullets.",
                           f"Excerpts:\n{excerpts}")
            bullets = []
            for b in (out.get("bullets") or [])[:6]:
                page = int(re.sub(r"\D", "", str(b.get("page", ""))) or 0)
                if page in pages and str(b.get("text", "")).strip():
                    bullets.append(f"{str(b['text']).strip()} (Annual Report - Page {page})")
            if bullets:
                return {"mode": "llm", "bullets": bullets}
            raise ValueError("no bullet with a valid page citation")
        except Exception as exc:
            LOG.warning("annual_report_summary", f"LLM summary failed ({type(exc).__name__}: {str(exc)[:150]}); using extractive summary.")
    bullets = []
    for e in evidence[:5]:
        sent = re.split(r"(?<=[.!?])\s+", e["text"])
        bullets.append((" ".join(sent[:2])[:320]).strip() + f" (Annual Report - Page {e['page']})")
    return {"mode": "extractive (no LLM)", "bullets": bullets}

def build_deterministic_insights(analysis: dict, weather: Optional[dict]) -> tuple[list[dict], list[dict]]:
    """Rule-based CALCULATED RESULT statements and RECOMMENDATIONS derived strictly from the computed numbers."""
    K, TR, TP, TC, CU, AN, WI = (analysis.get(k) or {} for k in ("kpis", "trends", "top_products", "top_countries", "customers", "anomalies", "weather_impact"))
    ins: list[Insight] = []; recs: list[Insight] = []
    def calc(text, ev): ins.append(Insight(category="CALCULATED RESULT", text=text, evidence=ev, source="deterministic"))
    def rec(text, ev): recs.append(Insight(category="RECOMMENDATION", text=text, evidence=ev, source="deterministic"))
    if K:
        calc(f"Gross sales revenue was GBP {K['total_revenue']:,.0f} from {K['total_orders']:,} orders and {K['unique_customers']:,} identified customers; average order value GBP {K['average_order_value']:,.2f}.", ["KPI: calculate_kpis"])
        calc(f"Returns/cancellations represent {K['return_rate_value_pct']}% of sales value and {K['return_rate_invoices_pct']}% of invoices.", ["KPI: calculate_kpis"])
    if TP.get("by_revenue"):
        t = TP["by_revenue"][0]
        calc(f"Top product by revenue: '{t['description']}' (GBP {t['revenue']:,.0f}, {t['revenue_share_pct']}% of product revenue); the top {len(TP['by_revenue'])} products make up {TP['top_n_revenue_share_pct']}%.", ["get_top_products"])
    if TC.get("top"):
        calc(f"{TC['top_country']} contributes {TC['top_country_share_pct']}% of revenue; all other countries together {TC['non_uk_share_pct']}%.", ["get_top_countries"])
    if TR.get("peak_month"):
        calc(f"Revenue peaked in {TR['peak_month']['month']} (GBP {TR['peak_month']['revenue']:,.0f}); the trend over complete months is {TR.get('trend_direction', 'n/a')}.", ["analyze_sales_trends"])
    if CU:
        calc(f"The top 10% of identified customers generate {CU['top10pct_revenue_share_pct']}% of identified revenue; {CU['repeat_customer_rate_pct']}% of customers ordered more than once; {CU['revenue_share_without_customer_id_pct']}% of revenue has no CustomerID.", ["analyze_customers"])
    if AN:
        calc(f"{AN['flagged_lines']:,} product lines ({AN['flagged_lines_pct']}%) are statistical anomalies, worth {AN['flagged_revenue_share_pct']}% of product revenue (unusual, not necessarily errors or fraud).", ["detect_anomalies"])
    if WI.get("available") and WI.get("pearson_temperature") is not None:
        calc(f"Daily revenue vs mean temperature: Pearson r = {WI['pearson_temperature']} over {WI['overlap_days']} days (source: {(weather or {}).get('source')}); association only.", ["analyze_weather_impact"])
    # recommendations (conditional rules)
    if CU.get("top10pct_revenue_share_pct", 0) >= 40:
        rec(f"Protect revenue concentration: the top 10% of customers drive {CU['top10pct_revenue_share_pct']}% of identified revenue - introduce account management/loyalty benefits for this segment.", ["analyze_customers"])
    if CU.get("repeat_customer_rate_pct", 100) < 70:
        rec(f"Raise repeat purchasing (currently {CU['repeat_customer_rate_pct']}% of customers reorder) with post-purchase follow-ups and bundles.", ["analyze_customers"])
    if TC.get("top_country_share_pct", 0) >= 75 and TC.get("top_non_uk"):
        names = ", ".join(r["Country"] for r in TC["top_non_uk"][:3])
        rec(f"Reduce dependence on {TC['top_country']} ({TC['top_country_share_pct']}% of revenue): pilot growth campaigns in the strongest export markets ({names}).", ["get_top_countries"])
    if K.get("return_rate_value_pct", 0) >= 3:
        rec(f"Investigate cancellations ({K['return_rate_value_pct']}% of sales value): review the largest return lines and product quality/description issues.", ["calculate_kpis", "detect_anomalies"])
    pk = TR.get("peak_month", {}).get("month", "")
    if pk[-2:] in {"10", "11", "12"}:
        rec(f"Plan inventory and marketing ahead of the autumn peak (peak month {pk}) to avoid stock-outs.", ["analyze_sales_trends"])
    if AN.get("flagged_lines", 0) > 0:
        rec("Review the flagged statistical anomalies (very large quantities/prices) to separate genuine bulk orders from data-entry or pricing issues.", ["detect_anomalies"])
    if CU.get("revenue_share_without_customer_id_pct", 0) >= 10:
        rec(f"Improve customer identification at checkout: {CU['revenue_share_without_customer_id_pct']}% of revenue cannot be linked to a customer.", ["analyze_customers"])
    if WI.get("available") and WI.get("pearson_temperature") is not None:
        r = abs(WI["pearson_temperature"])
        rec("Weather shows only a weak association with daily revenue in this data, so do not prioritise weather-based planning without better-matched local weather data." if r < 0.3
            else "Test weather-aware promotions in a controlled experiment; the association is notable but not proof of causation.", ["analyze_weather_impact"])
    return [i.model_dump() for i in ins], [r.model_dump() for r in recs]

def generate_llm_insights(digest: dict, evidence: list[dict], ar_summary: dict) -> tuple[list[dict], dict]:
    """Ask the LLM for FACT / LLM INTERPRETATION / RECOMMENDATION items, then ground-check them. Returns (items, meta)."""
    meta = {"requested": LLM is not None, "items_returned": 0, "fact_claimed": 0, "fact_grounded": 0, "fact_downgraded": 0, "items_with_unsupported_numbers": 0, "error": None}
    if LLM is None:
        return [], meta
    pages = {e["page"] for e in STORE.evidence}
    ev_text = "\n".join(f"[Page {e['page']}] {e['text'][:500]}" for e in STORE.evidence[:6])
    system = ("You are a senior business analyst. Sources: DIGEST (numbers calculated by Python) and EXCERPTS (Target Corporation annual report, with page numbers). "
              "IMPORTANT: the sales dataset is from a different (UK online) retailer than Target; annual-report content is external market context, never the retailer's own data. "
              "Rules: (1) use ONLY numbers that appear in DIGEST or EXCERPTS, never invent figures; (2) category FACT = a statement taken from EXCERPTS and MUST list evidence like 'Page 12'; "
              "(3) category LLM INTERPRETATION = your reasoned hypothesis linking digest results and excerpts, worded cautiously; (4) category RECOMMENDATION = a concrete action referencing digest numbers. "
              "Return ONLY JSON: {\"items\": [{\"category\": \"FACT|LLM INTERPRETATION|RECOMMENDATION\", \"text\": \"<=45 words\", \"evidence\": [\"Page N\" or \"digest: <field>\"]}]}. "
              "Produce 2-3 FACT, 3 LLM INTERPRETATION and 4 RECOMMENDATION items.")
    user = f"DIGEST:\n{json.dumps(digest, default=str)}\n\nEXCERPTS:\n{ev_text}"
    try:
        out = llm_json(system, user)
        raw_items = out.get("items", out) if isinstance(out, dict) else out
    except Exception as exc:
        meta["error"] = f"{type(exc).__name__}: {str(exc)[:200]}"
        LOG.warning("insight_llm", f"LLM insight generation failed: {meta['error']}")
        return [], meta
    allowed = json.dumps(digest, default=str) + ev_text
    cat_map = {"FACT": "FACT", "LLM INTERPRETATION": "LLM INTERPRETATION", "INTERPRETATION": "LLM INTERPRETATION", "RECOMMENDATION": "RECOMMENDATION",
               "RECOMMENDATIONS": "RECOMMENDATION", "CALCULATED RESULT": "LLM INTERPRETATION"}
    items: list[dict] = []
    for raw in (raw_items or [])[:14]:
        try:
            cat = cat_map.get(str(raw.get("category", "")).strip().upper())
            text = str(raw.get("text", "")).strip()
            if not cat or not text:
                continue
            evid = raw.get("evidence") or []
            evid = [str(evid)] if isinstance(evid, str) else [str(x) for x in evid]
            note = ""
            if cat == "FACT":
                meta["fact_claimed"] += 1
                cited = {int(n) for e in evid for n in re.findall(r"[Pp]age\s*(\d+)", e)}
                if cited & pages:
                    meta["fact_grounded"] += 1
                else:
                    cat, note = "LLM INTERPRETATION", "Downgraded from FACT: no valid page citation."
                    meta["fact_downgraded"] += 1
            bad = unsupported_numbers(text, allowed)
            if bad:
                meta["items_with_unsupported_numbers"] += 1
                note = (note + " " if note else "") + f"Contains numbers not found in the inputs: {', '.join(bad[:4])}."
            items.append(Insight(category=cat, text=text[:600], evidence=evid[:4], source="llm", note=note).model_dump())
        except Exception:
            continue
    meta["items_returned"] = len(items)
    return items, meta


# %% Cell 44 (code)
# ---------- Report Agent helpers ----------
def build_limitations(state: dict) -> list[str]:
    """Static + run-dependent limitations that are printed in the report."""
    ds = state.get("data_summary", {}).get("sources", {}); lm = state.get("llm_meta", {}); an = state.get("analysis_results", {})
    lim = ["The annual report is Target Corporation's, while the sales dataset is a different (UK online) retailer: report passages are external market context, not the retailer's own performance.",
           "Weather default location is Berlin (52.52N, 13.42E) while most customers are in the UK; correlation with sales is exploratory, is not causal, and may be confounded by seasonality.",
           "Currency is assumed to be GBP; the dataset has no currency field.",
           "The data covers roughly one year: seasonality conclusions rest on a single cycle. Partial months are excluded from growth/trend statistics.",
           "Anomalies are purely statistical (IQR k=3 / robust z-score): they flag unusual values, not fraud or errors.",
           "Annual-report retrieval is lexical (TF-IDF): it can miss paraphrases; page numbers are physical PDF pages and may differ from printed page numbers.",
           "Customer analytics covers only rows with a CustomerID (a substantial share of rows has none)."]
    if not lm.get("available"): lim.append("No Groq LLM was available for this run: interpretations, annual-report summary and executive text use deterministic fallbacks. Reason: " + str(lm.get("reason", "n/a"))[:200])
    if ds.get("weather", {}).get("fallback_used"): lim.append("The live Open-Meteo API was unavailable; the local CSV fallback was used.")
    if an.get("anomalies", {}).get("method") == "zscore": lim.append("Z-score method is sensitive to heavy tails in retail data.")
    return lim

def deterministic_exec_summary(digest: dict) -> dict:
    K = digest.get("kpis", {}); TC = digest.get("top_countries", [{}]); TP = digest.get("top_products", [{}]); CU = digest.get("customers", {}); AN = digest.get("anomalies", {})
    parts = []
    if K.get("total_revenue") is not None:
        parts.append(f"Between {K.get('date_start')} and {K.get('date_end')} the business generated GBP {K['total_revenue']:,.0f} in gross sales from {K['total_orders']:,} orders and {K['unique_customers']:,} identified customers (average order value GBP {K['average_order_value']:,.2f}).")
    if TC and TC[0].get("country"): parts.append(f"{TC[0]['country']} accounts for {TC[0]['share_pct']}% of revenue; the top product is '{TP[0].get('name', 'n/a')}'.")
    if digest.get("trend_direction"): parts.append(f"Revenue across complete months is {digest['trend_direction']}" + (f", peaking in {digest['peak_month']['month']}." if digest.get("peak_month") else "."))
    if K.get("return_rate_value_pct") is not None: parts.append(f"Returns are {K['return_rate_value_pct']}% of sales value.")
    if CU.get("top10pct_revenue_share_pct") is not None: parts.append(f"The top 10% of identified customers generate {CU['top10pct_revenue_share_pct']}% of identified revenue.")
    if AN.get("flagged_lines") is not None: parts.append(f"{AN['flagged_lines']:,} product lines were flagged as statistical anomalies for review.")
    concl = ("The analysis combined deterministic Pandas calculations with retrieved annual-report context and weather data. Priorities are retaining high-value customers, "
             "reducing dependence on the main market, understanding returns, and reviewing flagged anomalies. See the recommendations section for details.")
    return {"executive_summary": " ".join(parts) or "Summary unavailable.", "conclusion": concl, "mode": "deterministic"}

def generate_exec_summary(digest: dict) -> dict:
    """LLM-written executive summary/conclusion; falls back to a template when the LLM fails or uses numbers not in the digest."""
    base = deterministic_exec_summary(digest)
    if LLM is None:
        return base
    try:
        out = llm_json("You write executive summaries for a business report. Use ONLY facts and numbers from the DIGEST; do not invent figures. "
                       "Return JSON: {\"executive_summary\": \"110-160 words\", \"conclusion\": \"60-90 words\"}. Plain prose, no markdown.",
                       f"DIGEST:\n{json.dumps(digest, default=str)}")
        es, co = str(out.get("executive_summary", "")).strip(), str(out.get("conclusion", "")).strip()
        bad = unsupported_numbers(es + " " + co, json.dumps(digest, default=str))
        if not es or not co:
            raise ValueError("missing fields")
        if bad:
            LOG.warning("exec_summary", f"LLM text contained numbers not in the digest ({bad[:5]}); using deterministic summary.")
            return base
        return {"executive_summary": es, "conclusion": co, "mode": "llm"}
    except Exception as exc:
        LOG.warning("exec_summary", f"LLM summary failed ({type(exc).__name__}: {str(exc)[:150]}); using deterministic summary.")
        return base

def assemble_report_data(state: dict, exec_summary: dict) -> dict:
    ds = state.get("data_summary", {}).get("sources", {}); lm = state.get("llm_meta", {}); wm = state.get("weather_results", {})
    s, a, w = ds.get("sales", {}), ds.get("annual_report", {}), ds.get("weather", {})
    sources = [
        {"name": "Online Retail (Excel)", "status": "Loaded" if s.get("available") else "Missing", "detail": f"{s.get('file', '')}: {s.get('rows', 0):,} raw rows, {s.get('date_start', '')} to {s.get('date_end', '')}" if s.get("available") else str(s.get("error", ""))},
        {"name": "Target Annual Report (PDF, RAG)", "status": "Indexed" if a.get("available") else "Missing", "detail": f"{a.get('file', '')}: {a.get('pages', 0)} pages, {a.get('chunks', 0)} chunks; {a.get('evidence_chunks', 0)} evidence chunks retrieved" if a.get("available") else str(a.get("error", ""))},
        {"name": "Open-Meteo weather", "status": "Retrieved" if w.get("available") else "Missing", "detail": f"{w.get('source', '')}; fallback used: {w.get('fallback_used')}" if w.get("available") else str(w.get("error", ""))},
        {"name": "Groq LLM", "status": "Used" if lm.get("available") else "Unavailable", "detail": f"model {lm.get('model')}" if lm.get("available") else str(lm.get("reason", ""))[:160]}]
    return {"generated_at": dt.datetime.now().strftime("%Y-%m-%d %H:%M"), "user_request": state.get("user_request", ""), "sources": sources, "data_quality": state.get("data_quality_raw", {}) or state.get("data_quality", {}),
            "cleaning_report": state.get("cleaning_report", {}), "analysis": state.get("analysis_results", {}), "evidence": state.get("annual_report_evidence", []), "ar_summary": state.get("annual_report_summary", {}),
            "weather": wm, "insights": state.get("insights", []), "recommendations": state.get("recommendations", []), "exec_summary": exec_summary, "limitations": build_limitations(state), "llm_meta": lm}


# %% Cell 45 (code)
# ---------- Graph nodes ----------
@agent_node("Request Check", is_agent=False)
def node_request_check(state: dict) -> dict:
    try:
        req = WorkflowRequest(user_request=state.get("user_request", ""))
        LOG.info("Request Check", "Request validated")
        return {"request_valid": True, "user_request": req.user_request}
    except Exception as exc:
        msg = "; ".join(str(exc).splitlines()[:3])
        LOG.error("Request Check", f"Invalid request: {msg}")
        return {"request_valid": False, "errors": [{"node": "Request Check", "error": f"Invalid request: {msg}", "severity": "error"}]}

@agent_node("Data Retrieval Agent")
def node_retrieval(state: dict) -> dict:
    task = (f"User request: {state['user_request']}\nSources: Online Retail sales (Excel), Target annual report (PDF), Open-Meteo weather. "
            "Retrieve what is needed for this request.")
    trace = build_retrieval_agent().run(task)
    # sanity: if the LLM chose a weather date range that does not cover the sales period, redo with the sales range
    meta, raw = STORE.weather_meta, STORE.raw_df
    if raw is not None and meta and any(c["tool"] == "get_weather_data" for c in trace["llm_tool_calls"]):
        d = pd.to_datetime(raw["InvoiceDate"], errors="coerce")
        if not (meta.get("date_start", "9") <= str(d.min().date()) and meta.get("date_end", "0") >= str(d.max().date())) and str(meta.get("source", "")).startswith("open-meteo"):
            LOG.warning("Data Retrieval Agent", "LLM-chosen weather range does not cover the sales period; re-fetching with the sales range.")
            get_weather_data_tool(); trace["deterministic_fallback"].append("get_weather_data (range corrected)")
    s_res, w_res = RESULTS.get("load_sales_data", {}), RESULTS.get("get_weather_data", {})
    idx = STORE.rag_index
    summary = {"sources": {
        "sales": {"available": STORE.raw_df is not None, "file": Path(STORE.paths["sales"]).name if STORE.paths.get("sales") else None, "rows": s_res.get("rows"), "date_start": s_res.get("date_start"), "date_end": s_res.get("date_end"), "error": RESULT_ERRORS.get("load_sales_data")},
        "annual_report": {"available": idx is not None, "file": Path(STORE.paths["report"]).name if STORE.paths.get("report") else None, "pages": getattr(idx, "n_pages", None), "chunks": len(idx.chunks) if idx else None,
                          "queries_run": list(SEARCH_QUERIES), "evidence_chunks": len(STORE.evidence), "error": RESULT_ERRORS.get("search_annual_report")},
        "weather": {"available": STORE.weather_df is not None, "source": w_res.get("source"), "fallback_used": w_res.get("fallback_used"), "rows": w_res.get("rows"), "error": RESULT_ERRORS.get("get_weather_data")}}}
    errs = [{"node": "Data Retrieval Agent", "tool": t["tool"], "error": t["error"], "severity": "error" if t["tool"] == "load_sales_data" else "warning"} for t in trace["tool_errors"]]
    if trace["llm_error"]:
        errs.append({"node": "Data Retrieval Agent", "tool": "llm", "error": trace["llm_error"], "severity": "warning"})
    LOG.info("Data Retrieval Agent", f"Availability: sales={summary['sources']['sales']['available']}, annual_report={summary['sources']['annual_report']['available']}, weather={summary['sources']['weather']['available']}")
    return {"data_summary": summary, "weather_results": w_res if isinstance(w_res, dict) else {}, "annual_report_evidence": jsonable(STORE.evidence), "agent_traces": {"Data Retrieval Agent": trace}, "errors": errs,
            "llm_meta": {"available": LLM_STATUS["available"], "model": LLM_STATUS["model"], "reason": LLM_STATUS["reason"], "key_source": LLM_STATUS["key_source"]}}

@agent_node("Data Quality Check", is_agent=False)
def node_quality_check(state: dict) -> dict:
    cleaned = STORE.clean_df is not None
    df = STORE.clean_df if cleaned else STORE.raw_df
    if df is None:
        rep = DataQualityReport(rows=0, columns=0, blocking_issues=["Sales data could not be loaded"], acceptable=False).model_dump()
    else:
        rep = validate_sales_data(df, is_cleaned=cleaned)
    upd: dict[str, Any] = {"data_quality": rep}
    if not cleaned:
        upd["data_quality_raw"] = rep                  # keep the raw-data diagnosis for the report
    LOG.info("Data Quality Check", f"{'post-cleaning' if cleaned else 'raw'} check: acceptable={rep['acceptable']}, defect_rows={rep.get('defect_rows')}, blocking={rep['blocking_issues']}")
    return upd

def route_after_quality(state: dict) -> str:
    q = state.get("data_quality", {})
    # The analysis tools require engineered features; always pass through the
    # cleaning node once per run, even when the raw data has no defects.
    if not q.get("is_cleaned"): return "clean"
    if q.get("blocking_issues"): return "abort"
    return "analysis"          # cleaned, residual non-blocking issues only

@agent_node("Cleaning", is_agent=False)
def node_cleaning(state: dict) -> dict:
    clean, rep = clean_sales_data(STORE.raw_df)
    STORE.clean_df, STORE.cleaning_report = engineer_features(clean), rep
    LOG.info("Cleaning", f"{rep['rows_before']:,} -> {rep['rows_after']:,} rows; returns kept: {rep['returns_kept']:,}; features added")
    return {"cleaning_report": rep, "quality_attempts": state.get("quality_attempts", 0) + 1}

@agent_node("Data Analysis Agent")
def node_analysis(state: dict) -> dict:
    trace = build_analysis_agent().run("Run the complete business analysis for the request: " + state["user_request"])
    keys = {"kpis": "calculate_kpis", "trends": "analyze_sales_trends", "top_products": "get_top_products", "top_countries": "get_top_countries",
            "customers": "analyze_customers", "anomalies": "detect_anomalies", "weather_impact": "analyze_weather_impact"}
    analysis = {k: RESULTS.get(t) for k, t in keys.items() if RESULTS.get(t) and "error" not in RESULTS.get(t, {})}
    errs = [{"node": "Data Analysis Agent", "tool": e["tool"], "error": e["error"], "severity": "error"} for e in trace["tool_errors"]]
    if trace["llm_error"]: errs.append({"node": "Data Analysis Agent", "tool": "llm", "error": trace["llm_error"], "severity": "warning"})
    if "kpis" in analysis: LOG.info("Data Analysis Agent", f"KPI calculation completed: revenue={analysis['kpis']['total_revenue']:,.0f}, orders={analysis['kpis']['total_orders']:,}")
    return {"analysis_results": analysis, "agent_traces": {"Data Analysis Agent": trace}, "errors": errs}

@agent_node("Research / Insight Agent")
def node_insight(state: dict) -> dict:
    analysis, weather = state.get("analysis_results", {}), state.get("weather_results", {})
    digest = build_digest(analysis, weather)
    agent = ToolAgent(name="Research / Insight Agent", tools=INSIGHT_TOOLS, result_chars=1000,
                      system_prompt=("You are the Research/Insight Agent. The sales analysis is finished (digest below). Use search_annual_report 2-3 times with focused queries "
                                     "(e.g. revenue growth, customers/loyalty, seasonality, supply chain) to find annual-report context that relates to these results. Then reply DONE."),
                      required={"search_annual_report": lambda: _run_searches(INSIGHT_QUERIES)})
    trace = agent.run(f"Request: {state['user_request']}\nDIGEST: {json.dumps(digest, default=str)[:1800]}")
    if STORE.rag_index is not None and len(STORE.evidence) < 6:
        _run_searches(INSIGHT_QUERIES)
    evidence = jsonable(STORE.evidence[:8])
    summary = summarize_annual_report(evidence)
    LOG.info("Research / Insight Agent", f"Annual report retrieval completed: {len(evidence)} evidence chunks, summary mode={summary['mode']}")
    det_ins, det_recs = build_deterministic_insights(analysis, weather)
    llm_items, gmeta = generate_llm_insights(digest, evidence, summary)
    insights = det_ins + [i for i in llm_items if i["category"] != "RECOMMENDATION"]
    recs = [i for i in llm_items if i["category"] == "RECOMMENDATION"] + det_recs
    errs = []
    if LLM is None: errs.append({"node": "Research / Insight Agent", "tool": "llm", "error": "LLM unavailable: " + str(LLM_STATUS["reason"])[:200] + " (deterministic insights only)", "severity": "warning"})
    elif gmeta.get("error"): errs.append({"node": "Research / Insight Agent", "tool": "llm", "error": gmeta["error"], "severity": "warning"})
    if trace["llm_error"]: errs.append({"node": "Research / Insight Agent", "tool": "llm", "error": trace["llm_error"], "severity": "warning"})
    llm_meta = dict(state.get("llm_meta", {}), grounding=gmeta, annual_report_summary_mode=summary["mode"])
    LOG.info("Research / Insight Agent", f"Insight agent completed: {len(insights)} insights, {len(recs)} recommendations")
    return {"annual_report_evidence": evidence, "annual_report_summary": summary, "insights": insights, "recommendations": recs, "llm_meta": llm_meta, "agent_traces": {"Research / Insight Agent": trace}, "errors": errs}

@agent_node("Report Agent")
def node_report(state: dict) -> dict:
    errs, charts = [], {}
    try:
        charts = generate_charts(STORE.clean_df, state.get("analysis_results", {}), OUTPUT_DIR / "charts", STORE.weather_df)
        LOG.info("Report Agent", f"Charts generated: {sorted(charts)}")
    except Exception as exc:
        errs.append({"node": "Report Agent", "tool": "generate_charts", "error": f"{type(exc).__name__}: {exc}", "severity": "warning"})
    es = generate_exec_summary(build_digest(state.get("analysis_results", {}), state.get("weather_results", {})))
    rd = assemble_report_data(state, es)
    path = generate_pdf_report(rd, charts, OUTPUT_DIR / "multi_agent_business_report.pdf")
    LOG.info("Report Agent", f"PDF generated: {path}")
    return {"report_path": str(path), "chart_paths": charts, "exec_summary": es, "errors": errs}

@agent_node("Action Agent")
def node_action(state: dict) -> dict:
    rp = Path(state.get("report_path") or "")
    if not rp.exists() or rp.stat().st_size == 0:
        raise FileNotFoundError("Report PDF is missing - nothing to send/save.")
    from pypdf import PdfReader
    n_pages = len(PdfReader(str(rp)).pages)
    analysis_path = OUTPUT_DIR / "analysis_results.json"
    analysis_path.write_text(json.dumps(jsonable(state.get("analysis_results", {})), indent=2), encoding="utf-8")
    rd = assemble_report_data(state, state.get("exec_summary", {}))
    email = generate_email_summary(rd, rp, REPORT_RECIPIENT)
    email_path = save_email_summary(email, OUTPUT_DIR / "email_summary.txt")
    delivery = send_email_if_configured(email)
    LOG.info("Action Agent", f"Report verified ({n_pages} pages, {rp.stat().st_size/1024:.0f} KB); email summary saved; sending: {delivery['reason']}")
    return {"email_summary": dict(email, delivery=delivery), "artifacts": {"report_pdf": str(rp), "report_pages": n_pages, "analysis_results_json": str(analysis_path), "email_summary_txt": str(email_path)}}

@agent_node("Abort", is_agent=False)
def node_abort(state: dict) -> dict:
    LOG.error("Abort", "Workflow stopped: " + "; ".join(state.get("data_quality", {}).get("blocking_issues", []) or ["unknown reason"]))
    return {"errors": [{"node": "Abort", "error": "Workflow aborted due to blocking data issues: " + "; ".join(state.get("data_quality", {}).get("blocking_issues", [])), "severity": "error"}]}


# %% Cell 46 (markdown)
# # # 17. LangGraph Orchestration
# #
# # ```
# # START -> request_check --(invalid)--> END
# #              | valid
# #              v
# #          retrieval_agent -> quality_check --(acceptable)-------------> analysis_agent
# #                                  |  (needs cleaning)                        |
# #                                  +--> cleaning --> quality_check (re-check) v
# #                                  |  (blocking issue)                  insight_agent -> report_agent -> action_agent -> END
# #                                  +--> abort --> END
# # ```
# # Features used: typed **state**, **nodes**, **edges**, **conditional edges** (request validity, data quality), a **loop** (clean -> re-check), error handling inside every node, and structured logging.

# %% Cell 47 (code)
from langgraph.graph import StateGraph, START, END

def route_after_request(state: dict) -> str:
    return "retrieval_agent" if state.get("request_valid") else "end"

def build_workflow():
    g = StateGraph(AgentState)
    g.add_node("request_check", node_request_check)
    g.add_node("retrieval_agent", node_retrieval)
    g.add_node("quality_check", node_quality_check)
    g.add_node("cleaning", node_cleaning)
    g.add_node("analysis_agent", node_analysis)
    g.add_node("insight_agent", node_insight)
    g.add_node("report_agent", node_report)
    g.add_node("action_agent", node_action)
    g.add_node("abort", node_abort)
    g.add_edge(START, "request_check")
    g.add_conditional_edges("request_check", route_after_request, {"retrieval_agent": "retrieval_agent", "end": END})
    g.add_edge("retrieval_agent", "quality_check")
    g.add_conditional_edges("quality_check", route_after_quality, {"analysis": "analysis_agent", "clean": "cleaning", "abort": "abort"})
    g.add_edge("cleaning", "quality_check")
    g.add_edge("analysis_agent", "insight_agent")
    g.add_edge("insight_agent", "report_agent")
    g.add_edge("report_agent", "action_agent")
    g.add_edge("action_agent", END)
    g.add_edge("abort", END)
    return g.compile()

WORKFLOW_APP = build_workflow()
print("LangGraph workflow compiled.")
try:
    print("\nMermaid definition of the compiled graph:\n")
    print(WORKFLOW_APP.get_graph().draw_mermaid())
except Exception as exc:
    print("(Mermaid export not available in this langgraph version:", type(exc).__name__, ")")


# %% Cell 48 (code)
from matplotlib.patches import FancyBboxPatch

def draw_architecture(path: Path) -> Path:
    """Draw the system architecture with Matplotlib and save it as PNG."""
    fig, ax = plt.subplots(figsize=(13, 9)); ax.set_xlim(0, 1); ax.set_ylim(0, 1); ax.axis("off"); ax.grid(False)
    def box(x, y, w, h, text, fc, tc="white", fs=9):
        ax.add_patch(FancyBboxPatch((x - w / 2, y - h / 2), w, h, boxstyle="round,pad=0.008", fc=fc, ec="#1F3A5F", lw=1.2))
        ax.text(x, y, text, ha="center", va="center", color=tc, fontsize=fs, fontweight="bold")
    def arrow(x1, y1, x2, y2, style="-|>", ls="-", color="#1F3A5F"):
        ax.annotate("", xy=(x2, y2), xytext=(x1, y1), arrowprops=dict(arrowstyle=style, lw=1.3, ls=ls, color=color))
    cx = 0.45
    ys = {"user": 0.94, "orch": 0.85, "ret": 0.75, "dq": 0.65, "ana": 0.54, "ins": 0.44, "rep": 0.34, "act": 0.24, "out": 0.12}
    box(cx, ys["user"], 0.26, 0.05, "User business request", "#555555")
    box(cx, ys["orch"], 0.36, 0.06, "Orchestrator (LangGraph state machine)\nrequest check | routing | logging | error handling", "#1F3A5F", fs=8.5)
    box(cx, ys["ret"], 0.30, 0.055, "Data Retrieval Agent\n(Excel | PDF RAG | Weather API)", "#2A9D8F", fs=8.5)
    box(cx, ys["dq"], 0.30, 0.055, "Data Quality Validation\n(conditional routing)", "#E9C46A", tc="#1F3A5F", fs=8.5)
    box(0.12, ys["dq"], 0.17, 0.05, "Cleaning\n(keep returns)", "#F4A261", tc="#1F3A5F", fs=8.5)
    box(cx, ys["ana"], 0.30, 0.055, "Data Analysis Agent\n(Pandas tools: KPIs, trends, anomalies)", "#2A9D8F", fs=8.5)
    box(cx, ys["ins"], 0.30, 0.055, "Research / Insight Agent\n(RAG evidence + interpretation)", "#2A9D8F", fs=8.5)
    box(cx, ys["rep"], 0.30, 0.055, "Report Agent\n(charts + ReportLab PDF)", "#2A9D8F", fs=8.5)
    box(cx, ys["act"], 0.30, 0.055, "Action Agent\n(save files + e-mail summary)", "#2A9D8F", fs=8.5)
    box(cx, ys["out"], 0.36, 0.06, "Outputs: PDF report | JSON results | execution log | e-mail summary", "#264653", fs=8.5)
    order = ["user", "orch", "ret", "dq", "ana", "ins", "rep", "act", "out"]
    for a, b in zip(order[:-1], order[1:]):
        arrow(cx, ys[a] - 0.03, cx, ys[b] + 0.03)
    arrow(cx - 0.15, ys["dq"], 0.205, ys["dq"], color="#C0392B"); arrow(0.12, ys["dq"] + 0.027, 0.12, ys["dq"] + 0.06, style="-", color="#C0392B")
    arrow(0.12, ys["dq"] + 0.06, cx - 0.12, ys["dq"] + 0.028, color="#C0392B")
    ax.text(0.135, ys["dq"] - 0.045, "not acceptable -> clean -> re-check", fontsize=7.5, color="#C0392B")
    ax.text(cx + 0.02, ys["dq"] - 0.052, "acceptable", fontsize=7.5, color="#1F3A5F")
    src_x = 0.86
    for y, t, c in [(0.82, "Online Retail\n(Excel)", "#8D99AE"), (0.72, "Target Annual Report\n(PDF)", "#8D99AE"), (0.62, "Open-Meteo\n(API + CSV fallback)", "#8D99AE")]:
        box(src_x, y, 0.22, 0.06, t, c, fs=8.5); arrow(src_x - 0.11, y, cx + 0.15, ys["ret"], color="#555555")
    box(src_x, 0.42, 0.22, 0.09, "Groq LLM\n(tool calling, summaries,\ninterpretation)", "#6D597A", fs=8.5)
    for k in ("ret", "ana", "ins", "rep"):
        arrow(src_x - 0.11, 0.42, cx + 0.15, ys[k], ls="--", color="#6D597A")
    ax.text(0.5, 0.01, "Dashed = LLM calls (optional; deterministic fallback exists). Heavy DataFrames live in a DataStore, not in the graph state.", ha="center", fontsize=8, color="#555555")
    fig.savefig(path, dpi=130, bbox_inches="tight"); plt.close(fig)
    return path

ARCH_PATH = draw_architecture(OUTPUT_DIR / "architecture_diagram.png")
display(Image(filename=str(ARCH_PATH)))


# %% Cell 49 (markdown)
# # # 18. Workflow Execution
# # `run_workflow()` resets per-run caches, invokes the compiled graph, derives the final status and writes `workflow_results.json` and `execution_log.json`. `analysis_results.json`, `email_summary.txt` and the PDF are written by the Action Agent.

# %% Cell 50 (code)
def finalize_outputs(final: dict) -> dict:
    """Write workflow_results.json and execution_log.json (the API key is never part of the state)."""
    log_path, wf_path = OUTPUT_DIR / "execution_log.json", OUTPUT_DIR / "workflow_results.json"
    log_path.write_text(json.dumps(jsonable(final.get("execution_log", [])), indent=2), encoding="utf-8")
    slim = {k: v for k, v in final.items() if k != "execution_log"}
    wf_path.write_text(json.dumps(jsonable(slim), indent=2), encoding="utf-8")
    return {"workflow_results_json": str(wf_path), "execution_log_json": str(log_path)}

def run_workflow(user_request: str, app: Any = None) -> dict:
    """Execute the whole multi-agent workflow and return the final state (with status, wall time and artifact paths)."""
    app = app or WORKFLOW_APP
    LOG.reset(); RESULTS.clear(); RESULT_ERRORS.clear(); SEARCH_QUERIES.clear(); TOOL_STATS.clear(); TOOL_ERRORS.clear(); STORE.reset_run()
    LLM_STATUS["calls"], LLM_STATUS["errors"] = 0, 0
    t0 = time.perf_counter()
    LOG.info("workflow", "Workflow started")
    init = {"user_request": user_request, "request_valid": False, "quality_attempts": 0, "execution_log": [], "errors": [], "agents_executed": [], "agent_timings": {}, "agent_traces": {},
            "llm_meta": {"available": LLM_STATUS["available"], "model": LLM_STATUS["model"], "reason": LLM_STATUS["reason"], "key_source": LLM_STATUS["key_source"]}}
    try:
        final = app.invoke(init, config={"recursion_limit": 40})
    except Exception as exc:
        LOG.error("workflow", f"Graph execution failed: {type(exc).__name__}: {exc}")
        final = dict(init, errors=[{"node": "graph", "error": f"{type(exc).__name__}: {exc}", "severity": "error"}], execution_log=LOG.flush())
    final = dict(final)
    wall = time.perf_counter() - t0
    report_ok = bool(final.get("report_path")) and Path(final["report_path"]).exists()
    errs = [e for e in final.get("errors", []) if e.get("severity", "error") == "error"]
    warns = [e for e in final.get("errors", []) if e.get("severity") == "warning"]
    final["status"] = "failed" if not report_ok else ("completed_with_warnings" if (errs or warns) else "completed")
    LOG.info("workflow", f"Workflow {'completed' if report_ok else 'finished WITHOUT a report'} - status={final['status']} in {wall:.1f}s")
    final["execution_log"] = list(final.get("execution_log", [])) + LOG.flush()
    final["wall_time_sec"] = round(wall, 2)
    final["tool_timings"] = {k: {"calls": len(v), "total_sec": round(sum(v), 3), "max_sec": round(max(v), 3)} for k, v in TOOL_STATS.items()}
    final["artifacts"] = dict(final.get("artifacts", {}), **finalize_outputs(final))
    return final

FINAL_STATE = run_workflow(USER_REQUEST)
print("\nWORKFLOW STATUS:", FINAL_STATE["status"], f"| wall time {FINAL_STATE['wall_time_sec']}s")


# %% Cell 51 (markdown)
# # # 23. Testing
# # The tests below exercise the major tools with assertions. They are intentionally runnable without Groq; live API and LLM behavior is recorded as an optional capability.

# %% Cell 52 (code)
def run_test(name: str, fn: Callable[[], Any]) -> dict:
    started = time.perf_counter()
    try:
        result = fn()
        return {"Test": name, "Status": "PASS", "Details": str(result)[:220], "Execution Time": round(time.perf_counter()-started, 3)}
    except Exception as exc:
        return {"Test": name, "Status": "FAIL", "Details": f"{type(exc).__name__}: {exc}", "Execution Time": round(time.perf_counter()-started, 3)}

TEST_RESULTS = []
def _fixture_df() -> pd.DataFrame:
    return pd.DataFrame({
        "InvoiceNo": ["10001","10001","10002","C10003","10004"],
        "StockCode": ["10001","10002","10001","10001","POST"],
        "Description": ["A","B","A","A","Shipping"],
        "Quantity": [2,1,3,-1,1], "InvoiceDate": pd.to_datetime(["2025-01-01"]*5),
        "UnitPrice": [10,20,10,10,5], "CustomerID": [1,1,2,2,None], "Country": ["UK"]*5})

TEST_RESULTS.append(run_test("Dataset loading", lambda: (load_sales_data(DATA_PATHS["sales"]) if DATA_PATHS.get("sales") else "skipped: file unavailable")))
TEST_RESULTS.append(run_test("Data quality validation", lambda: validate_sales_data(_fixture_df())))
TEST_RESULTS.append(run_test("KPI calculations", lambda: calculate_kpis(engineer_features(clean_sales_data(_fixture_df())[0]))))
TEST_RESULTS.append(run_test("Top products", lambda: get_top_products(engineer_features(clean_sales_data(_fixture_df())[0]))))
TEST_RESULTS.append(run_test("Anomaly detection", lambda: detect_anomalies(engineer_features(clean_sales_data(_fixture_df())[0]))))
TEST_RESULTS.append(run_test("PDF retrieval", lambda: search_annual_report("revenue", 2) if DATA_PATHS.get("report") else "skipped: file unavailable"))
TEST_RESULTS.append(run_test("Weather API/fallback", lambda: get_weather_data(mode="historical")))
TEST_RESULTS.append(run_test("Report generation", lambda: Path(FINAL_STATE.get("report_path", "")).exists() if FINAL_STATE.get("report_path") else "skipped: end-to-end data unavailable"))
TEST_RESULTS.append(run_test("Email summary generation", lambda: generate_email_summary(FINAL_STATE, FINAL_STATE.get("report_path", OUTPUT_DIR/"multi_agent_business_report.pdf"))))
TEST_RESULTS.append(run_test("End-to-end workflow", lambda: FINAL_STATE.get("status", "unknown")))
TEST_RESULTS_DF = pd.DataFrame(TEST_RESULTS)
display(TEST_RESULTS_DF)


# %% Cell 53 (markdown)
# # # 24. Evaluation
# # The evaluation reports only metrics measured during this notebook run: workflow status, wall time, per-agent timings, tool timings, test pass rate, and whether the PDF artifact exists. No performance values are invented.

# %% Cell 54 (code)
EVALUATION = {
    "reliability": {"workflow_status": FINAL_STATE.get("status"), "report_exists": bool(FINAL_STATE.get("report_path") and Path(FINAL_STATE["report_path"]).exists()), "test_pass_rate_pct": round(100 * float((TEST_RESULTS_DF["Status"] == "PASS").mean()), 1) if len(TEST_RESULTS_DF) else 0},
    "accuracy": {"deterministic_kpis": "Pandas tools used", "retrieval_grounding": FINAL_STATE.get("annual_report_summary", {}).get("mode", "unavailable"), "evidence_count": len(FINAL_STATE.get("annual_report_evidence", []))},
    "efficiency": {"workflow_wall_time_sec": FINAL_STATE.get("wall_time_sec"), "agent_timings_sec": FINAL_STATE.get("agent_timings", {}), "tool_timings": FINAL_STATE.get("tool_timings", {})},
}
print(json.dumps(jsonable(EVALUATION), indent=2))


# %% Cell 55 (markdown)
# # # 25. Limitations
# # - The weather coordinates are Berlin and should be treated as an exploratory proxy for a predominantly UK customer base, not causal evidence.
# # - The Target annual report and the Online Retail dataset are separate business contexts; annual-report evidence is contextual rather than a direct explanation of the retail dataset.
# # - Groq, live weather, and PDF extraction can be unavailable; deterministic fallbacks keep the workflow usable and record the limitation.
# # - Statistical anomalies are unusual observations, not evidence of fraud.

# %% Cell 56 (markdown)
# # # 26. Future Improvements
# # Production extensions could add a persistent vector store, stronger entity alignment between sources, scheduled monitoring, data contracts, a feature store, model evaluation datasets, role-based access, and a managed notification provider.

# %% Cell 57 (markdown)
# # # 27. Requirements
# # Create a real requirements file in the configured output directory and display it.

# %% Cell 58 (code)
REQUIREMENTS_TEXT = """pandas>=2.0\nnumpy>=1.24\nopenpyxl>=3.1\nmatplotlib>=3.7\nscikit-learn>=1.2\npypdf>=4.0\nreportlab>=4.0\nrequests>=2.31\npydantic>=2.0\nlanggraph>=0.2\nlangchain-core>=0.3\nlangchain-groq>=0.2\n"""
REQUIREMENTS_PATH = OUTPUT_DIR / "requirements.txt"
REQUIREMENTS_PATH.write_text(REQUIREMENTS_TEXT, encoding="utf-8")
print(REQUIREMENTS_PATH)
print(REQUIREMENTS_PATH.read_text())


# %% Cell 59 (markdown)
# # # 28. Final Summary
# # This notebook implements a complete AI Business Analyst system. The Retrieval Agent gathers Excel/PDF/weather sources; the quality and cleaning nodes validate and preserve returns; the Analysis Agent computes deterministic KPIs and anomalies; the Research/Insight Agent retrieves cited annual-report evidence and produces grounded interpretations; the Report Agent creates charts and a ReportLab PDF; and the Action Agent persists JSON, the PDF, and an email-ready summary. LangGraph supplies typed state, nodes, conditional routing, a cleaning loop, logging, and error capture. Groq is optional for LLM reasoning and real tool calling; deterministic fallbacks ensure Kaggle execution remains functional when the key or network is unavailable.

# %% Cell 60 (code)
print("FINAL DEMONSTRATION")
for label, value in [
    ("Workflow Status", FINAL_STATE.get("status")),
    ("Execution Time", FINAL_STATE.get("wall_time_sec")),
    ("Agents Executed", FINAL_STATE.get("agents_executed")),
    ("Generated Report Path", FINAL_STATE.get("report_path")),
    ("Email Summary Path", FINAL_STATE.get("artifacts", {}).get("email_summary_txt"))
]:
    print(f"{label}: {value}")

````

````python
# %% [markdown]
# # 11. PDF Retrieval / RAG
#
# Lightweight, dependency-free-on-GPU RAG over the Target annual report:
#
# ```
# PDF -> page text (pypdf) -> overlapping word chunks (page-aware) -> TF-IDF (1-2 grams) -> cosine similarity -> top-k evidence -> LLM summary
# ```
# TF-IDF was chosen over neural embeddings because it is fast, deterministic, needs no model download and works reliably offline on Kaggle. Each chunk keeps its **PDF page number**, so citations read `Annual Report - Page X` (physical PDF page, which may differ from the printed page number).

# %%
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel

class AnnualReportIndex:
    """Page-aware TF-IDF retrieval index over a PDF."""

    def __init__(self, pdf_path: str | Path, chunk_words: int = 180, overlap: int = 40, min_chunk_words: int = 25) -> None:
        self.pdf_path = Path(pdf_path)
        self.chunk_words, self.overlap, self.min_chunk_words = chunk_words, overlap, min_chunk_words
        self.chunks: list[dict] = []
        self.n_pages = 0
        self.pages_without_text = 0
        self._vectorizer: Optional[TfidfVectorizer] = None
        self._matrix = None

    def build(self) -> "AnnualReportIndex":
        """Extract text, chunk it and fit the TF-IDF index. Raises on unreadable/empty PDFs."""
        if not self.pdf_path.exists():
            raise FileNotFoundError(f"Annual report not found: {self.pdf_path}")
        try:
            from pypdf import PdfReader
            reader = PdfReader(str(self.pdf_path))
            self.n_pages = len(reader.pages)
        except Exception as exc:
            raise ValueError(f"Could not open PDF '{self.pdf_path.name}': {type(exc).__name__}: {exc}") from exc
        stride = max(1, self.chunk_words - self.overlap)
        for page_no, page in enumerate(reader.pages, start=1):
            try:
                text = page.extract_text() or ""
            except Exception:
                text = ""
            words = re.sub(r"\s+", " ", text).strip().split(" ")
            words = [w for w in words if w]
            if len(words) < self.min_chunk_words:
                self.pages_without_text += 1 if not words else 0
                if len(words) >= 10:
                    self.chunks.append({"chunk_id": len(self.chunks), "page": page_no, "text": " ".join(words)})
                continue
            for start in range(0, len(words), stride):
                piece = words[start:start + self.chunk_words]
                if len(piece) >= self.min_chunk_words:
                    self.chunks.append({"chunk_id": len(self.chunks), "page": page_no, "text": " ".join(piece)})
        if not self.chunks:
            raise ValueError("PDF text extraction produced no text (scanned/image-only PDF? OCR would be required).")
        texts = [c["text"] for c in self.chunks]
        self._vectorizer = TfidfVectorizer(stop_words="english", ngram_range=(1, 2), sublinear_tf=True, max_df=0.9)
        try:
            self._matrix = self._vectorizer.fit_transform(texts)
        except ValueError:
            # Very small or repetitive PDFs can prune every term with the
            # default English stop-word/max_df settings. Relax only the
            # vectorizer, never the page-aware chunking or citation logic.
            self._vectorizer = TfidfVectorizer(stop_words=None, ngram_range=(1, 2), sublinear_tf=True, max_df=1.0)
            self._matrix = self._vectorizer.fit_transform(texts)
        return self

    def search(self, query: str, top_k: int = 3, min_score: float = 0.02) -> list[dict]:
        """Return the top_k chunks most similar to the query, each with page, score, text and citation."""
        if not query or not query.strip():
            raise ValueError("Empty search query.")
        if self._vectorizer is None:
            raise RuntimeError("Index not built.")
        sims = linear_kernel(self._vectorizer.transform([query]), self._matrix).ravel()
        order = np.argsort(-sims)[:max(1, int(top_k))]
        return [{"chunk_id": self.chunks[i]["chunk_id"], "page": self.chunks[i]["page"], "score": round(float(sims[i]), 4),
                 "text": self.chunks[i]["text"], "citation": f"Annual Report - Page {self.chunks[i]['page']}"}
                for i in order if sims[i] >= min_score]

def get_rag_index() -> AnnualReportIndex:
    """Build the index lazily (once) from the discovered PDF path."""
    if STORE.rag_index is None:
        if not STORE.paths.get("report"):
            raise FileNotFoundError(f"Annual report PDF not found. Expected: {EXPECTED_NAMES['report']}")
        t0 = time.perf_counter()
        STORE.rag_index = AnnualReportIndex(STORE.paths["report"]).build()
        LOG.info("rag", f"Index built: {STORE.rag_index.n_pages} pages, {len(STORE.rag_index.chunks)} chunks in {time.perf_counter()-t0:.1f}s")
    return STORE.rag_index

@track_tool("search_annual_report")
def search_annual_report(query: str, top_k: int = 3) -> list[dict]:
    """Retrieve the annual-report passages most relevant to `query`. New chunks are added to STORE.evidence."""
    results = get_rag_index().search(query, top_k=top_k)
    known = {e["chunk_id"]: e for e in STORE.evidence}
    for r in results:
        item = dict(r, query=query)
        if r["chunk_id"] not in known or r["score"] > known[r["chunk_id"]]["score"]:
            known[r["chunk_id"]] = item
    STORE.evidence = sorted(known.values(), key=lambda e: -e["score"])
    return results

if STORE.paths.get("report"):
    try:
        _demo = search_annual_report("net sales growth and comparable sales", top_k=3)
        print(f"Indexed {STORE.rag_index.n_pages} pages / {len(STORE.rag_index.chunks)} chunks. Example retrieval:\n")
        for r in _demo:
            print(f"[{r['citation']}]  score={r['score']}\n   {r['text'][:350]}...\n")
        STORE.evidence = []   # demo retrieval is not part of the workflow evidence
    except Exception as exc:
        print("RAG demo failed:", exc)
else:
    print("Annual report PDF not found - RAG skipped.")

# %% [markdown]
# # 12. External API Tools (Open-Meteo)
#
# `get_weather_data()` calls the **live Open-Meteo API** (no key needed):
# * historical daily data from `archive-api.open-meteo.com` (temperature mean/max/min, precipitation),
# * or current conditions from `api.open-meteo.com`.
#
# Robustness: timeout + retries; empty/invalid responses are treated as failures; on any failure the tool **falls back to the local CSV** and says so in its result (`fallback_used=True`).
#
# *Important caveat:* the provided coordinates are Berlin, while the retailer's customers are mostly in the UK. Weather analysis is therefore an **exploratory proxy**, not evidence of causation (this is repeated in the report).

# %%
import requests

OPEN_METEO_ARCHIVE = "https://archive-api.open-meteo.com/v1/archive"
OPEN_METEO_FORECAST = "https://api.open-meteo.com/v1/forecast"

def _parse_open_meteo_csv(path: str | Path) -> pd.DataFrame:
    """Parse an Open-Meteo CSV export (metadata header + data table) into a DAILY DataFrame.

    Output columns (when available): date, temp_mean, temp_max, temp_min, precip_sum. Hourly data is aggregated to daily.
    """
    lines = Path(path).read_text(encoding="utf-8-sig", errors="replace").splitlines()
    idx = next((i for i, l in enumerate(lines) if l.strip().lower().startswith("time")), None)
    if idx is None:
        raise ValueError("Open-Meteo CSV has no 'time' header row.")
    raw = pd.read_csv(io.StringIO("\n".join(lines[idx:])))
    tcol = raw.columns[0]
    raw[tcol] = pd.to_datetime(raw[tcol], errors="coerce")
    raw = raw.dropna(subset=[tcol])
    if raw.empty:
        raise ValueError("Open-Meteo CSV contains no valid timestamps.")
    mapping: dict[str, str] = {}
    for c in raw.columns[1:]:
        lc = c.lower()
        if "temperature" in lc and "max" in lc: mapping[c] = "temp_max"
        elif "temperature" in lc and "min" in lc: mapping[c] = "temp_min"
        elif "temperature" in lc: mapping.setdefault(c, "temp_mean")
        elif "precip" in lc or "rain" in lc: mapping.setdefault(c, "precip_sum")
    if not mapping:
        raise ValueError(f"No temperature/precipitation columns recognised in {list(raw.columns)}")
    data = pd.DataFrame({"date": raw[tcol].dt.normalize()})
    for src, dst in mapping.items():
        data[dst] = pd.to_numeric(raw[src], errors="coerce")
    g = data.groupby("date")
    daily = pd.DataFrame(index=g.size().index)
    if "temp_mean" in data: daily["temp_mean"] = g["temp_mean"].mean()
    if "temp_max" in data: daily["temp_max"] = g["temp_max"].max()
    elif "temp_mean" in data: daily["temp_max"] = g["temp_mean"].max()
    if "temp_min" in data: daily["temp_min"] = g["temp_min"].min()
    elif "temp_mean" in data: daily["temp_min"] = g["temp_mean"].min()
    if "temp_mean" not in daily and {"temp_max", "temp_min"} <= set(daily.columns):
        daily["temp_mean"] = (daily["temp_max"] + daily["temp_min"]) / 2
    if "precip_sum" in data: daily["precip_sum"] = g["precip_sum"].sum(min_count=1)
    return daily.reset_index()

@track_tool("load_local_weather")
def load_local_weather(path: str | Path) -> pd.DataFrame:
    """Load the local Open-Meteo CSV as a daily DataFrame."""
    if not path or not Path(path).exists():
        raise FileNotFoundError(f"Local weather CSV not found. Expected: {EXPECTED_NAMES['weather']}")
    return _parse_open_meteo_csv(path)

@track_tool("fetch_open_meteo_api")
def fetch_open_meteo_api(start_date: str, end_date: str, latitude: float, longitude: float,
                         mode: str = "historical", timeout: int = 20, retries: int = 2) -> pd.DataFrame | dict:
    """Call the Open-Meteo API. historical -> daily DataFrame; current -> dict. Raises on timeout/HTTP error/empty payload."""
    if mode == "current":
        url, params = OPEN_METEO_FORECAST, {"latitude": latitude, "longitude": longitude, "current_weather": "true", "timezone": "auto"}
    else:
        url = OPEN_METEO_ARCHIVE
        params = {"latitude": latitude, "longitude": longitude, "start_date": start_date, "end_date": end_date, "timezone": "auto",
                  "daily": "temperature_2m_mean,temperature_2m_max,temperature_2m_min,precipitation_sum"}
    last_exc: Optional[Exception] = None
    for attempt in range(retries + 1):
        try:
            resp = requests.get(url, params=params, timeout=timeout)
            resp.raise_for_status()
            payload = resp.json()
            if mode == "current":
                if not payload.get("current_weather"):
                    raise ValueError("Empty current_weather payload")
                return payload["current_weather"]
            daily = payload.get("daily") or {}
            if not daily.get("time"):
                raise ValueError("Empty daily payload from Open-Meteo")
            df = pd.DataFrame({"date": pd.to_datetime(daily["time"]), "temp_mean": daily.get("temperature_2m_mean"),
                               "temp_max": daily.get("temperature_2m_max"), "temp_min": daily.get("temperature_2m_min"),
                               "precip_sum": daily.get("precipitation_sum")})
            return df
        except Exception as exc:
            last_exc = exc
            if attempt < retries:
                time.sleep(1.5 * (attempt + 1))
    raise RuntimeError(f"Open-Meteo API failed after {retries + 1} attempts: {type(last_exc).__name__}: {last_exc}")

def _weather_summary(df: pd.DataFrame) -> dict:
    out = {"rows": int(len(df)), "date_start": str(df["date"].min().date()), "date_end": str(df["date"].max().date())}
    if "temp_mean" in df: out.update(temp_mean_avg=round(float(df["temp_mean"].mean()), 2))
    if "temp_min" in df: out.update(temp_min=round(float(df["temp_min"].min()), 2))
    if "temp_max" in df: out.update(temp_max=round(float(df["temp_max"].max()), 2))
    if "precip_sum" in df: out.update(precip_total_mm=round(float(df["precip_sum"].sum()), 1), rainy_days=int((df["precip_sum"] > 0.1).sum()))
    return out

@track_tool("get_weather_data")
def get_weather_data(start_date: str = "", end_date: str = "", latitude: float = WEATHER_LAT, longitude: float = WEATHER_LON,
                     mode: str = "historical", store: bool = True) -> dict:
    """Retrieve weather via the live Open-Meteo API, falling back to the local CSV if the API is unavailable.

    Args:
        start_date/end_date: 'YYYY-MM-DD'. Empty = the sales-data date range. (Ignored for mode='current'.)
        latitude/longitude: location (defaults to the coordinates of the provided CSV).
        mode: 'historical' (daily archive) or 'current' (latest conditions).
        store: keep the daily DataFrame in STORE.weather_df for later correlation analysis.
    """
    mode = (mode or "historical").lower()
    if mode not in {"historical", "current"}:
        raise ValueError("mode must be 'historical' or 'current'")
    latitude = WEATHER_LAT if latitude is None else float(latitude)
    longitude = WEATHER_LON if longitude is None else float(longitude)
    if not (-90 <= latitude <= 90 and -180 <= longitude <= 180):
        raise ValueError("Invalid coordinates")

    if mode == "current":
        try:
            cur = fetch_open_meteo_api("", "", latitude, longitude, mode="current")
            return jsonable({"source": "open-meteo API (current)", "api_available": True, "fallback_used": False, "mode": mode,
                             "latitude": latitude, "longitude": longitude, "current": cur})
        except Exception as exc:
            LOG.warning("get_weather_data", f"Current-weather API unavailable ({exc}); using latest row of local CSV (NOT current).")
            local = load_local_weather(STORE.paths.get("weather"))
            return jsonable({"source": "local CSV (API fallback, latest available day - not current)", "api_available": False,
                             "fallback_used": True, "mode": mode, "api_error": str(exc)[:300], "latest_local_day": _weather_summary(local.tail(1))})

    if not start_date or not end_date:
        ref = STORE.clean_df if STORE.clean_df is not None else STORE.raw_df
        if ref is None:
            raise ValueError("No dates given and no sales data loaded to derive them from.")
        d = pd.to_datetime(ref["InvoiceDate"], errors="coerce")
        start_date, end_date = str(d.min().date()), str(d.max().date())
    s, e = pd.to_datetime(start_date), pd.to_datetime(end_date)
    if pd.isna(s) or pd.isna(e) or s > e:
        raise ValueError(f"Invalid date range: {start_date} -> {end_date}")
    latest_allowed = pd.Timestamp.today().normalize() - pd.Timedelta(days=5)   # archive lags a few days
    e = min(e, latest_allowed)
    s_str, e_str = str(s.date()), str(e.date())

    api_error, df, source = None, None, None
    try:
        df = fetch_open_meteo_api(s_str, e_str, latitude, longitude, mode="historical")
        source = "open-meteo API (historical archive)"
    except Exception as exc:
        api_error = str(exc)
        LOG.warning("get_weather_data", f"API unavailable -> falling back to local CSV. Reason: {api_error[:200]}")
        local = load_local_weather(STORE.paths.get("weather"))
        df = local[(local["date"] >= s) & (local["date"] <= e)]
        source = "local CSV (API fallback)"
        if df.empty:
            raise ValueError("Local CSV has no rows in the requested range either.") from exc
    df = df.dropna(how="all", subset=[c for c in df.columns if c != "date"]).reset_index(drop=True)
    if df.empty:
        raise ValueError("Weather data is empty after cleaning.")
    summary = _weather_summary(df)
    result = {"source": source, "api_available": api_error is None, "fallback_used": api_error is not None, "mode": mode,
              "latitude": latitude, "longitude": longitude, "api_error": api_error[:300] if api_error else None, **summary}
    if store:
        STORE.weather_df, STORE.weather_meta = df, result
    return jsonable(result)

# %% [markdown]
# # 13. LLM Configuration (Groq)
#
# `init_llm()` creates a `ChatGroq` client with `GROQ_MODEL`, runs a tiny connectivity test and **automatically falls back** to `GROQ_FALLBACK_MODEL`. If no key / no model works, `LLM` stays `None` and every LLM-dependent step uses its deterministic fallback.
#
# | Component | Needs Groq? | Fallback when Groq is unavailable |
# |---|---|---|
# | Retrieval Agent tool selection | Optional | deterministic tool sequence |
# | Analysis Agent tool selection | Optional | deterministic tool sequence |
# | Annual-report summary | **Yes** | extractive snippets with page citations |
# | Insights / interpretations | **Yes** | rule-based calculated results and recommendations |
# | Executive summary / conclusion | **Yes** | template text built from KPIs |
# | All numbers, charts, PDF, e-mail summary | No | - |

# %%
LLM = None
LLM_STATUS: dict[str, Any] = {"available": False, "model": None, "key_source": _GROQ_KEY_SOURCE, "reason": "not initialised", "calls": 0, "errors": 0}

def init_llm() -> None:
    """Initialise the Groq chat model (preferred model, then fallback model). Never prints the key."""
    global LLM
    LLM = None
    LLM_STATUS.update(available=False, model=None, reason="", calls=0, errors=0)
    if not _GROQ_KEY:
        LLM_STATUS["reason"] = "No GROQ_API_KEY found (Kaggle Secrets / configuration cell)."
        LOG.warning("llm", LLM_STATUS["reason"] + " Running in deterministic mode.")
        return
    try:
        from langchain_groq import ChatGroq
    except Exception as exc:
        LLM_STATUS["reason"] = f"langchain_groq import failed: {exc}"
        LOG.warning("llm", LLM_STATUS["reason"]); return
    reasons = []
    for model in [GROQ_MODEL, GROQ_FALLBACK_MODEL]:
        try:
            candidate = ChatGroq(model=model, api_key=_GROQ_KEY, temperature=0, max_retries=2, timeout=60)
            reply = candidate.invoke("Reply with the single word: OK")
            if not str(getattr(reply, "content", "")).strip():
                raise ValueError("empty reply")
            LLM = candidate
            LLM_STATUS.update(available=True, model=model, reason="ok")
            LOG.info("llm", f"Groq model ready: {model}" + ("" if model == GROQ_MODEL else f" (fallback; preferred {GROQ_MODEL} unavailable)"))
            return
        except Exception as exc:
            reasons.append(f"{model}: {type(exc).__name__}: {str(exc)[:160]}")
            LOG.warning("llm", f"Model {model} unusable - {type(exc).__name__}")
    LLM_STATUS["reason"] = " | ".join(reasons)
    LOG.warning("llm", "No Groq model available. Running in deterministic mode.")

init_llm()
print("LLM status:", {k: v for k, v in LLM_STATUS.items()})

# %%
def llm_invoke(runnable: Any, messages: Any, retries: int = 3) -> Any:
    """Invoke an LLM runnable with retries for rate limits/timeouts/5xx (other errors are raised immediately)."""
    last: Optional[Exception] = None
    for attempt in range(retries):
        try:
            LLM_STATUS["calls"] += 1
            return runnable.invoke(messages)
        except Exception as exc:
            LLM_STATUS["errors"] += 1
            last = exc
            text = f"{type(exc).__name__} {exc}".lower()
            if any(k in text for k in ("rate", "429", "timeout", "timed out", "503", "502", "500", "overloaded", "connection")):
                time.sleep(min(25, 6 * (attempt + 1)))
                continue
            raise
    raise last  # type: ignore[misc]

def extract_json(text: str) -> Any:
    """Extract the first JSON object/array from an LLM reply (handles markdown fences and surrounding prose)."""
    if not isinstance(text, str) or not text.strip():
        raise ValueError("empty LLM reply")
    cleaned = re.sub(r"```(?:json)?", "", text).strip()
    for opener, closer in (("{", "}"), ("[", "]")):
        a, b = cleaned.find(opener), cleaned.rfind(closer)
        if a != -1 and b > a:
            try:
                return json.loads(cleaned[a:b + 1])
            except json.JSONDecodeError:
                continue
    raise ValueError("no valid JSON found in LLM reply")

def llm_text(system: str, user: str) -> str:
    """Single LLM call returning text. Raises if the LLM is unavailable."""
    if LLM is None:
        raise RuntimeError("LLM unavailable")
    from langchain_core.messages import SystemMessage, HumanMessage
    reply = llm_invoke(LLM, [SystemMessage(content=system), HumanMessage(content=user)])
    return str(reply.content).strip()

def llm_json(system: str, user: str, attempts: int = 2) -> Any:
    """LLM call that must return JSON; re-asks once if the reply cannot be parsed."""
    last: Optional[Exception] = None
    for i in range(attempts):
        text = llm_text(system, user + ("" if i == 0 else "\n\nYour previous reply was not valid JSON. Reply with ONLY one valid JSON object."))
        try:
            return extract_json(text)
        except Exception as exc:
            last = exc
    raise ValueError(f"LLM did not return valid JSON: {last}")
````

````python
# %%
# ---------- Report Agent helpers ----------
def build_limitations(state: dict) -> list[str]:
    """Static + run-dependent limitations that are printed in the report."""
    ds = state.get("data_summary", {}).get("sources", {}); lm = state.get("llm_meta", {}); an = state.get("analysis_results", {})
    lim = ["The annual report is Target Corporation's, while the sales dataset is a different (UK online) retailer: report passages are external market context, not the retailer's own performance.",
           "Weather default location is Berlin (52.52N, 13.42E) while most customers are in the UK; correlation with sales is exploratory, is not causal, and may be confounded by seasonality.",
           "Currency is assumed to be GBP; the dataset has no currency field.",
           "The data covers roughly one year: seasonality conclusions rest on a single cycle. Partial months are excluded from growth/trend statistics.",
           "Anomalies are purely statistical (IQR k=3 / robust z-score): they flag unusual values, not fraud or errors.",
           "Annual-report retrieval is lexical (TF-IDF): it can miss paraphrases; page numbers are physical PDF pages and may differ from printed page numbers.",
           "Customer analytics covers only rows with a CustomerID (a substantial share of rows has none)."]
    if not lm.get("available"): lim.append("No Groq LLM was available for this run: interpretations, annual-report summary and executive text use deterministic fallbacks. Reason: " + str(lm.get("reason", "n/a"))[:200])
    if ds.get("weather", {}).get("fallback_used"): lim.append("The live Open-Meteo API was unavailable; the local CSV fallback was used.")
    if an.get("anomalies", {}).get("method") == "zscore": lim.append("Z-score method is sensitive to heavy tails in retail data.")
    return lim

def deterministic_exec_summary(digest: dict) -> dict:
    K = digest.get("kpis", {}); TC = digest.get("top_countries", [{}]); TP = digest.get("top_products", [{}]); CU = digest.get("customers", {}); AN = digest.get("anomalies", {})
    parts = []
    if K.get("total_revenue") is not None:
        parts.append(f"Between {K.get('date_start')} and {K.get('date_end')} the business generated GBP {K['total_revenue']:,.0f} in gross sales from {K['total_orders']:,} orders and {K['unique_customers']:,} identified customers (average order value GBP {K['average_order_value']:,.2f}).")
    if TC and TC[0].get("country"): parts.append(f"{TC[0]['country']} accounts for {TC[0]['share_pct']}% of revenue; the top product is '{TP[0].get('name', 'n/a')}'.")
    if digest.get("trend_direction"): parts.append(f"Revenue across complete months is {digest['trend_direction']}" + (f", peaking in {digest['peak_month']['month']}." if digest.get("peak_month") else "."))
    if K.get("return_rate_value_pct") is not None: parts.append(f"Returns are {K['return_rate_value_pct']}% of sales value.")
    if CU.get("top10pct_revenue_share_pct") is not None: parts.append(f"The top 10% of identified customers generate {CU['top10pct_revenue_share_pct']}% of identified revenue.")
    if AN.get("flagged_lines") is not None: parts.append(f"{AN['flagged_lines']:,} product lines were flagged as statistical anomalies for review.")
    concl = ("The analysis combined deterministic Pandas calculations with retrieved annual-report context and weather data. Priorities are retaining high-value customers, "
             "reducing dependence on the main market, understanding returns, and reviewing flagged anomalies. See the recommendations section for details.")
    return {"executive_summary": " ".join(parts) or "Summary unavailable.", "conclusion": concl, "mode": "deterministic"}

def generate_exec_summary(digest: dict) -> dict:
    """LLM-written executive summary/conclusion; falls back to a template when the LLM fails or uses numbers not in the digest."""
    base = deterministic_exec_summary(digest)
    if LLM is None:
        return base
    try:
        out = llm_json("You write executive summaries for a business report. Use ONLY facts and numbers from the DIGEST; do not invent figures. "
                       "Return JSON: {\"executive_summary\": \"110-160 words\", \"conclusion\": \"60-90 words\"}. Plain prose, no markdown.",
                       f"DIGEST:\n{json.dumps(digest, default=str)}")
        es, co = str(out.get("executive_summary", "")).strip(), str(out.get("conclusion", "")).strip()
        bad = unsupported_numbers(es + " " + co, json.dumps(digest, default=str))
        if not es or not co:
            raise ValueError("missing fields")
        if bad:
            LOG.warning("exec_summary", f"LLM text contained numbers not in the digest ({bad[:5]}); using deterministic summary.")
            return base
        return {"executive_summary": es, "conclusion": co, "mode": "llm"}
    except Exception as exc:
        LOG.warning("exec_summary", f"LLM summary failed ({type(exc).__name__}: {str(exc)[:150]}); using deterministic summary.")
        return base

def assemble_report_data(state: dict, exec_summary: dict) -> dict:
    ds = state.get("data_summary", {}).get("sources", {}); lm = state.get("llm_meta", {}); wm = state.get("weather_results", {})
    s, a, w = ds.get("sales", {}), ds.get("annual_report", {}), ds.get("weather", {})
    sources = [
        {"name": "Online Retail (Excel)", "status": "Loaded" if s.get("available") else "Missing", "detail": f"{s.get('file', '')}: {s.get('rows', 0):,} raw rows, {s.get('date_start', '')} to {s.get('date_end', '')}" if s.get("available") else str(s.get("error", ""))},
        {"name": "Target Annual Report (PDF, RAG)", "status": "Indexed" if a.get("available") else "Missing", "detail": f"{a.get('file', '')}: {a.get('pages', 0)} pages, {a.get('chunks', 0)} chunks; {a.get('evidence_chunks', 0)} evidence chunks retrieved" if a.get("available") else str(a.get("error", ""))},
        {"name": "Open-Meteo weather", "status": "Retrieved" if w.get("available") else "Missing", "detail": f"{w.get('source', '')}; fallback used: {w.get('fallback_used')}" if w.get("available") else str(w.get("error", ""))},
        {"name": "Groq LLM", "status": "Used" if lm.get("available") else "Unavailable", "detail": f"model {lm.get('model')}" if lm.get("available") else str(lm.get("reason", ""))[:160]}]
    return {"generated_at": dt.datetime.now().strftime("%Y-%m-%d %H:%M"), "user_request": state.get("user_request", ""), "sources": sources, "data_quality": state.get("data_quality_raw", {}) or state.get("data_quality", {}),
            "cleaning_report": state.get("cleaning_report", {}), "analysis": state.get("analysis_results", {}), "evidence": state.get("annual_report_evidence", []), "ar_summary": state.get("annual_report_summary", {}),
            "weather": wm, "insights": state.get("insights", []), "recommendations": state.get("recommendations", []), "exec_summary": exec_summary, "limitations": build_limitations(state), "llm_meta": lm}

# %%
# ---------- Graph nodes ----------
@agent_node("Request Check", is_agent=False)
def node_request_check(state: dict) -> dict:
    try:
        req = WorkflowRequest(user_request=state.get("user_request", ""))
        LOG.info("Request Check", "Request validated")
        return {"request_valid": True, "user_request": req.user_request}
    except Exception as exc:
        msg = "; ".join(str(exc).splitlines()[:3])
        LOG.error("Request Check", f"Invalid request: {msg}")
        return {"request_valid": False, "errors": [{"node": "Request Check", "error": f"Invalid request: {msg}", "severity": "error"}]}

@agent_node("Data Retrieval Agent")
def node_retrieval(state: dict) -> dict:
    task = (f"User request: {state['user_request']}\nSources: Online Retail sales (Excel), Target annual report (PDF), Open-Meteo weather. "
            "Retrieve what is needed for this request.")
    trace = build_retrieval_agent().run(task)
    # sanity: if the LLM chose a weather date range that does not cover the sales period, redo with the sales range
    meta, raw = STORE.weather_meta, STORE.raw_df
    if raw is not None and meta and any(c["tool"] == "get_weather_data" for c in trace["llm_tool_calls"]):
        d = pd.to_datetime(raw["InvoiceDate"], errors="coerce")
        if not (meta.get("date_start", "9") <= str(d.min().date()) and meta.get("date_end", "0") >= str(d.max().date())) and str(meta.get("source", "")).startswith("open-meteo"):
            LOG.warning("Data Retrieval Agent", "LLM-chosen weather range does not cover the sales period; re-fetching with the sales range.")
            get_weather_data_tool(); trace["deterministic_fallback"].append("get_weather_data (range corrected)")
    s_res, w_res = RESULTS.get("load_sales_data", {}), RESULTS.get("get_weather_data", {})
    idx = STORE.rag_index
    summary = {"sources": {
        "sales": {"available": STORE.raw_df is not None, "file": Path(STORE.paths["sales"]).name if STORE.paths.get("sales") else None, "rows": s_res.get("rows"), "date_start": s_res.get("date_start"), "date_end": s_res.get("date_end"), "error": RESULT_ERRORS.get("load_sales_data")},
        "annual_report": {"available": idx is not None, "file": Path(STORE.paths["report"]).name if STORE.paths.get("report") else None, "pages": getattr(idx, "n_pages", None), "chunks": len(idx.chunks) if idx else None,
                          "queries_run": list(SEARCH_QUERIES), "evidence_chunks": len(STORE.evidence), "error": RESULT_ERRORS.get("search_annual_report")},
        "weather": {"available": STORE.weather_df is not None, "source": w_res.get("source"), "fallback_used": w_res.get("fallback_used"), "rows": w_res.get("rows"), "error": RESULT_ERRORS.get("get_weather_data")}}}
    errs = [{"node": "Data Retrieval Agent", "tool": t["tool"], "error": t["error"], "severity": "error" if t["tool"] == "load_sales_data" else "warning"} for t in trace["tool_errors"]]
    if trace["llm_error"]:
        errs.append({"node": "Data Retrieval Agent", "tool": "llm", "error": trace["llm_error"], "severity": "warning"})
    LOG.info("Data Retrieval Agent", f"Availability: sales={summary['sources']['sales']['available']}, annual_report={summary['sources']['annual_report']['available']}, weather={summary['sources']['weather']['available']}")
    return {"data_summary": summary, "weather_results": w_res if isinstance(w_res, dict) else {}, "annual_report_evidence": jsonable(STORE.evidence), "agent_traces": {"Data Retrieval Agent": trace}, "errors": errs,
            "llm_meta": {"available": LLM_STATUS["available"], "model": LLM_STATUS["model"], "reason": LLM_STATUS["reason"], "key_source": LLM_STATUS["key_source"]}}

@agent_node("Data Quality Check", is_agent=False)
def node_quality_check(state: dict) -> dict:
    cleaned = STORE.clean_df is not None
    df = STORE.clean_df if cleaned else STORE.raw_df
    if df is None:
        rep = DataQualityReport(rows=0, columns=0, blocking_issues=["Sales data could not be loaded"], acceptable=False).model_dump()
    else:
        rep = validate_sales_data(df, is_cleaned=cleaned)
    upd: dict[str, Any] = {"data_quality": rep}
    if not cleaned:
        upd["data_quality_raw"] = rep                  # keep the raw-data diagnosis for the report
    LOG.info("Data Quality Check", f"{'post-cleaning' if cleaned else 'raw'} check: acceptable={rep['acceptable']}, defect_rows={rep.get('defect_rows')}, blocking={rep['blocking_issues']}")
    return upd

def route_after_quality(state: dict) -> str:
    q = state.get("data_quality", {})
    # The analysis tools require engineered features; always pass through the
    # cleaning node once per run, even when the raw data has no defects.
    if not q.get("is_cleaned"): return "clean"
    if q.get("blocking_issues"): return "abort"
    return "analysis"          # cleaned, residual non-blocking issues only

@agent_node("Cleaning", is_agent=False)
def node_cleaning(state: dict) -> dict:
    clean, rep = clean_sales_data(STORE.raw_df)
    STORE.clean_df, STORE.cleaning_report = engineer_features(clean), rep
    LOG.info("Cleaning", f"{rep['rows_before']:,} -> {rep['rows_after']:,} rows; returns kept: {rep['returns_kept']:,}; features added")
    return {"cleaning_report": rep, "quality_attempts": state.get("quality_attempts", 0) + 1}

@agent_node("Data Analysis Agent")
def node_analysis(state: dict) -> dict:
    trace = build_analysis_agent().run("Run the complete business analysis for the request: " + state["user_request"])
    keys = {"kpis": "calculate_kpis", "trends": "analyze_sales_trends", "top_products": "get_top_products", "top_countries": "get_top_countries",
            "customers": "analyze_customers", "anomalies": "detect_anomalies", "weather_impact": "analyze_weather_impact"}
    analysis = {k: RESULTS.get(t) for k, t in keys.items() if RESULTS.get(t) and "error" not in RESULTS.get(t, {})}
    errs = [{"node": "Data Analysis Agent", "tool": e["tool"], "error": e["error"], "severity": "error"} for e in trace["tool_errors"]]
    if trace["llm_error"]: errs.append({"node": "Data Analysis Agent", "tool": "llm", "error": trace["llm_error"], "severity": "warning"})
    if "kpis" in analysis: LOG.info("Data Analysis Agent", f"KPI calculation completed: revenue={analysis['kpis']['total_revenue']:,.0f}, orders={analysis['kpis']['total_orders']:,}")
    return {"analysis_results": analysis, "agent_traces": {"Data Analysis Agent": trace}, "errors": errs}

@agent_node("Research / Insight Agent")
def node_insight(state: dict) -> dict:
    analysis, weather = state.get("analysis_results", {}), state.get("weather_results", {})
    digest = build_digest(analysis, weather)
    agent = ToolAgent(name="Research / Insight Agent", tools=INSIGHT_TOOLS, result_chars=1000,
                      system_prompt=("You are the Research/Insight Agent. The sales analysis is finished (digest below). Use search_annual_report 2-3 times with focused queries "
                                     "(e.g. revenue growth, customers/loyalty, seasonality, supply chain) to find annual-report context that relates to these results. Then reply DONE."),
                      required={"search_annual_report": lambda: _run_searches(INSIGHT_QUERIES)})
    trace = agent.run(f"Request: {state['user_request']}\nDIGEST: {json.dumps(digest, default=str)[:1800]}")
    if STORE.rag_index is not None and len(STORE.evidence) < 6:
        _run_searches(INSIGHT_QUERIES)
    evidence = jsonable(STORE.evidence[:8])
    summary = summarize_annual_report(evidence)
    LOG.info("Research / Insight Agent", f"Annual report retrieval completed: {len(evidence)} evidence chunks, summary mode={summary['mode']}")
    det_ins, det_recs = build_deterministic_insights(analysis, weather)
    llm_items, gmeta = generate_llm_insights(digest, evidence, summary)
    insights = det_ins + [i for i in llm_items if i["category"] != "RECOMMENDATION"]
    recs = [i for i in llm_items if i["category"] == "RECOMMENDATION"] + det_recs
    errs = []
    if LLM is None: errs.append({"node": "Research / Insight Agent", "tool": "llm", "error": "LLM unavailable: " + str(LLM_STATUS["reason"])[:200] + " (deterministic insights only)", "severity": "warning"})
    elif gmeta.get("error"): errs.append({"node": "Research / Insight Agent", "tool": "llm", "error": gmeta["error"], "severity": "warning"})
    if trace["llm_error"]: errs.append({"node": "Research / Insight Agent", "tool": "llm", "error": trace["llm_error"], "severity": "warning"})
    llm_meta = dict(state.get("llm_meta", {}), grounding=gmeta, annual_report_summary_mode=summary["mode"])
    LOG.info("Research / Insight Agent", f"Insight agent completed: {len(insights)} insights, {len(recs)} recommendations")
    return {"annual_report_evidence": evidence, "annual_report_summary": summary, "insights": insights, "recommendations": recs, "llm_meta": llm_meta, "agent_traces": {"Research / Insight Agent": trace}, "errors": errs}

@agent_node("Report Agent")
def node_report(state: dict) -> dict:
    errs, charts = [], {}
    try:
        charts = generate_charts(STORE.clean_df, state.get("analysis_results", {}), OUTPUT_DIR / "charts", STORE.weather_df)
        LOG.info("Report Agent", f"Charts generated: {sorted(charts)}")
    except Exception as exc:
        errs.append({"node": "Report Agent", "tool": "generate_charts", "error": f"{type(exc).__name__}: {exc}", "severity": "warning"})
    es = generate_exec_summary(build_digest(state.get("analysis_results", {}), state.get("weather_results", {})))
    rd = assemble_report_data(state, es)
    path = generate_pdf_report(rd, charts, OUTPUT_DIR / "multi_agent_business_report.pdf")
    LOG.info("Report Agent", f"PDF generated: {path}")
    return {"report_path": str(path), "chart_paths": charts, "exec_summary": es, "errors": errs}

@agent_node("Action Agent")
def node_action(state: dict) -> dict:
    rp = Path(state.get("report_path") or "")
    if not rp.exists() or rp.stat().st_size == 0:
        raise FileNotFoundError("Report PDF is missing - nothing to send/save.")
    from pypdf import PdfReader
    n_pages = len(PdfReader(str(rp)).pages)
    analysis_path = OUTPUT_DIR / "analysis_results.json"
    analysis_path.write_text(json.dumps(jsonable(state.get("analysis_results", {})), indent=2), encoding="utf-8")
    rd = assemble_report_data(state, state.get("exec_summary", {}))
    email = generate_email_summary(rd, rp, REPORT_RECIPIENT)
    email_path = save_email_summary(email, OUTPUT_DIR / "email_summary.txt")
    delivery = send_email_if_configured(email)
    LOG.info("Action Agent", f"Report verified ({n_pages} pages, {rp.stat().st_size/1024:.0f} KB); email summary saved; sending: {delivery['reason']}")
    return {"email_summary": dict(email, delivery=delivery), "artifacts": {"report_pdf": str(rp), "report_pages": n_pages, "analysis_results_json": str(analysis_path), "email_summary_txt": str(email_path)}}

@agent_node("Abort", is_agent=False)
def node_abort(state: dict) -> dict:
    LOG.error("Abort", "Workflow stopped: " + "; ".join(state.get("data_quality", {}).get("blocking_issues", []) or ["unknown reason"]))
    return {"errors": [{"node": "Abort", "error": "Workflow aborted due to blocking data issues: " + "; ".join(state.get("data_quality", {}).get("blocking_issues", [])), "severity": "error"}]}

# %% [markdown]
# # 17. LangGraph Orchestration
#
# ```
# START -> request_check --(invalid)--> END
#              | valid
#              v
#          retrieval_agent -> quality_check --(acceptable)-------------> analysis_agent
#                                  |  (needs cleaning)                        |
#                                  +--> cleaning --> quality_check (re-check) v
#                                  |  (blocking issue)                  insight_agent -> report_agent -> action_agent -> END
#                                  +--> abort --> END
# ```
# Features used: typed **state**, **nodes**, **edges**, **conditional edges** (request validity, data quality), a **loop** (clean -> re-check), error handling inside every node, and structured logging.

# %%
from langgraph.graph import StateGraph, START, END

def route_after_request(state: dict) -> str:
    return "retrieval_agent" if state.get("request_valid") else "end"

def build_workflow():
    g = StateGraph(AgentState)
    g.add_node("request_check", node_request_check)
    g.add_node("retrieval_agent", node_retrieval)
    g.add_node("quality_check", node_quality_check)
    g.add_node("cleaning", node_cleaning)
    g.add_node("analysis_agent", node_analysis)
    g.add_node("insight_agent", node_insight)
    g.add_node("report_agent", node_report)
    g.add_node("action_agent", node_action)
    g.add_node("abort", node_abort)
    g.add_edge(START, "request_check")
    g.add_conditional_edges("request_check", route_after_request, {"retrieval_agent": "retrieval_agent", "end": END})
    g.add_edge("retrieval_agent", "quality_check")
    g.add_conditional_edges("quality_check", route_after_quality, {"analysis": "analysis_agent", "clean": "cleaning", "abort": "abort"})
    g.add_edge("cleaning", "quality_check")
    g.add_edge("analysis_agent", "insight_agent")
    g.add_edge("insight_agent", "report_agent")
    g.add_edge("report_agent", "action_agent")
    g.add_edge("action_agent", END)
    g.add_edge("abort", END)
    return g.compile()

WORKFLOW_APP = build_workflow()
print("LangGraph workflow compiled.")
try:
    print("\nMermaid definition of the compiled graph:\n")
    print(WORKFLOW_APP.get_graph().draw_mermaid())
except Exception as exc:
    print("(Mermaid export not available in this langgraph version:", type(exc).__name__, ")")

# %%
from matplotlib.patches import FancyBboxPatch

def draw_architecture(path: Path) -> Path:
    """Draw the system architecture with Matplotlib and save it as PNG."""
    fig, ax = plt.subplots(figsize=(13, 9)); ax.set_xlim(0, 1); ax.set_ylim(0, 1); ax.axis("off"); ax.grid(False)
    def box(x, y, w, h, text, fc, tc="white", fs=9):
        ax.add_patch(FancyBboxPatch((x - w / 2, y - h / 2), w, h, boxstyle="round,pad=0.008", fc=fc, ec="#1F3A5F", lw=1.2))
        ax.text(x, y, text, ha="center", va="center", color=tc, fontsize=fs, fontweight="bold")
    def arrow(x1, y1, x2, y2, style="-|>", ls="-", color="#1F3A5F"):
        ax.annotate("", xy=(x2, y2), xytext=(x1, y1), arrowprops=dict(arrowstyle=style, lw=1.3, ls=ls, color=color))
    cx = 0.45
    ys = {"user": 0.94, "orch": 0.85, "ret": 0.75, "dq": 0.65, "ana": 0.54, "ins": 0.44, "rep": 0.34, "act": 0.24, "out": 0.12}
    box(cx, ys["user"], 0.26, 0.05, "User business request", "#555555")
    box(cx, ys["orch"], 0.36, 0.06, "Orchestrator (LangGraph state machine)\nrequest check | routing | logging | error handling", "#1F3A5F", fs=8.5)
    box(cx, ys["ret"], 0.30, 0.055, "Data Retrieval Agent\n(Excel | PDF RAG | Weather API)", "#2A9D8F", fs=8.5)
    box(cx, ys["dq"], 0.30, 0.055, "Data Quality Validation\n(conditional routing)", "#E9C46A", tc="#1F3A5F", fs=8.5)
    box(0.12, ys["dq"], 0.17, 0.05, "Cleaning\n(keep returns)", "#F4A261", tc="#1F3A5F", fs=8.5)
    box(cx, ys["ana"], 0.30, 0.055, "Data Analysis Agent\n(Pandas tools: KPIs, trends, anomalies)", "#2A9D8F", fs=8.5)
    box(cx, ys["ins"], 0.30, 0.055, "Research / Insight Agent\n(RAG evidence + interpretation)", "#2A9D8F", fs=8.5)
    box(cx, ys["rep"], 0.30, 0.055, "Report Agent\n(charts + ReportLab PDF)", "#2A9D8F", fs=8.5)
    box(cx, ys["act"], 0.30, 0.055, "Action Agent\n(save files + e-mail summary)", "#2A9D8F", fs=8.5)
    box(cx, ys["out"], 0.36, 0.06, "Outputs: PDF report | JSON results | execution log | e-mail summary", "#264653", fs=8.5)
    order = ["user", "orch", "ret", "dq", "ana", "ins", "rep", "act", "out"]
    for a, b in zip(order[:-1], order[1:]):
        arrow(cx, ys[a] - 0.03, cx, ys[b] + 0.03)
    arrow(cx - 0.15, ys["dq"], 0.205, ys["dq"], color="#C0392B"); arrow(0.12, ys["dq"] + 0.027, 0.12, ys["dq"] + 0.06, style="-", color="#C0392B")
    arrow(0.12, ys["dq"] + 0.06, cx - 0.12, ys["dq"] + 0.028, color="#C0392B")
    ax.text(0.135, ys["dq"] - 0.045, "not acceptable -> clean -> re-check", fontsize=7.5, color="#C0392B")
    ax.text(cx + 0.02, ys["dq"] - 0.052, "acceptable", fontsize=7.5, color="#1F3A5F")
    src_x = 0.86
    for y, t, c in [(0.82, "Online Retail\n(Excel)", "#8D99AE"), (0.72, "Target Annual Report\n(PDF)", "#8D99AE"), (0.62, "Open-Meteo\n(API + CSV fallback)", "#8D99AE")]:
        box(src_x, y, 0.22, 0.06, t, c, fs=8.5); arrow(src_x - 0.11, y, cx + 0.15, ys["ret"], color="#555555")
    box(src_x, 0.42, 0.22, 0.09, "Groq LLM\n(tool calling, summaries,\ninterpretation)", "#6D597A", fs=8.5)
    for k in ("ret", "ana", "ins", "rep"):
        arrow(src_x - 0.11, 0.42, cx + 0.15, ys[k], ls="--", color="#6D597A")
    ax.text(0.5, 0.01, "Dashed = LLM calls (optional; deterministic fallback exists). Heavy DataFrames live in a DataStore, not in the graph state.", ha="center", fontsize=8, color="#555555")
    fig.savefig(path, dpi=130, bbox_inches="tight"); plt.close(fig)
    return path

ARCH_PATH = draw_architecture(OUTPUT_DIR / "architecture_diagram.png")
display(Image(filename=str(ARCH_PATH)))

# %% [markdown]
# # 18. Workflow Execution
# `run_workflow()` resets per-run caches, invokes the compiled graph, derives the final status and writes `workflow_results.json` and `execution_log.json`. `analysis_results.json`, `email_summary.txt` and the PDF are written by the Action Agent.

# %%
def finalize_outputs(final: dict) -> dict:
    """Write workflow_results.json and execution_log.json (the API key is never part of the state)."""
    log_path, wf_path = OUTPUT_DIR / "execution_log.json", OUTPUT_DIR / "workflow_results.json"
    log_path.write_text(json.dumps(jsonable(final.get("execution_log", [])), indent=2), encoding="utf-8")
    slim = {k: v for k, v in final.items() if k != "execution_log"}
    wf_path.write_text(json.dumps(jsonable(slim), indent=2), encoding="utf-8")
    return {"workflow_results_json": str(wf_path), "execution_log_json": str(log_path)}

def run_workflow(user_request: str, app: Any = None) -> dict:
    """Execute the whole multi-agent workflow and return the final state (with status, wall time and artifact paths)."""
    app = app or WORKFLOW_APP
    LOG.reset(); RESULTS.clear(); RESULT_ERRORS.clear(); SEARCH_QUERIES.clear(); TOOL_STATS.clear(); TOOL_ERRORS.clear(); STORE.reset_run()
    LLM_STATUS["calls"], LLM_STATUS["errors"] = 0, 0
    t0 = time.perf_counter()
    LOG.info("workflow", "Workflow started")
    init = {"user_request": user_request, "request_valid": False, "quality_attempts": 0, "execution_log": [], "errors": [], "agents_executed": [], "agent_timings": {}, "agent_traces": {},
            "llm_meta": {"available": LLM_STATUS["available"], "model": LLM_STATUS["model"], "reason": LLM_STATUS["reason"], "key_source": LLM_STATUS["key_source"]}}
    try:
        final = app.invoke(init, config={"recursion_limit": 40})
    except Exception as exc:
        LOG.error("workflow", f"Graph execution failed: {type(exc).__name__}: {exc}")
        final = dict(init, errors=[{"node": "graph", "error": f"{type(exc).__name__}: {exc}", "severity": "error"}], execution_log=LOG.flush())
    final = dict(final)
    wall = time.perf_counter() - t0
    report_ok = bool(final.get("report_path")) and Path(final["report_path"]).exists()
    errs = [e for e in final.get("errors", []) if e.get("severity", "error") == "error"]
    warns = [e for e in final.get("errors", []) if e.get("severity") == "warning"]
    final["status"] = "failed" if not report_ok else ("completed_with_warnings" if (errs or warns) else "completed")
    LOG.info("workflow", f"Workflow {'completed' if report_ok else 'finished WITHOUT a report'} - status={final['status']} in {wall:.1f}s")
    final["execution_log"] = list(final.get("execution_log", [])) + LOG.flush()
    final["wall_time_sec"] = round(wall, 2)
    final["tool_timings"] = {k: {"calls": len(v), "total_sec": round(sum(v), 3), "max_sec": round(max(v), 3)} for k, v in TOOL_STATS.items()}
    final["artifacts"] = dict(final.get("artifacts", {}), **finalize_outputs(final))
    return final

FINAL_STATE = run_workflow(USER_REQUEST)
print("\nWORKFLOW STATUS:", FINAL_STATE["status"], f"| wall time {FINAL_STATE['wall_time_sec']}s")
````

````python
# %% [markdown]
# # 14. Agent Tools
#
# ### 14.1 Deterministic analytics (Pandas) — *the LLM never does this arithmetic*
#
# **KPI definitions**
# * **Sales lines** = non-return lines with `Quantity > 0` (after cleaning). **Returns** = lines flagged `IsReturn`.
# * `Total Revenue` = sum of `Revenue` over sales lines (gross, before returns). `Net Revenue` = Total Revenue - value of returns.
# * `Total Orders` = distinct sales `InvoiceNo`. `Unique Customers` = distinct non-missing `CustomerID` in sales.
# * `Unique Products` = distinct product `StockCode` (service/fee codes excluded).
# * `Average Order Value` = Total Revenue / Total Orders. `Average Items per Order` = units sold / Total Orders.
# * `Return Rate (invoices)` = cancellation invoices / (sales invoices + cancellation invoices). `Return Rate (value)` = returns value / Total Revenue.
# * The **last month in the data may be partial**; it is flagged and excluded from growth/trend statistics.

# %%
@track_tool("calculate_kpis")
def calculate_kpis(df: pd.DataFrame) -> dict:
    """Compute the deterministic KPI set (see definitions above). Returns a KPIResult dict."""
    df = _ensure_features(df)
    s, r = _sales(df), _returns(df)
    if s.empty:
        raise ValueError("No sales lines available for KPI calculation.")
    total_revenue = float(s["Revenue"].sum())
    returns_value = float(abs(r["Revenue"].sum()))
    orders, canc = int(s["InvoiceNo"].nunique()), int(r["InvoiceNo"].nunique())
    customers = int(s["CustomerID"].nunique())
    kpi = KPIResult(
        total_revenue=round(total_revenue, 2), returns_value=round(returns_value, 2), net_revenue=round(total_revenue - returns_value, 2),
        total_orders=orders, cancellation_invoices=canc, unique_customers=customers,
        unique_products=int(s.loc[~s["IsNonProduct"], "StockCode"].nunique()),
        average_order_value=round(_safe_div(total_revenue, orders), 2),
        average_items_per_order=round(_safe_div(s["Quantity"].sum(), orders), 2),
        average_revenue_per_customer=round(_safe_div(s.loc[s["CustomerID"].notna(), "Revenue"].sum(), customers), 2),
        return_rate_invoices_pct=round(100 * _safe_div(canc, orders + canc), 2),
        return_rate_value_pct=round(100 * _safe_div(returns_value, total_revenue), 2),
        transaction_lines=int(len(s)), date_start=str(s["InvoiceDate"].min().date()), date_end=str(s["InvoiceDate"].max().date()))
    return jsonable(kpi.model_dump())

@track_tool("analyze_sales_trends")
def analyze_sales_trends(df: pd.DataFrame) -> dict:
    """Monthly revenue/orders/customers/AOV, MoM growth, partial-month flags, trend direction, weekday and hour profiles."""
    s = _sales(_ensure_features(df))
    if s.empty:
        raise ValueError("No sales lines for trend analysis.")
    m = (s.groupby("YearMonth").agg(revenue=("Revenue", "sum"), orders=("InvoiceNo", "nunique"),
                                   customers=("CustomerID", "nunique"), units=("Quantity", "sum"))
           .reset_index().sort_values("YearMonth").reset_index(drop=True))
    m["aov"] = m["revenue"] / m["orders"]
    first, last = s["InvoiceDate"].min(), s["InvoiceDate"].max()
    partial_first = first.normalize() > first.to_period("M").start_time
    partial_last = last.normalize() < (last + pd.offsets.MonthEnd(0)).normalize()
    m["is_partial"] = False
    if partial_first: m.loc[m.index[0], "is_partial"] = True
    if partial_last: m.loc[m.index[-1], "is_partial"] = True
    m["mom_growth_pct"] = m["revenue"].pct_change() * 100
    m.loc[m["is_partial"] | m["is_partial"].shift(1, fill_value=False), "mom_growth_pct"] = np.nan
    complete = m[~m["is_partial"]]
    result: dict[str, Any] = {"monthly": m.round(2).to_dict("records"), "partial_months": m.loc[m["is_partial"], "YearMonth"].tolist()}
    if len(complete) >= 3:
        slope = float(np.polyfit(np.arange(len(complete)), complete["revenue"].to_numpy(), 1)[0])
        rel = slope / float(complete["revenue"].mean())
        result.update(trend_slope_per_month=round(slope, 2), trend_relative_slope=round(rel, 4),
                      trend_direction="increasing" if rel > 0.02 else "decreasing" if rel < -0.02 else "roughly stable")
    if not complete.empty:
        pk, lo = complete.loc[complete["revenue"].idxmax()], complete.loc[complete["revenue"].idxmin()]
        result.update(peak_month={"month": pk["YearMonth"], "revenue": round(float(pk["revenue"]), 2)},
                      lowest_month={"month": lo["YearMonth"], "revenue": round(float(lo["revenue"]), 2)})
    order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
    dow = s.groupby("DayOfWeek")["Revenue"].sum().reindex(order).fillna(0)
    result["revenue_by_weekday"] = {k: round(float(v), 2) for k, v in dow.items()}
    hour = s.groupby("Hour")["Revenue"].sum()
    result["peak_hour"] = int(hour.idxmax())
    result["revenue_by_hour"] = {int(k): round(float(v), 2) for k, v in hour.items()}
    return jsonable(result)

@track_tool("get_top_products")
def get_top_products(df: pd.DataFrame, n: int = 10) -> dict:
    """Top-n products (service/fee lines excluded) by revenue and by units."""
    n = max(1, min(int(n), 50))
    s = _sales(_ensure_features(df))
    non_product_revenue = float(s.loc[s["IsNonProduct"], "Revenue"].sum())
    p = s[~s["IsNonProduct"]]
    if p.empty:
        raise ValueError("No product lines available.")
    desc = p.groupby("StockCode")["Description"].agg(lambda x: x.value_counts().index[0])
    agg = p.groupby("StockCode").agg(revenue=("Revenue", "sum"), quantity=("Quantity", "sum"), orders=("InvoiceNo", "nunique"))
    agg["description"] = desc
    total = float(agg["revenue"].sum())
    def fmt(frame: pd.DataFrame) -> list[dict]:
        out = frame.reset_index()[["StockCode", "description", "revenue", "quantity", "orders"]]
        out["revenue_share_pct"] = 100 * out["revenue"] / total
        return out.round(2).to_dict("records")
    by_rev = agg.sort_values("revenue", ascending=False).head(n)
    by_qty = agg.sort_values("quantity", ascending=False).head(n)
    return jsonable({"by_revenue": fmt(by_rev), "by_quantity": fmt(by_qty), "n_products": int(len(agg)),
                     "top_n_revenue_share_pct": round(100 * float(by_rev["revenue"].sum()) / total, 2),
                     "non_product_lines_revenue": round(non_product_revenue, 2)})

@track_tool("get_top_countries")
def get_top_countries(df: pd.DataFrame, n: int = 10) -> dict:
    """Top-n countries by revenue with orders, customers and revenue share."""
    n = max(1, min(int(n), 50))
    s = _sales(_ensure_features(df))
    agg = s.groupby("Country").agg(revenue=("Revenue", "sum"), orders=("InvoiceNo", "nunique"), customers=("CustomerID", "nunique"))
    total = float(agg["revenue"].sum())
    agg["revenue_share_pct"] = 100 * agg["revenue"] / total
    agg["aov"] = agg["revenue"] / agg["orders"]
    top = agg.sort_values("revenue", ascending=False).head(n).reset_index()
    non_home = agg.drop(index="United Kingdom", errors="ignore").sort_values("revenue", ascending=False)
    return jsonable({"top": top.round(2).to_dict("records"), "n_countries": int(len(agg)),
                     "top_country": str(top.iloc[0]["Country"]), "top_country_share_pct": round(float(top.iloc[0]["revenue_share_pct"]), 2),
                     "non_uk_share_pct": round(float(100 * non_home["revenue"].sum() / total), 2),
                     "top_non_uk": non_home.head(5).reset_index().round(2).to_dict("records")})

@track_tool("analyze_customers")
def analyze_customers(df: pd.DataFrame) -> dict:
    """Customer behaviour: spend distribution, concentration, repeat rate, frequency segments (identified customers only)."""
    s = _sales(_ensure_features(df))
    ident = s[s["CustomerID"].notna()]
    if ident.empty:
        raise ValueError("No sales lines with CustomerID.")
    per = ident.groupby("CustomerID").agg(spend=("Revenue", "sum"), orders=("InvoiceNo", "nunique"), units=("Quantity", "sum"))
    per = per.sort_values("spend", ascending=False)
    k = max(1, int(math.ceil(0.10 * len(per))))
    orders = per["orders"]
    return jsonable({
        "identified_customers": int(len(per)),
        "revenue_share_from_identified_pct": round(100 * float(ident["Revenue"].sum()) / float(s["Revenue"].sum()), 2),
        "revenue_share_without_customer_id_pct": round(100 * float(s.loc[s["CustomerID"].isna(), "Revenue"].sum()) / float(s["Revenue"].sum()), 2),
        "mean_spend": round(float(per["spend"].mean()), 2), "median_spend": round(float(per["spend"].median()), 2),
        "spend_percentiles": {f"p{q}": round(float(per["spend"].quantile(q / 100)), 2) for q in (50, 75, 90, 99)},
        "top10pct_customers_count": k, "top10pct_revenue_share_pct": round(100 * float(per["spend"].iloc[:k].sum()) / float(per["spend"].sum()), 2),
        "repeat_customer_rate_pct": round(100 * float((orders >= 2).mean()), 2),
        "avg_orders_per_customer": round(float(orders.mean()), 2),
        "segments": {"one_time": int((orders == 1).sum()), "occasional_2_5": int(((orders >= 2) & (orders <= 5)).sum()), "loyal_6_plus": int((orders >= 6).sum())},
        "top_customers": per.head(10).reset_index().round(2).to_dict("records"),
    })

@track_tool("detect_anomalies")
def detect_anomalies(df: pd.DataFrame, method: str = "iqr", iqr_k: float = 3.0, z_threshold: float = 4.0) -> dict:
    """Detect **statistical anomalies** (NOT fraud) on product sales lines.

    Line level: Quantity, UnitPrice and line Revenue above an upper threshold
      * method='iqr':    Q3 + iqr_k x IQR  (default k=3 -> extreme outliers only)
      * method='zscore': mean + z_threshold x std
    Day level: robust z-score (median/MAD, Iglewicz-Hoaglin) on daily revenue, |z| > 3.5.
    """
    method = (method or "iqr").lower()
    if method not in {"iqr", "zscore"}:
        raise ValueError("method must be 'iqr' or 'zscore'")
    full = _ensure_features(df)
    p = _sales(full)
    p = p[~p["IsNonProduct"]].copy()
    if p.empty:
        raise ValueError("No product sales lines for anomaly detection.")
    thresholds, flags = {}, pd.DataFrame(index=p.index)
    for col in ("Quantity", "UnitPrice", "Revenue"):
        x = p[col].astype(float)
        thr = float(x.quantile(0.75) + iqr_k * (x.quantile(0.75) - x.quantile(0.25))) if method == "iqr" else float(x.mean() + z_threshold * x.std())
        thresholds[col] = round(thr, 2)
        flags[col] = x > thr
    any_flag = flags.any(axis=1)
    flagged = p[any_flag].copy()
    flagged["reasons"] = flags[any_flag].apply(lambda r: ", ".join(c for c in flags.columns if r[c]), axis=1)
    top = flagged.sort_values("Revenue", ascending=False).head(10)
    examples = [{"InvoiceNo": r.InvoiceNo, "date": str(r.InvoiceDate.date()), "description": str(r.Description)[:45], "quantity": int(r.Quantity),
                 "unit_price": round(float(r.UnitPrice), 2), "revenue": round(float(r.Revenue), 2), "country": r.Country, "reasons": r.reasons}
                for r in top.itertuples()]
    daily = p.groupby("Date")["Revenue"].sum()
    med = float(daily.median()); mad = float((daily - med).abs().median())
    rz = 0.6745 * (daily - med) / mad if mad > 0 else pd.Series(0.0, index=daily.index)
    bad_days = rz[rz.abs() > 3.5].sort_values(key=lambda x: -x.abs())
    returns = _returns(full).sort_values("Revenue").head(5)
    return jsonable({
        "label": "Statistical Anomalies (unusual values, not proof of fraud or error)", "method": method,
        "parameters": {"iqr_k": iqr_k} if method == "iqr" else {"z_threshold": z_threshold},
        "thresholds": thresholds, "lines_analysed": int(len(p)),
        "per_metric_counts": {c: int(flags[c].sum()) for c in flags.columns},
        "flagged_lines": int(any_flag.sum()), "flagged_lines_pct": round(100 * float(any_flag.mean()), 3),
        "flagged_revenue": round(float(flagged["Revenue"].sum()), 2),
        "flagged_revenue_share_pct": round(100 * float(flagged["Revenue"].sum()) / float(p["Revenue"].sum()), 2),
        "top_examples": examples,
        "daily": {"method": "robust z-score (median/MAD), |z|>3.5", "days_analysed": int(len(daily)), "flagged_days": int(len(bad_days)),
                  "examples": [{"date": str(d.date()), "revenue": round(float(daily[d]), 2), "robust_z": round(float(z), 2)} for d, z in bad_days.head(8).items()]},
        "largest_returns": [{"InvoiceNo": r.InvoiceNo, "description": str(r.Description)[:45], "quantity": int(r.Quantity), "value": round(float(r.Revenue), 2)} for r in returns.itertuples()],
    })

@track_tool("analyze_weather_impact")
def analyze_weather_impact(df: pd.DataFrame, weather_df: Optional[pd.DataFrame]) -> dict:
    """Exploratory correlation between daily sales revenue and daily weather (association only, not causation)."""
    if weather_df is None or weather_df.empty:
        return {"available": False, "reason": "No weather data retrieved."}
    daily = _sales(_ensure_features(df)).groupby("Date")["Revenue"].sum().rename("revenue").reset_index().rename(columns={"Date": "date"})
    w = weather_df.copy(); w["date"] = pd.to_datetime(w["date"]).dt.normalize()
    j = daily.merge(w, on="date", how="inner").dropna(subset=["revenue"])
    if len(j) < 20:
        return {"available": False, "reason": f"Only {len(j)} overlapping days (need >= 20)."}
    out: dict[str, Any] = {"available": True, "overlap_days": int(len(j)),
                           "caveat": "Weather location may differ from the customer base (provided coordinates = Berlin; most sales are UK). Correlation is not causation."}
    for col, key in (("temp_mean", "temperature"), ("precip_sum", "precipitation")):
        if col in j and j[col].notna().sum() >= 20 and j[col].nunique() > 1:
            out[f"pearson_{key}"] = round(float(j["revenue"].corr(j[col])), 3)
            out[f"spearman_{key}"] = round(float(j["revenue"].corr(j[col], method="spearman")), 3)
    if "temp_mean" in j and j["temp_mean"].nunique() >= 4:
        try:
            j["band"] = pd.qcut(j["temp_mean"], 4, duplicates="drop")
            g = j.groupby("band", observed=True).agg(avg_daily_revenue=("revenue", "mean"), days=("revenue", "size"), t_min=("temp_mean", "min"), t_max=("temp_mean", "max"))
            out["by_temperature_quartile"] = [{"temp_range_c": f"{r.t_min:.1f} to {r.t_max:.1f}", "avg_daily_revenue": round(float(r.avg_daily_revenue), 2), "days": int(r.days)} for r in g.itertuples()]
        except Exception:
            pass
    return jsonable(out)

# %% [markdown]
# ### 14.2 LLM-facing tool wrappers (what the Groq model can actually call)
# Each wrapper has a docstring that becomes the tool description sent to the model. Wrappers catch exceptions and return `{"error": ...}` so one failing tool never crashes the agent loop. Full results are cached in `RESULTS`; the LLM receives a size-limited JSON copy (to respect free-tier token limits).

# %%
from langchain_core.tools import StructuredTool

RESULTS: dict[str, Any] = {}          # tool name -> latest full result (used by agents/nodes)
RESULT_ERRORS: dict[str, str] = {}    # tool name -> latest error message
SEARCH_QUERIES: list[str] = []       # queries actually executed by the report-search tool
ANALYSIS_DF_ERROR = "Clean sales data is not available yet (run retrieval + quality/cleaning first)."

def _run_tool(name: str, fn: Callable, *args, **kwargs) -> Any:
    try:
        res = fn(*args, **kwargs)
        RESULTS[name] = jsonable(res)
        RESULT_ERRORS.pop(name, None)
        return RESULTS[name]
    except Exception as exc:
        RESULT_ERRORS[name] = f"{type(exc).__name__}: {exc}"
        return {"error": RESULT_ERRORS[name]}

def _clean_df() -> pd.DataFrame:
    if STORE.clean_df is None:
        raise RuntimeError(ANALYSIS_DF_ERROR)
    return STORE.clean_df

# ---- retrieval tools ----
def load_sales_data_tool() -> dict:
    """Load the Online Retail sales dataset (Excel) into memory. Call this first. Returns row count, columns and date range."""
    def _do():
        if not STORE.paths.get("sales"):
            raise FileNotFoundError(f"Sales file not found. Expected: {EXPECTED_NAMES['sales']}")
        STORE.raw_df = load_sales_data(STORE.paths["sales"])
        d = pd.to_datetime(STORE.raw_df["InvoiceDate"], errors="coerce")
        return {"rows": len(STORE.raw_df), "columns": list(STORE.raw_df.columns), "date_start": str(d.min().date()), "date_end": str(d.max().date()),
                "file": Path(STORE.paths["sales"]).name}
    return _run_tool("load_sales_data", _do)

def search_annual_report_tool(query: str, top_k: int = 3) -> dict:
    """Search the Target annual report (PDF) for passages relevant to a query, e.g. 'net sales growth' or 'supply chain risk'. Returns snippets with page numbers."""
    res = _run_tool("search_annual_report", search_annual_report, query, int(top_k or 3))
    if isinstance(res, dict) and "error" in res:
        return res
    SEARCH_QUERIES.append(query)
    RESULTS["search_annual_report"] = {"queries_run": list(SEARCH_QUERIES)}
    return {"query": query, "results": [{"page": r["page"], "score": r["score"], "snippet": r["text"][:280]} for r in res]}

def get_weather_data_tool(start_date: str = "", end_date: str = "", latitude: float = WEATHER_LAT, longitude: float = WEATHER_LON, mode: str = "historical") -> dict:
    """Get weather data from the live Open-Meteo API (falls back to a local CSV). mode='historical' needs start_date/end_date (YYYY-MM-DD; leave empty to use the sales date range); mode='current' returns current conditions."""
    return _run_tool("get_weather_data", get_weather_data, start_date or "", end_date or "", latitude, longitude, mode or "historical")

# ---- analysis tools ----
def calculate_kpis_tool() -> dict:
    """Calculate the deterministic business KPIs (revenue, orders, customers, AOV, items per order, return rates)."""
    return _run_tool("calculate_kpis", lambda: calculate_kpis(_clean_df()))
def analyze_sales_trends_tool() -> dict:
    """Analyse monthly revenue trends, growth, seasonality peaks, weekday and hourly profiles."""
    return _run_tool("analyze_sales_trends", lambda: analyze_sales_trends(_clean_df()))
def get_top_products_tool(n: int = 10) -> dict:
    """Get the top-n products by revenue and by units sold."""
    return _run_tool("get_top_products", lambda: get_top_products(_clean_df(), int(n or 10)))
def get_top_countries_tool(n: int = 10) -> dict:
    """Get the top-n countries by revenue with orders, customers and revenue share."""
    return _run_tool("get_top_countries", lambda: get_top_countries(_clean_df(), int(n or 10)))
def analyze_customers_tool() -> dict:
    """Analyse customer behaviour: spend distribution, concentration, repeat purchase rate and segments."""
    return _run_tool("analyze_customers", lambda: analyze_customers(_clean_df()))
def detect_anomalies_tool(method: str = "iqr") -> dict:
    """Detect statistical anomalies in transactions (unusually high quantity/price/revenue) using 'iqr' or 'zscore'."""
    return _run_tool("detect_anomalies", lambda: detect_anomalies(_clean_df(), method or "iqr"))
def analyze_weather_impact_tool() -> dict:
    """Correlate daily revenue with the retrieved weather data (exploratory, association only)."""
    return _run_tool("analyze_weather_impact", lambda: analyze_weather_impact(_clean_df(), STORE.weather_df))

def _mk(fn: Callable, name: str) -> StructuredTool:
    return StructuredTool.from_function(func=fn, name=name, description=(fn.__doc__ or name).strip())

RETRIEVAL_TOOLS = [_mk(load_sales_data_tool, "load_sales_data"), _mk(search_annual_report_tool, "search_annual_report"), _mk(get_weather_data_tool, "get_weather_data")]
ANALYSIS_TOOLS = [_mk(calculate_kpis_tool, "calculate_kpis"), _mk(analyze_sales_trends_tool, "analyze_sales_trends"), _mk(get_top_products_tool, "get_top_products"),
                  _mk(get_top_countries_tool, "get_top_countries"), _mk(analyze_customers_tool, "analyze_customers"), _mk(detect_anomalies_tool, "detect_anomalies"),
                  _mk(analyze_weather_impact_tool, "analyze_weather_impact")]
INSIGHT_TOOLS = [RETRIEVAL_TOOLS[1]]
print("LLM-callable tools:", [t.name for t in RETRIEVAL_TOOLS + ANALYSIS_TOOLS])
````
